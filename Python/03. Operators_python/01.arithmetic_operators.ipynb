### **Operators in Python**

Operators are special symbols used to perform operations on variables and values.

#### **Types of Operators in Python**

1. Arithmetic Operators

2. Comparison (Relational) Operators

3. Logical Operators

4. Assignment Operators

5. Identity Operators

6. Membership Operators

7. Bitwise Operators

**1. Arithmetic Operators**

Arithmetic operators are used to perform mathematical calculations such as addition, subtraction, multiplication, and division.

**1.1 Addition**

```python
# Addition
a = 2
b = 3
print('Addition: ', a + b)
```

**1.2 Subtraction**

```python
# Subtraction
c = 5
d = 2
print('Subtraction: ', c - d)  # Output: 3
```

**1.3 Multiplication(*)**

```python
# Multiplication
e = 2
f = 3
print('Multiplication: ', e * f)
```

**1.4 Division**


```python
g = 6
h = 3
print('Division: ', g / h) 
```

**1.5 Modulus**
The modulus operator (%) returns the remainder when one operand is divided by another. It can be useful for tasks such as checking for even or odd numbers, or performing divisibility tests.

```python
# Checking if a number is even or odd:

num = 7
if num % 2 == 0:
    print('Even')
else:
    print('Odd')  

    
# Divisibility test:

dividend = 10
divisor = 2
if dividend % divisor == 0:
    print(f'{dividend} is divisible by {divisor}') 
else:
    print(f'{dividend} is not divisible by {divisor}')
```

**1.6 Exponentiation**


```python
k = 2
l = 3
print('Exponentiation: ', k ** l)
```

**1.7 Floor Division Operator**

The floor division operator (//) returns the largest possible integer that is less than or equal to the result of the division (i.e., it performs integer division).

```python
print('7 // 2 = ', 7 // 2) 
print('11 // 4 = ', 11 // 4)
```
