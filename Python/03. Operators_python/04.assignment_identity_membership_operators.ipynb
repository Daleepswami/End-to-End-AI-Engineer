### **Assignment_Operator::==>>**
Python assignment operators are used to assign values to variables. The most commonly used assignment operator is = .

```python
a = 10
```

```python
# Assignment Operators

a = 10      # assign the value

a += 5      # Addition and assign
print(a)

a -= 3      # Subtraction and assign
print(a)

a *= 2      # multipaly and assign
print(a)

a /= 4      # divisin and assign
print(a)

a //= 2     # floor division and assign
print(a)

a **= 2     # Exponentiation and assign
print(a)

a %= 3       # module and assign
print(a)
```

### Identity Operators
Identity operators in Python are is and is not. They are used to check if two variables refer to the same object in memory.

**is Operator ::==>>**

The is operator returns True if the operands refer to the same object, otherwise it returns False.

**is not Operator::==>>**

The is not operator returns True if the operands do not refer to the same object, otherwise it returns False.

```python
# Identity Operators

a = 20
b = 20
c = a

print(a is b)
print(a is c)
print(a is not b)

x = [1, 2, 3]
y = [1, 2, 3]
z = x

print(x is y)
print(x is z)
print(x is not y)

print(x == y)
print(x != y)
```

### Membership Operators

Membership operators are used to test if a value or variable is found in a sequence (string, list, tuple, set, etc.). There are two membership operators in Python:


1. in
2. not in

**1. in Operator**
The in operator is used to check if a value exists in a sequence. It returns True if the value is found, otherwise it returns False

```python
# in operator
a = 'Mayank'
print('aya' in a)
print('May' in a)
print('may' in a)
print('ank' in a)

numbers = [2, 3, 5, 7]
print(2 in numbers)
print(11 in numbers)
print(5 in numbers)
print(4 in numbers)
```

**2.not in Operator**

The not in operator is used to check if a value does not exist in a sequence. It returns True if the value is not found, otherwise it returns False.

```python
a = 'Mayank'
print('may' not in a)
print('May' not in a)
print('xyz' not in a)
print('ank' not in a)

numbers = [1, 2, 3]
print(11 not in numbers)
print(2 not in numbers)
print(4 not in numbers)
print(1 not in numbers)
```
