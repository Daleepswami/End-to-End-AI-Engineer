### Logical Operators in Python
Logical operators are used to combine conditions and check whether a condition is True or False.

Python has three logical operators:

1. and

2. or

3. not




**1. and Operator**

The and operator returns True only when both conditions are True.

```python
a = 6
b = 20
c = 50

print(a < b and b < c)
```

```python
print(a < b and b > a) 
print(a > b and b > a)  
print(a < b and b < a)  
print(a > b and b < a)
```

**2. or Operator**

The or operator returns True when at least one condition is True.

```python
a = 6
b = 20
c = 50

print(a > b or b < c)
```

```python
print(a < b or b > a)   
print(a > b or b < a)   
print(a < b or b < a)   
print(a > b or b > a)
```


**3. not Operator**

The not operator reverses the Boolean result: True becomes False, and False becomes True.

```python
a = 6
b = 20

print(not (a > b))
```

```python
print(not (a < b))      
print(not (a > b))      
print(not (a == b))     
print(not (a != b))
```
