### ==>> Comments in Python
Comments are used to explain code and make it easier to understand. Python ignores comments during execution.

**Single-Line Comments**

A single-line comment starts with the # symbol.

```python
# Display a message
print("Hello Python")
```

**Uses of Comments**

To explain the purpose of code.

To improve code readability.

To temporarily disable a line of code during testing.

```python
# print("Hello Python")
print("Hello World")  #The first line is ignored because it is commented out.
```

**Multi-line Comments in Python**

Python does not have a dedicated syntax for multi-line comments. You can use # at the beginning of each line.

Triple quotes (''' or """) can span multiple lines, but they create a string, not a true comment. If the string is not assigned or used, it may simply be ignored.

```python
# This is a multi-line comment.
# It explains the purpose of the code.
# Python ignores these lines.

print("Hello Python")
```

```python
"""
This is a multi-line string.
It is not a true Python comment.
"""

print("Hello Python")
```

### ==>> Syntax Errors vs Runtime Errors

Errors are problems in a program that prevent it from working as expected. Two common types are syntax errors and runtime errors.







**==>> 1. Syntax Errors**

A syntax error occurs when code violates Python's syntax rules. Python detects this error before executing the affected code.

**Common causes:**

Misspelled keywords or function names in invalid syntax.

Missing parentheses or quotation marks.

Incorrect indentation in places where indentation is required.

```python
prin("Hello Python")
```

Note: This example actually produces a NameError, not a SyntaxError, because prin is a valid identifier but is not defined. A true syntax error is shown below.

print("Hello Python")

This produces a SyntaxError because the closing parenthesis is missing.

##### ==>> **2.Runtime Errors (Exceptions)**

Runtime errors occur while a program is executing. Python raises an exception when it encounters an operation it cannot complete.



```python
#Example: Division by Zero

print(10 / 0)
```

The error occurs because division by zero is not allowed.
