## Predefined Functions in Python

Python provides predefined functions that can be used without defining them ourselves. They are mainly categorized into:

**Built-in Functions** – Available directly in Python.

**Functions from Modules** – Available by importing a module.

**1. Built-in Functions**

Built-in functions are ready-to-use functions provided by Python.




**1.1 print() Function**

The print() function is used to display text, values, and results on the screen.

```python
print("Hello Python")
print(10 + 20)
```

**1.2 The end Argument**

By default, print() moves to a new line after displaying the output. The end argument changes what is printed at the end.

```python
print("Hello", end=" ")
print("Python")
```

**1.3 The sep Argument**

The sep argument specifies the separator between multiple values passed to print(). By default, the separator is a space.


```python
print("Hello", "Python", sep="-")
```
