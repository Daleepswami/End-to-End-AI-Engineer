### **Identifiers in Python**
Identifiers are names given to program elements such as variables, functions, classes, and modules. They are used to identify these elements in a Python program.

#### 1. Rules for Naming Identifiers

1. Reserved Keywords: An identifier cannot be a Python keyword, such as if, for, class, or True.

2. Starting Character: An identifier cannot begin with a number.

3. Allowed Characters: Identifiers can contain letters (A–Z, a–z), digits (0–9), and underscores (_).

4. No Other Special Characters: Spaces and special characters such as @, !, and - are not allowed.

5. Case-Sensitive: Uppercase and lowercase letters are treated differently.

**2. Valid and Invalid Identifiers**

**Valid identifiers:**

    variable1
    
    _variable
    
    variable_name

    studentMarks

**Invalid identifiers:**

    1variable — starts with a number.

    variable-name — contains a hyphen.

    variable! — contains a special character.

    class — a reserved keyword.

```python
# Valid identifiers
variable1 = 10
_variable = 20
variable_name = 30

print(variable1)
print(_variable)
print(variable_name)
```

```python
# Case-Sensitivity Example

name = "Python"
Name = "Programming"

print(name)
print(Name)
```

```python
# Checking Python Keywords
    # Python provides the keyword module to check reserved keywords.

import keyword

print(keyword.iskeyword("if"))
print(keyword.iskeyword("variable"))
```

**Key Points**

Identifiers name variables, functions, classes, and other program elements.

Identifiers cannot start with a digit.

Only letters, digits, and underscores are allowed.

Identifiers are case-sensitive.

Python keywords cannot be used as identifiers.
