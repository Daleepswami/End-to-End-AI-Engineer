### **Data Types in Python**
A data type defines the type of value a variable stores or refers to. Python determines the data type automatically based on the assigned value, so we do not need to declare it explicitly.

**1. Integer (int)**

Integers represent whole numbers, including positive numbers, negative numbers, and zero, without a fractional part.

```python
a = 10
b = -5
c = 0

print(a)
print(type(a))
print(type(b))
print(type(c))
```

**2. Float (float)**

Floats represent floating-point numbers, which commonly include a decimal point. They can also be written using exponential notation

```python
a = 3.14
b = -2.7
c = 1.05e3

print(a)
print(b)
print(c)
print(type(a))
```

**3. Complex (complex)**

Complex numbers contain a real part and an imaginary part. Python uses j to represent the imaginary unit.

```python
a = 3 + 4j
b = 2.5 - 1.5j

print(a)
print(type(a))
print(a.real)
print(a.imag)
```

**4. Boolean (bool)**

The Boolean data type represents one of two values: True or False. It is commonly used in conditions and comparisons.

```python
a = True
b = False

print(a)
print(b)
print(type(a))
print(type(b))
```

**5. String (str)**

A string is a sequence of characters enclosed in single quotes, double quotes, or triple quotes.

```python
name = "Python"
message = 'Hello World'

print(name)
print(message)
print(type(name))
```

**Strings in Python:** A string is a sequence of characters enclosed in single quotes (' '), double quotes (" "), or triple quotes (""" """).

Strings are used to store text, such as names, messages, and sentences.

**1. Creating Strings**

```python
single = 'Hello'
double = "World"
multi = """This is
multi-line"""

print(single)
print(double)
print(multi)
```

**Access a character using indexing**

```python
s="Python"
print(s[0])
print(s[-1])
```

**Strings are immutable**

```python
s = "python"
# This causes a TypeError:
# s[0] = "P"


# 'str' object does not support item assignment
```

#### **Useful String Operations**

**1. upper:->** Convert to uppercase

```python
s = "python"
print(s.upper())
```

**2. capitalize():=>** Capitalize the first letter

```python
print(s.capitalize())
```

**3. count():=>** count the characters

```python
print(len(s))
```

**4. lower():->** Convert to uppercase

```python
s="PYTHON"
print(s.lower())
```
