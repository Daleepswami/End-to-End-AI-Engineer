### **Data Types in Python**
A data type defines the type of value a variable stores or refers to. Python determines the data type automatically based on the assigned value, so we do not need to declare it explicitly.

**1. Integer (int)**

Integers represent whole numbers, including positive numbers, negative numbers, and zero, without a fractional part.

```python
a = 10
b = -5
c = 0

print(a)
print(type(a))
print(type(b))
print(type(c))
```

**2. Float (float)**

Floats represent floating-point numbers, which commonly include a decimal point. They can also be written using exponential notation

```python
a = 3.14
b = -2.7
c = 1.05e3

print(a)
print(b)
print(c)
print(type(a))
```

**3. Complex (complex)**

Complex numbers contain a real part and an imaginary part. Python uses j to represent the imaginary unit.

```python
a = 3 + 4j
b = 2.5 - 1.5j

print(a)
print(type(a))
print(a.real)
print(a.imag)
```

**4. Boolean (bool)**

The Boolean data type represents one of two values: True or False. It is commonly used in conditions and comparisons.

```python
a = True
b = False

print(a)
print(b)
print(type(a))
print(type(b))
```

**5. String (str)**

A string is a sequence of characters enclosed in single quotes, double quotes, or triple quotes.

```python
name = "Python"
message = 'Hello World'

print(name)
print(message)
print(type(name))
```

A string is a sequence of characters enclosed in single quotes (' '), 
double quotes (" "), or triple quotes (""" """).


**6. Dictionary (dict)**

Stores data as key-value pairs. Written in curly brackets as key: value. Keys must be unique and immutable (str, int, tuple). Values can be anything.

```python
student = {
    "name": "Aman",
    "age": 20,
    "city": "Jaipur"
}
```

```python
print(student["name"])             
print(student.get("age"))          
print(student.get("phone"))        
print(student.get("phone", "NA"))  
```

**7. Set**

Unordered, no duplicates, no indexing. Written in curly brackets { }.

```python
nums = {1, 2, 3, 3, 2}
print(nums)   
```

```python
s = set()    # correct

d = {}       # this is an empty dictionary, not a set
```

**8. Tuple**

Ordered, immutable, duplicates allowed. Written in round brackets ( ).

```python
point = (10, 20, 30)
print(point[0])     
print(point[-1])    
print(point[0:2])   
```

**9. List**

Ordered, mutable, duplicates allowed. Written in square brackets [ ].

```python
fruits = ["apple", "banana", "mango"]
print(fruits)
```

**10. NoneType**

None means “nothing” or no value. It is a special Python object, and its type is NoneType.

```python
x = None
print(x)         
print(type(x))   
```
