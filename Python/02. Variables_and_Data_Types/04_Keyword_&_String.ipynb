### **Keywords in Python**
#### **What are keywords?**

Keywords are reserved words that have a special meaning to the Python interpreter.

They cannot be used as identifiers (variable, function, or class names).

You can list them with the built-in keyword module.

#### **Listing all keywords**

```python
import keyword

words = keyword.kwlist
print(len(words), "keywords")

for i in range(0, len(words), 6):
    print("  ".join(f"{w:<9}" for w in words[i:i + 6]))
```

### **Python Strings**



What is a String?

- A string (`str`) is an **ordered sequence** of Unicode characters.
- Strings are **immutable**: once created, their characters cannot be changed in place. Every modification produces a **new string**.

```python
name = "Python"
print(type(name), len(name))
print("P" in name, "java" in name.lower())
```

#### **Creating a String**

| Way | Example |
|---|---|
| Single quotes | `'single quotes'` |
| Double quotes | `"double quotes"` |
| Triple quotes (multi-line) | `"""line1 line2"""` |
| Mix quotes to include an apostrophe | `"It's easy"` |
| Escape sequences | `'Line 1\nLine 2\tTabbed'` |
| Raw string (backslashes kept as-is) | `r"C:\new\folder"` |

**Escape sequences:**

| Escape | Meaning |
|---|---|
| `\n` | New line |
| `\t` | Tab |
| `\\` | Backslash |
| `\'` and `\"` | Quote characters |

```python
s1 = 'single quotes'
s2 = "double quotes"
s3 = """triple quotes
can span several lines"""
s4 = "It's easy"
s5 = 'Line 1\nLine 2\tTabbed'
s6 = r"C:\new\folder"

for s in (s1, s2, s3, s4, s5, s6):
    print(s)
```

#### **Indexing and Slicing**

- **Positive index**: starts from 0 (from the left).
- **Negative index**: starts from -1 (from the end).
- **Slicing**: `s[start:stop:step]`, where `stop` is **excluded**.

The string `"PYTHON"`:

| P | Y | T | H | O | N |
|---|---|---|---|---|---|
| 0 | 1 | 2 | 3 | 4 | 5 |
| -6 | -5 | -4 | -3 | -2 | -1 |

```python
s = "PYTHON"
print(s[0], s[-1])
print(s[0:3])
print(s[2:])
print(s[:4])
print(s[::2])
print(s[::-1])
```

#### **split() and join()**

- `split()` breaks a string into a **list**.
- `join()` combines a list back into a **string**.
- `split()` with no argument splits on **any whitespace** and removes extra spaces.

```python
csv_line = "Asha,28,Chennai,Data Analyst"
parts = csv_line.split(",")
print(parts)
print(" | ".join(parts))

sentence = "  Python   is   fun  "
print(sentence.split())
```

#### **Common String Methods**

```python
text = "  Data Analytics with Python  "
print(text.strip())
print(text.upper())
print(text.lower())
print(text.strip().title())
print(text.replace("Python", "SQL"))
print(text.find("with"), text.count("a"))
print(text.strip().startswith("Data"), "2026".isdigit())
```

| Method | What it does |
|---|---|
| `strip()` / `lstrip()` / `rstrip()` | Remove surrounding whitespace |
| `upper()` / `lower()` / `title()` / `capitalize()` | Change case |
| `replace(old, new)` | Replace text |
| `find(sub)` / `index(sub)` | Position of `sub` (`find` gives -1, `index` raises an error if not found) |
| `count(sub)` | Number of occurrences |
| `startswith()` / `endswith()` | Check the beginning or end |
| `isdigit()` / `isalpha()` / `isalnum()` | Check the kind of characters |
| `split()` / `join()` | Break into a list / combine a list |

#### **Deleting a String**

- Because strings are immutable, you **cannot delete or change a single character** (`word[2] = "t"` raises `TypeError`).
- You can build a **new string**, or delete the whole variable with `del`.

```python
word = "Pyhton"
# word[2] = "t"            # would raise TypeError
fixed = word[:2] + "th" + word[4:]
print(fixed)
del fixed
print(fixed)
```

#### **String Formatting**

Formatting inserts values into text. **f-strings** (Python 3.6+) are the clearest and fastest option.

```python
name, marks, total = "Asha", 456, 500
pct = marks / total * 100

print(f"{name} scored {marks}/{total} ({pct:.1f}%)")

print("{} scored {}/{}".format(name, marks, total))

print("%s scored %d/%d (%.1f%%)" % (name, marks, total, pct))
```

**Alignment example:**

```python
print(f"{'Item':<10}{'Price':>8}")
print(f"{'Laptop':<10}{54999.5:>8,.1f}")
print(f"{0.256:.1%}")    # 25.6%
```

**Format specifiers:**

| Spec | Meaning | Example -> Result |
|---|---|---|
| `:.2f` | 2 decimal places | `f"{3.14159:.2f}"` -> `3.14` |
| `:,` | Thousands separator | `f"{1500000:,}"` -> `1,500,000` |
| `:<10` / `:>10` / `:^10` | Left / right / center align | `f"{'hi':>5}"` -> `'   hi'` |
| `:.1%` | Percentage | `f"{0.256:.1%}"` -> `25.6%` |
