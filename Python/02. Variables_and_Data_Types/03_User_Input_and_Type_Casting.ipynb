### User Input and Type Casting in Python

**1. User Input (input())**

The input() function is used to take input from the user. By default, it returns the input as a string (str).

```python
name = input("Enter your name: ")
print("Hello", name)
```

**2. Type Casting**

Type casting is the process of converting a value from one data type to another.

Python supports two types of type conversion:

*1. Implicit Type Casting*

*2. Explicit Type Casting*

**2.1 Implicit Type Casting**

Python automatically converts one data type into another when appropriate.

```python
a = 10
b = 2.5

result = a + b

print(result)
print(type(result))
```

**2.2 Explicit Type Casting**

Explicit type casting means converting a value into a specific data type using functions such as int(), float(), str(), and bool().

```python
# Using int()
a = int("25")
print(a)
print(type(a))
```

```python
# Using float()

a = float("3.14")

print(a)
print(type(a))
```

```python
# Using str()

a = str(100)

print(a)
print(type(a))
```

```python
# Using bool()

print(bool(1))
print(bool(0))
print(bool(""))
```

**Taking Numeric Input**

Because input() returns a string, convert the input before performing numeric calculations.

```python
age = int(input("Enter your age: "))

print("Your age is:", age)
print(type(age))
```
