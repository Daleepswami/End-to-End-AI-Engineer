# **02. Basic Exception Handling**

### **2.1 Introduction to `try` and `except`**

**1. What are `try` and `except`?**

- ***To prevent your program from crashing when an exception occurs, Python provides the `try`-`except` block.***

- ***This mechanism allows you to "try" to execute a block of code, and if an exception occurs, to "catch" it and execute a different block of code (the exception handler) instead of crashing.***

**2. Syntax:**

```python
try:
    # Code that might raise an exception
except SomeExceptionType:
    # Code to execute if SomeExceptionType occurs in the 'try' block
    # This is the exception handler
```

**3. Example: Handling `ZeroDivisionError`**

````python
try:
    result = 10 / 0                       # This line will cause a ZeroDivisionError
    print(f"Result: {result}")            # This line will NOT be executed
except ZeroDivisionError:
    print("Error: Cannot divide by zero!")

print("Program continues here.")          # Program did not crash
````

### **2.2 Working of `try` Block**

**1. Working of the `try` Block**

- ***The `try` block contains the code that you want to monitor for exceptions.***

- ***Python runs the lines of the `try` block one by one.***

- ***If no exception occurs, the whole block runs and the `except` block is skipped.***

- ***If an exception occurs, the remaining lines of the `try` block are skipped.***

````python
try:
    print("Line 1 runs")
    value = int("abc")                    # exception occurs here
    print("Line 3 never runs")
except ValueError:
    print("Jumped to except")
````

### **2.3 Working of `except` Block**

**1. Working of the `except` Block**

- ***If an exception of `SomeExceptionType` (for example, `ValueError` or `ZeroDivisionError`) occurs within the `try` block, the execution jumps immediately to this `except` block.***

- ***The code inside the `except` block is executed, and then the program continues normally after the entire `try`-`except` structure.***

**2. Example: Handling Invalid User Input**

- ***This program asks for a number, but robustly handles cases where the user types non-numeric input.***

````python
try:
    num_str = input("Enter an integer: ")
    number = int(num_str)        # This might raise a ValueError if input is not an int
    print(f"You entered: {number}")
except ValueError:               # Catch only ValueError
    print("Invalid input! Please enter a whole number.")

print("End of program.")
````

**3. Interaction**

```
# Scenario 1: Valid input
Enter an integer: 42
You entered: 42
End of program.

# Scenario 2: Invalid input
Enter an integer: hello
Invalid input! Please enter a whole number.
End of program.
```

### **2.4 Handling Specific Exceptions**

**1. Why Handle Specific Exceptions?**

- ***It is good practice to handle specific exceptions rather than catching all possible errors.***

- ***This allows you to provide precise error messages and appropriate recovery actions.***

- ***Use `except ValueError`, not a bare `except`.***

**2. Catching the Exception Object (`as e`)**

- ***You can capture the exception object itself using `as e` (or any other variable name).***

- ***This object often contains useful information about the error.***

````python
try:
    data = [1, 2, 3]
    index = int(input("Enter an index: "))
    value = data[index]                       # Could raise IndexError
    print(f"Value at index {index}: {value}")
except IndexError as e:                       # 'e' will hold the IndexError object
    print(f"Error accessing list: {e}")
    print("The index you entered is out of range.")
except ValueError as e:                       # 'e' will hold the ValueError object
    print(f"Error converting input: {e}")
    print("Please enter a valid integer for the index.")
````

### **2.5 Handling Multiple Exceptions**

**1. Multiple `except` Blocks**

- ***You can use multiple `except` blocks to handle different types of exceptions in different ways.***

- ***Python will check `except` blocks in order from top to bottom.***

````python
try:
    num1 = int(input("Enter first number: "))
    num2 = int(input("Enter second number: "))
    result = num1 / num2
    print(f"Result: {result}")
except ValueError:                      # Handles non-integer input
    print("Error: Please enter valid numbers.")
except ZeroDivisionError:               # Handles division by zero
    print("Error: You cannot divide by zero!")
except Exception:                       # A general block (catches any other unhandled exception)
    print("An unexpected error occurred.")
````

**2. Important Note on Order**

- ***Place the most specific exception handlers first, followed by more general ones.***

- ***If `except Exception:` (which catches all exceptions) were first, it would catch everything, and the subsequent specific `except` blocks would never be reached.***

### **2.6 Handling Multiple Exceptions in a Single `except` Block**

**1. Using a Tuple**

- ***Several exception types can be caught in one `except` block by writing them as a tuple.***

- ***The same handler runs for any of the listed exceptions.***

````python
for value in ["10", "abc", None]:
    try:
        print(100 / int(value))
    except (ValueError, TypeError) as e:          # several types in one block
        print("Bad value:", e)
````

**2. Combined Example**

````python
def to_ratio(values, index):
    try:
        return 100 / int(values[index])
    except (ValueError, TypeError) as e:                      # several types in one block
        return f"Bad value: {e}"
    except ZeroDivisionError:
        return "Value is zero"
    except IndexError:
        return "No such position"
    except Exception as e:                                    # last resort
        return f"Unexpected error: {e}"

data = ["4", "zero", "0"]
for i in range(4):
    print(i, "->", to_ratio(data, i))
````

**Warning:**

- ***Order matters: list specific exceptions first and the general `Exception` last.***

### **2.7 Generic Exception Handling (`Exception`)**

**1. What is Generic Exception Handling?**

- ***`except Exception as e` is a catch-all handler: it catches almost every exception.***

- ***It is useful as a last resort, placed after all the specific handlers.***

````python
try:
    num = int(input('Enter a number: '))
    result = 10 / num
except ZeroDivisionError:
    print('Cannot divide by zero!')
except ValueError:
    print('Please enter a valid integer.')
except Exception as e:                  # Catch-all
    print(f'Unexpected error: {e}')
````

**2. Catching Different Errors with One Handler**

````python
actions = [lambda: 10 / 0, lambda: int("x"), lambda: [][1]]

for action in actions:
    try:
        action()
    except Exception as e:
        print(type(e).__name__, "->", e)
````

**3. Warning About a Bare `except:`**

- ***Avoid a bare `except:` because it also hides bugs and keyboard interrupts.***

- ***Always catch specific exceptions, not a bare `except:`.***
