# **01. Introduction to Exception Handling**

### **1.1 What is Exception Handling?**

**1. What is Exception Handling?**

- ***Exception handling is a mechanism that allows a program to respond to problems that occur while it is running, instead of crashing.***

- ***Errors are inevitable: users type letters instead of numbers, files go missing, and networks fail.***

- ***Python provides the `try`-`except` block for this purpose.***

- ***It allows you to "try" to execute a block of code, and if an exception occurs, to "catch" it and execute a different block of code (the exception handler) instead of crashing.***

**2. Basic Idea**

````python
try:
    result = 10 / 0                      # This line causes a ZeroDivisionError
except ZeroDivisionError:
    print("Error: Cannot divide by zero!")

print("Program continues here.")         # Program did not crash
````

**Key Takeaway:**

- ***Exception handling is about making your programs resilient. Instead of crashing, they can provide meaningful feedback to the user or recover from unexpected situations.***

### **1.2 What is an Exception?**

**1. What is an Exception?**

- ***An exception is a runtime error that interrupts normal program flow.***

- ***Exceptions occur during the execution of your program, even if the syntax is perfectly valid.***

- ***Python raises exceptions for invalid operations.***

- ***If not handled, an exception will cause your program to crash and terminate abruptly.***

- ***Python provides various built-in exception types to describe different kinds of problems.***

**2. Common Exceptions**

| Exception | When it Occurs |
|-----------|----------------|
| `ZeroDivisionError` | Division by zero |
| `ValueError` | Right type, but invalid value for an operation |
| `TypeError` | Operation on a wrong data type |
| `IndexError` | List index out of range |
| `KeyError` | Dictionary key not found |
| `NameError` | Variable not defined |
| `FileNotFoundError` | File does not exist |
| `AttributeError` | Object has no such attribute |
| `ImportError` | Module not found |
| `RecursionError` | Maximum recursion depth exceeded |

**3. Example**

````python
tests = [
    ("ZeroDivisionError", lambda: 10 / 0),
    ("ValueError",        lambda: int("abc")),
    ("TypeError",         lambda: "hello" + 5),
    ("IndexError",        lambda: [1, 2, 3][3]),
    ("KeyError",          lambda: {"name": "Alice"}["age"]),
    ("NameError",         lambda: undefined_variable),
    ("FileNotFoundError", lambda: open("non_existent_file.txt", "r")),
]

for name, action in tests:
    try:
        action()
    except Exception as e:
        print(f"{name:18} -> {e}")
````

### **1.3 Why Exception Handling is Needed**

**1. Reasons for Using Exception Handling**

1. **Prevents Crashes:** The program does not terminate abruptly when an error occurs.
2. **Meaningful Feedback:** The user gets a clear message instead of a technical error.
3. **Recovery:** The program can recover from unexpected situations and continue.
4. **Unpredictable Conditions:** Real-world conditions (wrong input, missing files, network failures) are unpredictable.
5. **Cleanup:** Resources such as files can be released properly even when an error occurs.

**2. Example: Handling Invalid Input**

````python
for text in ["42", "hello"]:
    try:
        number = int(text)               # This might raise a ValueError if input is not an int
        print(f"You entered: {number}")
    except ValueError:
        print("Invalid input! Please enter a whole number.")
````

### **1.4 Difference Between Syntax Errors and Exceptions**

**1. Syntax Errors**

- ***Syntax errors are mistakes in the structure of your code that violate Python's grammar rules.***

- ***The Python interpreter catches these before your code even starts running (at "compile time" or "parse time").***

- ***Your program will not run if it has a syntax error.***

- ***Examples: missing colons, unclosed parentheses, incorrect keywords.***

```
>>> if x > 5 # Missing colon
SyntaxError: expected ':'

>>> print("Hello" # Missing closing parenthesis
SyntaxError: unexpected EOF while parsing
```

**2. Exceptions (Run-time Errors)**

- ***Exceptions occur during the execution of the program, even if the syntax is perfectly valid.***

- ***They can be handled using `try`-`except`.***

| Feature | Syntax Error | Exception |
|---------|--------------|-----------|
| When it occurs | Before the program runs (parse time) | While the program is running (run time) |
| Does the program start? | No | Yes, until the error occurs |
| Cause | Code violates Python's grammar | Something unexpected happens during execution |
| Can be handled with `try`-`except`? | No, the code must be fixed | Yes |
| Example | Missing colon | `10 / 0` |

**3. Logic Errors (a Third Kind of Error)**

- ***Logic errors are the most insidious type of error because they do not cause your program to crash or raise an exception.***

- ***The program runs successfully from Python's perspective, but it produces incorrect or unintended results.***

- ***Debugging logic errors often requires careful testing, tracing variable values, and reviewing your algorithm.***

````python
num1 = 2
num2 = 2
# Intention: Add num1 and num2
result = num1 * num2      # Logic error: Used multiplication instead of addition
print(result)
````

### **1.5 How Exceptions Affect Program Execution**

**1. Unhandled Exception**

- ***If an exception is not handled, the program crashes and terminates abruptly.***

- ***The remaining code after the error is never executed.***

```python
print("Before error")
print(10 / 0)            # ZeroDivisionError: the program stops here
print("After error")     # never executed
```

```
Before error
ZeroDivisionError: division by zero
```

**2. Handled Exception**

- ***When an exception occurs inside the `try` block, the execution jumps immediately to the `except` block.***

- ***The remaining lines of the `try` block are skipped.***

- ***After the `except` block, the program continues normally after the entire `try`-`except` structure.***

````python
print("Step 1: start")
try:
    print("Step 2: inside try, before the error")
    result = 10 / 0
    print("Step 3: this line is skipped")
except ZeroDivisionError:
    print("Step 4: inside except")
print("Step 5: program continues")
````
