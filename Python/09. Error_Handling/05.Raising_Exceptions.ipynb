# **05. Raising Exceptions**

### **5.1 Introduction to `raise` Keyword**

**1. What is the `raise` Keyword?**

- ***Sometimes you might need to force an exception to occur if a certain condition is met (or not met).***

- ***The `raise` statement allows you to trigger an exception manually.***

**2. Uses of `raise`**

1. **Input Validation:** If a function receives invalid arguments.
2. **Custom Errors:** Creating and raising your own specific error types.
3. **Signaling Problems:** Indicating that a situation occurred that your function cannot handle.

**3. Syntax:**

```python
raise ExceptionType("Optional error message")
```

### **5.2 Raising Built-in Exceptions**

**1. Raising Built-in Exceptions**

- ***Python's built-in exception types (such as `ValueError` and `TypeError`) can be raised using `raise`.***

- ***Choose the exception type that best describes the problem.***

````python
def set_age(age):
    if not isinstance(age, int):                      # Check if it's an integer
        raise TypeError("Age must be an integer.")
    if age < 0 or age > 120:
        raise ValueError("Age must be between 0 and 120.")
    print(f"Age set to: {age}")

for value in [30, 200, "abc"]:
    try:
        set_age(value)
    except (TypeError, ValueError) as e:
        print(type(e).__name__, ":", e)
````

### **5.3 Raising Exceptions with Custom Messages**

**1. Custom Messages**

- ***The text inside the brackets is the error message.***

- ***A clear message tells the caller what went wrong and why.***

````python
def check_age(age):
    if age < 0:
        raise ValueError(f'Age cannot be negative: {age}')
    if age > 150:
        raise ValueError(f'Age seems unrealistic: {age}')
    return f'Age {age} is valid'

try:
    print(check_age(-5))
except ValueError as e:
    print(e)                                          # Age cannot be negative: -5
````

**2. Validate and Retry**

- ***Raising an exception inside `try` sends control to the matching `except` block.***

````python
def read_age():
    for attempt in range(3):
        raw = input("Age: ")
        try:
            age = int(raw)
            if age < 0:
                raise ValueError("age cannot be negative")
            return age
        except ValueError as e:
            print("Invalid:", e)
    return None

print("Saved age:", read_age())
````

### **5.4 Using `raise` Without Arguments**

**1. What Does a Bare `raise` Do?**

- ***A `raise` statement without any exception re-raises the exception that is currently being handled.***

- ***It is used inside an `except` block.***

- ***Outside an `except` block, there is no active exception, so Python raises a `RuntimeError`.***

````python
try:
    try:
        int("abc")
    except ValueError:
        print("Inner: error noticed, passing it on")
        raise                                          # re-raises the same ValueError
except ValueError as e:
    print("Outer received:", e)
````

````python
try:
    raise                                              # no active exception here
except RuntimeError as e:
    print("RuntimeError:", e)
````

### **5.5 Re-raising Exceptions**

**1. Why Re-raise?**

- ***Sometimes a function wants to do something (such as logging or cleanup) and then let the caller handle the error.***

- ***It handles the exception partly, and then raises it again with `raise`.***

- ***Using `raise` keeps the original error and its traceback.***

````python
def process(value):
    try:
        return 100 / value
    except ZeroDivisionError:
        print("Log: division by zero happened in process()")
        raise                                          # let the caller handle it

try:
    process(0)
except ZeroDivisionError as e:
    print("Caller handled:", e)
````

**2. Re-raise After Cleanup with `finally`**

````python
def read_first_line(path):
    f = None
    try:
        f = open(path)
        return f.readline()
    except FileNotFoundError:
        print("Log: file missing:", path)
        raise
    finally:
        if f:
            f.close()

try:
    read_first_line("no_such_file.txt")
except FileNotFoundError as e:
    print("Caller handled:", e)
````
