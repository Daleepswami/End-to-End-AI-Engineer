# **03. Advanced Exception Handling Blocks**

### **3.1 Introduction to `else` Block**

**1. What is the `else` Block?**

- ***The `try`-`except` structure can be extended with `else` and `finally` blocks for more control over execution flow.***

- ***The code in the `else` block is executed only if the `try` block completes successfully, without any exceptions being raised.***

- ***It is a good place for code that should only run if no errors occurred.***

### **3.2 Working of `else` Block**

**1. How `else` Works**

- ***If no exception occurs in `try`, the `else` block runs after it.***

- ***If an exception occurs, the matching `except` block runs and the `else` block is skipped.***

````python
for text in ["25", "abc"]:
    try:
        number = int(text)
    except ValueError:
        print("Invalid:", text)
    else:                                      # runs only if no exception occurred
        print("Converted successfully:", number)
````

### **3.3 Introduction to `finally` Block**

**1. What is the `finally` Block?**

- ***The code in the `finally` block is always executed, regardless of whether an exception occurred in the `try` block or not, and regardless of whether it was handled by an `except` block.***

- ***It is typically used for cleanup operations that must happen, such as closing files or releasing network connections, even if an error crashes the main logic.***

### **3.4 Working of `finally` Block**

**1. `finally` Runs Every Time**

````python
for b in [4, 0]:
    try:
        print("Result:", 10 / b)
    except ZeroDivisionError:
        print("Cannot divide by zero.")
    finally:
        print("Done with 10 /", b)
````

**2. Cleanup Using `finally`**

- ***Use `finally` to release resources (close files, database connections).***

````python
f = None
try:
    f = open("File/finally_demo.txt", "w")
    f.write("data")
    result = 10 / 0                            # error after the file is opened
except ZeroDivisionError:
    print("An error occurred.")
finally:
    if f:
        f.close()                              # always closes the file
    print("File closed:", f.closed)
````

**3. `finally` with `return`**

- ***The `finally` block runs even when the function returns from the `try` block.***

````python
def demo():
    try:
        return "returned from try"
    finally:
        print("finally still runs")

print(demo())
````

### **3.5 Complete `try`-`except`-`else`-`finally` Structure**

**1. Syntax with `else` and `finally`:**

```python
try:
    # Code that might raise an exception
except SpecificException1:
    # Handler for SpecificException1
except SpecificException2:
    # Handler for SpecificException2
else:
    # Code to run if NO exception occurs in the 'try' block
finally:
    # Code that ALWAYS runs, regardless of exceptions
```

**2. Example**

````python
def divide_numbers(a, b):
    try:
        result = a / b
    except ZeroDivisionError:
        print("Caught an error: Cannot divide by zero!")
        return None  # Return None to indicate failure
    except TypeError:
        print("Caught an error: Invalid types for division (must be numbers)!")
        return None
    else:  # This block runs ONLY if no exception occurred in the 'try' block
        print("Division successful!")
        return result
    finally:  # This block ALWAYS runs
        print("--- Division attempt finished. ---")

# Test cases
print("Scenario 1: Successful division")
res1 = divide_numbers(10, 2)
if res1 is not None:
    print(f"Result: {res1}\n")

print("Scenario 2: Division by zero")
res2 = divide_numbers(10, 0)
if res2 is not None:
    print(f"Result: {res2}\n")

print("Scenario 3: Type error")
res3 = divide_numbers(10, "two")
if res3 is not None:
    print(f"Result: {res3}\n")
````

**3. Example: All Four Blocks Together**

````python
def safe_divide(a, b):
    try:
        result = a / b
    except ZeroDivisionError:
        print("Cannot divide by zero.")
    else:
        print("Result:", result)
    finally:
        print("Done with", a, "/", b)

safe_divide(10, 4)
safe_divide(10, 0)
````

**Interview Point:**

- ***What is the difference between `else` and `finally`? `else` runs only when the `try` block raised no exception; `finally` runs every time, with or without an exception.***

### **3.6 Nested `try`-`except` Blocks**

**1. What is a Nested `try`-`except`?**

- ***A `try`-`except` block can be placed inside another `try` or `except` block.***

- ***The inner block handles the errors it knows about.***

- ***If the inner block does not handle an exception, it moves to the outer block.***

````python
data = ["10", "0", "abc"]

try:
    for item in data:
        try:
            print(100 / int(item))
        except ZeroDivisionError:                  # inner handler
            print("Inner: cannot divide by zero for", item)
    print("Outer: loop completed")                 # skipped, because of the error below
except ValueError as e:                            # outer handler
    print("Outer caught:", e)
````

**2. Nested `try` Inside an `except` Block**

- ***An error can also occur while handling another error, so a `try` can be placed inside an `except` block.***

````python
try:
    value = int("abc")
except ValueError:
    print("Conversion failed, trying a fallback...")
    try:
        value = 10 / 0
    except ZeroDivisionError:
        print("Fallback also failed.")
        value = 0

print("value =", value)
````
