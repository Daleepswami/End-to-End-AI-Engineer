# **06. Custom User-Defined Exceptions**

### **6.1 What are Custom Exceptions?**

**1. What are Custom Exceptions?**

- ***Custom exceptions are exception classes created by the programmer.***

- ***They are raised with `raise` when a business rule is broken.***

- ***Custom exceptions make business rules explicit.***

**2. Why Use Custom Exceptions?**

1. ***The error name describes the problem clearly (for example, `InsufficientFundsError`).***
2. ***They can carry extra information, such as the balance and the amount.***
3. ***Callers can catch exactly the error they care about.***

### **6.2 Creating Custom Exception Classes**

**1. Simplest Custom Exception**

- ***A custom exception is just a class that inherits from `Exception`.***

- ***The body can be `pass`.***

```python
class InvalidMarksError(Exception):
    pass

try:
    raise InvalidMarksError("Marks must be between 0 and 100")
except InvalidMarksError as e:
    print("Error:", e)
```

### **6.3 Inheriting from the Exception Class**

**1. Inheriting from `Exception`**

- ***Create your own exception classes by inheriting from `Exception`.***

- ***Do not inherit from `BaseException` directly; it is meant for exits and interrupts (for example, `KeyboardInterrupt`).***

**2. Building a Hierarchy**

- ***A base custom exception can be created, and more specific exceptions can inherit from it.***

- ***Catching the base exception catches all the child exceptions as well.***

```python
class AppError(Exception):
    """Base class for all errors of this application."""

class ValidationError(AppError):
    pass

class DatabaseError(AppError):
    pass

try:
    raise ValidationError("Name cannot be empty")
except AppError as e:                                  # catches any AppError child
    print(type(e).__name__, "caught as AppError:", e)

print([c.__name__ for c in ValidationError.__mro__])
```

### **6.4 Adding Custom Error Messages**

**1. Custom Message Using `__init__`**

- ***Override `__init__` and pass a formatted message to `super().__init__()`.***

- ***You can also store extra values as attributes.***

```python
class InsufficientFundsError(Exception):
    def __init__(self, balance, amount):
        super().__init__(f'Balance {balance} is less than {amount}')
        self.balance = balance
        self.amount = amount

try:
    raise InsufficientFundsError(100, 500)
except InsufficientFundsError as e:
    print(e)                                           # Balance 100 is less than 500
    print("Short by:", e.amount - e.balance)
```

### **6.5 Using Custom Exceptions in Functions**

**1. Raising a Custom Exception in a Function**

- ***A function raises the custom exception when a rule is broken.***

- ***The caller handles it with `try`-`except`.***

```python
class InsufficientBalanceError(Exception):
    def __init__(self, balance, amount):
        super().__init__(f"Balance {balance} is less than requested {amount}")

def withdraw(balance, amount):
    if amount > balance:
        raise InsufficientBalanceError(balance, amount)
    return balance - amount

try:
    print(withdraw(5000, 2000))
    print(withdraw(5000, 8000))
except InsufficientBalanceError as e:
    print("Error:", e)
```

### **6.6 Practical Examples of Custom Exceptions**

**1. Example: `InvalidMarksError`**

- ***Raise `InvalidMarksError` when marks are outside 0 to 100.***

```python
class InvalidMarksError(Exception):
    def __init__(self, marks):
        super().__init__(f"Invalid marks: {marks}. Marks must be between 0 and 100.")
        self.marks = marks

def grade(marks):
    if not 0 <= marks <= 100:
        raise InvalidMarksError(marks)
    return "Pass" if marks >= 40 else "Fail"

for m in [75, 30, 120, -5]:
    try:
        print(m, "->", grade(m))
    except InvalidMarksError as e:
        print("Error:", e)
```

**2. Example: Age Validation**

```python
class UnderAgeError(Exception):
    def __init__(self, age, minimum=18):
        super().__init__(f"Age {age} is below the minimum age {minimum}")

def register(name, age):
    if age < 18:
        raise UnderAgeError(age)
    return f"{name} registered"

for name, age in [("Asha", 25), ("Ravi", 15)]:
    try:
        print(register(name, age))
    except UnderAgeError as e:
        print("Registration failed:", e)
```
