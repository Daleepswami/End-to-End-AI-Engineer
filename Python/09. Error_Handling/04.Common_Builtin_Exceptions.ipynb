# **04. Common Built-in Exceptions**

### **4.1 `ZeroDivisionError`**

**1. What is `ZeroDivisionError`?**

- ***It occurs when you attempt to divide a number by zero.***

- ***It is raised by `/`, `//` and `%` when the divisor is zero.***

```python
for expr in ["10 / 0", "10 // 0", "10 % 0"]:
    try:
        eval(expr)
    except ZeroDivisionError as e:
        print(expr, "->", e)
```

**2. Fix: Check the Divisor First**

```python
a, b = 10, 0
if b != 0:
    print(a / b)
else:
    print("Cannot divide by zero.")
```

### **4.2 `TypeError`**

**1. What is `TypeError`?**

- ***It occurs when an operation is performed on an inappropriate type.***

- ***Example: adding a string and an integer, or changing a tuple (tuples are immutable).***

```python
try:
    "hello" + 5
except TypeError as e:
    print("Error:", e)

point = (10, 20)
try:
    point[0] = 99
except TypeError as e:
    print("Error:", e)
```

**2. Fix: Convert to the Correct Type**

```python
print("hello" + str(5))
```

### **4.3 `ValueError`**

**1. What is `ValueError`?**

- ***It occurs when a function receives an argument of the correct type but an inappropriate value.***

- ***Example: `'abc'` is a string, but it cannot be converted to `int`.***

```python
try:
    int("abc")
except ValueError as e:
    print("Error:", e)
```

**2. Fix: Wrap Risky Conversions in `try`-`except ValueError`**

- ***This is important when the value comes from users.***

```python
text = "abc"
try:
    number = int(text)
except ValueError:
    number = 0
    print("Invalid number, using default 0")
print(number)
```

### **4.4 `NameError`**

**1. What is `NameError`?**

- ***It occurs when you try to use a variable or function name that has not been defined.***

- ***It also occurs when a variable has been deleted using `del`.***

```python
try:
    print(undefined_variable)
except NameError as e:
    print("Error:", e)

word = "python"
del word
try:
    print(word)
except NameError as e:
    print("Error:", e)
```

### **4.5 `IndexError`**

**1. What is `IndexError`?**

- ***It occurs when you try to access an index that is out of bounds for a sequence (list, string, tuple).***

```python
my_list = [1, 2, 3]
try:
    print(my_list[3])
except IndexError as e:
    print("Error:", e)
```

**2. Fix: Check the Length First**

```python
index = 3
if index < len(my_list):
    print(my_list[index])
else:
    print("Index out of range.")
```

### **4.6 `KeyError`**

**1. What is `KeyError`?**

- ***It occurs when you try to access a non-existent key in a dictionary.***

```python
my_dict = {"name": "Alice"}
try:
    print(my_dict["age"])
except KeyError as e:
    print("Missing key:", e)
```

**2. Fix: Use `get()` with a Default Value**

```python
print(my_dict.get("age", 0))                  # 0 (default if key missing)
```

### **4.7 `AttributeError`**

**1. What is `AttributeError`?**

- ***It occurs when you use an attribute (or method) that does not exist on an object.***

- ***Private attributes with a double underscore are name-mangled, so accessing them from outside also raises `AttributeError`.***

```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.__balance = balance                       # private attribute

    @property
    def balance(self):                                 # read-only access
        return self.__balance

acc = BankAccount("Asha", 5000)
print(acc.balance)

try:
    print(acc.__balance)
except AttributeError as e:
    print("Error:", e)

try:
    "hello".foo()
except AttributeError as e:
    print("Error:", e)
```

### **4.8 `FileNotFoundError`**

**1. What is `FileNotFoundError`?**

- ***It occurs when you try to open a file that does not exist.***

```python
try:
    open("non_existent_file.txt", "r")
except FileNotFoundError as e:
    print("Error:", e)
```

### **4.9 `FileExistsError`**

**1. What is `FileExistsError`?**

- ***It occurs in exclusive creation mode (`"x"`) when the file already exists.***

- ***This is useful to ensure you are creating a new file.***

```python
import os

if os.path.exists("unique_file.txt"):
    os.remove("unique_file.txt")               # start fresh so the demo can be re-run

for attempt in (1, 2):
    try:
        with open("unique_file.txt", "x") as f:
            f.write("Created once.")
        print("Attempt", attempt, ": file created.")
    except FileExistsError as e:
        print("Attempt", attempt, ": Error -", e)
```

### **4.10 `ImportError` and `ModuleNotFoundError`**

**1. What are `ImportError` and `ModuleNotFoundError`?**

- ***These errors occur when Python cannot find a module you are trying to import, or encounters an issue during the import process itself.***

- ***`ModuleNotFoundError` is more specific. It typically means the module file simply does not exist or is not in Python's search path.***

- ***`ImportError` is a broader category that can include `ModuleNotFoundError` or other issues preventing a successful import (for example, a name that does not exist in the module).***

```python
try:
    import non_existent_module
except ModuleNotFoundError as e:
    print("ModuleNotFoundError:", e)

try:
    from math import power               # 'power' is not a direct function in math module (use math.pow)
except ImportError as e:
    print("ImportError:", e)

print(issubclass(ModuleNotFoundError, ImportError))
```

### **4.11 `RuntimeError`**

**1. What is `RuntimeError`?**

- ***`RuntimeError` is a general run-time error that does not fit into any other category.***

- ***`RecursionError` (maximum recursion depth exceeded) is a special kind of `RuntimeError`.***

- ***Python's default recursion limit is 1000.***

```python
try:
    raise RuntimeError("Something went wrong at run time")
except RuntimeError as e:
    print("RuntimeError:", e)

def infinite(n):
    return infinite(n + 1)

try:
    infinite(0)
except RecursionError as e:
    print("RecursionError:", e)

print(issubclass(RecursionError, RuntimeError))
```

### **4.12 `OverflowError`**

**1. What is `OverflowError`?**

- ***It occurs when the result of a numeric calculation is too large to be represented.***

- ***It usually appears with floating-point calculations.***

- ***Python integers have unlimited precision, so normal integer calculations do not overflow.***

```python
import math

try:
    math.exp(1000)
except OverflowError as e:
    print("Error:", e)

try:
    2.0 ** 10000
except OverflowError as e:
    print("Error:", e)

print(len(str(2 ** 10000)), "digits - no error for int")
```

### **4.13 `MemoryError`**

**1. What is `MemoryError`?**

- ***It occurs when an operation runs out of memory.***

- ***Example: trying to create an extremely large list.***

```python
try:
    huge = [0] * (10 ** 13)
except MemoryError:
    print("Not enough memory to create this list.")
```

### **4.14 `OSError` and its Common Subclasses**

**1. What is `OSError`?**

- ***`OSError` is the base class for errors related to the operating system and file operations.***

- ***Many common file errors are its subclasses, so `except OSError` can catch all of them.***

| Exception | When it Occurs |
|-----------|----------------|
| `FileNotFoundError` | The file or folder does not exist |
| `FileExistsError` | The file already exists (for example, in `"x"` mode) |
| `PermissionError` | No rights to read or write the file |
| `IsADirectoryError` | A folder was given instead of a file |
| `NotADirectoryError` | A file was given where a folder was expected |

```python
for exc in [FileNotFoundError, FileExistsError, PermissionError, IsADirectoryError, NotADirectoryError]:
    print(exc.__name__, "is a subclass of OSError:", issubclass(exc, OSError))

try:
    open("missing_folder/data.txt")
except OSError as e:                         # catches FileNotFoundError too
    print(type(e).__name__, "|", e.errno, "|", e.strerror)
```
