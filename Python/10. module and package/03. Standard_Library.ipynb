# **08. Important Standard Library Modules**

### **8.1 What is the Python Standard Library?**

**1. What is the Standard Library?**

- ***The Python Standard Library is a large collection of built-in modules that come with Python.***

- ***No installation is needed, so a module can be used with a simple `import`.***

- ***It provides ready-made modules for files, maths, dates, networking and more.***

**2. Important Standard Library Modules**

| Module | Key Uses |
|--------|----------|
| `math` | `sqrt`, `floor`, `ceil`, `pi`, `factorial`, `log` |
| `random` | `random()`, `randint()`, `choice()`, `shuffle()`, `sample()` |
| `datetime` | `date`, `time`, `datetime`, `timedelta` (dates and times) |
| `os` | File system, environment variables |
| `sys` | `argv`, `path`, `exit()`, `version` |
| `re` | Regular expressions: `search`, `match`, `findall`, `sub` |
| `json` | `load`, `dump`: read and write JSON |
| `collections` | `defaultdict`, `Counter`, `deque`, `OrderedDict`, `namedtuple` |
| `itertools` | `product`, `combinations`, `permutations`, `chain` |
| `functools` | `reduce`, `partial`, `lru_cache` (memoization) |

```python
import sys

try:
    print(len(sys.stdlib_module_names), "standard modules (Python 3.10+)")
    print("math" in sys.stdlib_module_names, "numpy" in sys.stdlib_module_names)
except AttributeError:
    print("sys.stdlib_module_names needs Python 3.10 or newer")
```

### **8.2 `math` – Mathematical Operations**

**1. What is the `math` Module?**

- ***The `math` module provides mathematical functions and constants.***

| Function / Constant | Description |
|---------------------|-------------|
| `math.sqrt(x)` | Square root |
| `math.floor(x)` | Round down to the nearest integer |
| `math.ceil(x)` | Round up to the nearest integer |
| `math.pi` | The value of pi |
| `math.factorial(n)` | Factorial of n |
| `math.log(x, base)` | Logarithm (natural log if no base is given) |
| `math.pow(x, y)` | x raised to the power y (returns a float) |
| `math.gcd(a, b)` | Greatest common divisor |

```python
import math

print(math.sqrt(16))              # 4.0
print(math.floor(4.7))            # 4
print(math.ceil(4.2))             # 5
print(math.pi)                    # 3.141592653589793
print(math.factorial(5))          # 120
print(math.log(100, 10))          # 2.0
print(math.pow(2, 3))             # 8.0
print(math.gcd(12, 18))           # 6
```

### **8.3 `random` – Random Number Generation**

**1. What is the `random` Module?**

- ***The `random` module generates random numbers and makes random choices.***

| Function | Description |
|----------|-------------|
| `random.random()` | Random float from 0.0 up to (but not including) 1.0 |
| `random.randint(a, b)` | Random integer from a to b (both included) |
| `random.choice(seq)` | One random item from a sequence |
| `random.shuffle(lst)` | Shuffles a list in place |
| `random.sample(seq, k)` | k unique random items |
| `random.seed(n)` | Makes the random numbers repeatable |

```python
import random

random.seed(42)                                   # same seed -> same random numbers every time

print(random.random())
print(random.randint(1, 10))
print(random.choice(["red", "green", "blue"]))

cards = [1, 2, 3, 4, 5]
random.shuffle(cards)
print(cards)

print(random.sample(range(1, 50), 5))
```

**2. Example: Rolling Two Dice**

```python
die1 = random.randint(1, 6)
die2 = random.randint(1, 6)
print("Dice:", die1, die2, "| Total:", die1 + die2)
```

### **8.4 `datetime` – Date and Time**

**1. What is the `datetime` Module?**

- ***The `datetime` module works with dates and times.***

- ***Its main classes are `date`, `time`, `datetime` and `timedelta`.***

| Class | Description |
|-------|-------------|
| `date` | A calendar date (year, month, day) |
| `time` | A time of day (hour, minute, second) |
| `datetime` | A date and a time together |
| `timedelta` | A duration, used for date arithmetic |

```python
from datetime import date, time, datetime, timedelta

d = date(2026, 1, 1)
print(d, d.year, d.month, d.day)

t = time(14, 30)
print(t)

now = datetime.now()                                    # current date and time
print(now.strftime("%Y-%m-%d %H:%M:%S"))               # format as text
```

**2. Date Arithmetic with `timedelta`**

```python
print(d + timedelta(days=30))                           # 30 days later
print((date(2026, 12, 25) - date(2026, 1, 1)).days)    # days between two dates

parsed = datetime.strptime("2026-03-15 10:30", "%Y-%m-%d %H:%M")      # text -> datetime
print(parsed)
```

### **8.6 `sys` – System-Specific Functions**

**1. What is the `sys` Module?**

- ***The `sys` module gives access to details about the Python interpreter and the system.***

| Item | Description |
|------|-------------|
| `sys.argv` | List of command-line arguments (`sys.argv[0]` is the script name) |

**2. Command-Line Arguments with `sys.argv`**

```python
import subprocess

with open("args_demo.py", "w") as f:
    f.write("""import sys
print("Script name:", sys.argv[0])
print("Arguments  :", sys.argv[1:])
""")

result = subprocess.run([sys.executable, "args_demo.py", "hello", "5"], capture_output=True, text=True)
print(result.stdout)
```

- ***`sys.exit()` raises a `SystemExit` exception, which ends the program.***

```python
try:
    sys.exit("Stopping the program")
except SystemExit as e:
    print("SystemExit caught:", e)
```

### **8.7 `re` – Regular Expressions**

**1. What is the `re` Module?**

- ***The `re` module is used for pattern matching in text with regular expressions.***

| Function | Description |
|----------|-------------|
| `re.search(pattern, text)` | Finds the first match anywhere in the text |
| `re.match(pattern, text)` | Matches only at the beginning of the text |
| `re.findall(pattern, text)` | Returns a list of all matches |
| `re.sub(pattern, new, text)` | Replaces matches with new text |

```python
import re

text = "Contact: asha@example.com, ravi@test.org, phone 98765-43210"

print(re.findall(r"\w+@\w+\.\w+", text))              # all email addresses
print(re.search(r"\d{5}-\d{5}", text).group())        # first phone number
print(bool(re.match(r"Contact", text)))               # does the text start with "Contact"?
print(re.sub(r"\d", "#", "Room 101"))                 # replace every digit
```

### **8.8 `json` – JSON Data Handling**

**1. What is the `json` Module?**

- ***The `json` module reads and writes JSON data.***

- ***`dump()` and `load()` work with files, while `dumps()` and `loads()` work with strings.***

```python
import json

data = {'name': 'Alice', 'scores': [90, 85, 92]}

with open('data.json', 'w') as f:
    json.dump(data, f, indent=4)                      # Python -> JSON file

with open('data.json') as f:
    loaded = json.load(f)                             # JSON file -> Python
    print(loaded['name'])                             # Alice

text = json.dumps(data)                               # Python -> JSON string
print(text)
print(json.loads(text)["scores"])                     # JSON string -> Python
```

### **8.9 `collections` – Specialized Data Structures**

**1. What is the `collections` Module?**

- ***The `collections` module provides special container types that extend the built-in list, dict and tuple.***

| Type | Description |
|------|-------------|
| `Counter` | Counts how many times each item appears |
| `defaultdict` | A dictionary that creates a default value for a missing key |
| `deque` | A fast double-ended queue |
| `OrderedDict` | A dictionary with extra order-related methods |
| `namedtuple` | A tuple whose items can be accessed by name |

```python
from collections import Counter, defaultdict

print(Counter("mississippi").most_common(2))          # [('i', 4), ('s', 4)]

groups = defaultdict(list)
for name, dept in [("Asha", "tech"), ("Ravi", "sales"), ("Meena", "tech")]:
    groups[dept].append(name)
print(dict(groups))                                   # {'tech': ['Asha', 'Meena'], 'sales': ['Ravi']}
```

```python
from collections import deque, OrderedDict, namedtuple

queue = deque()
queue.append("Customer 1"); queue.append("Customer 2"); queue.append("Customer 3")
print("Serving:", queue.popleft())
print("Waiting:", list(queue))

od = OrderedDict(a=1, b=2, c=3)
od.move_to_end("a")                                   # move a key to the end
print(list(od))

Point = namedtuple('Point', ['x', 'y'])
p = Point(3, 7)
print(p.x, p.y)                                       # 3 7
print(p[0])                                           # 3 (still index-accessible)
```

### **8.10 `itertools` – Iteration Tools**

**1. What is the `itertools` Module?**

- ***The `itertools` module provides fast tools for working with iterators and for creating combinations.***

| Function | Description |
|----------|-------------|
| `product(a, b)` | All pairs: one item from each iterable |
| `combinations(seq, r)` | All groups of r items, order ignored |
| `permutations(seq, r)` | All arrangements of r items, order matters |
| `chain(a, b)` | Joins several iterables into one |

```python
from itertools import product, combinations, permutations, chain

print(list(product([1, 2], ['a', 'b'])))              # [(1, 'a'), (1, 'b'), (2, 'a'), (2, 'b')]
print(list(combinations('ABC', 2)))                   # [('A', 'B'), ('A', 'C'), ('B', 'C')]
print(list(permutations([1, 2, 3], 2)))               # 6 arrangements
print(list(chain([1, 2], [3, 4])))                    # [1, 2, 3, 4]
```

### **8.11 `functools` – Functional Programming Tools**

**1. What is the `functools` Module?**

- ***The `functools` module provides tools for working with functions and for functional-style programming.***

| Function | Description |
|----------|-------------|
| `reduce(func, seq)` | Combines all items of a sequence into one value |
| `partial(func, ...)` | Creates a new function with some arguments already fixed |
| `lru_cache` | Remembers results of a function (memoization) |

```python
from functools import reduce, partial, lru_cache

print(reduce(lambda a, b: a * b, [1, 2, 3, 4, 5]))    # 120

def power(base, exp):
    return base ** exp

square = partial(power, exp=2)                        # exp is fixed to 2
cube = partial(power, exp=3)
print(square(5), cube(5))                             # 25 125
```

**2. Memoization with `lru_cache`**

- ***`lru_cache` stores the result for each input, so a repeated call does not need to recalculate.***

```python
@lru_cache(maxsize=None)
def fib(n):
    if n <= 1:
        return n
    return fib(n - 1) + fib(n - 2)

print(fib(30))                                        # 832040
print(fib.cache_info())
```
