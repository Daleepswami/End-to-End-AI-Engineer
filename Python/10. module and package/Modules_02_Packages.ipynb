# **05. Introduction to Packages**

### **5.1 What is a Package?**

**1. What is a Package?**

- ***A package is a directory containing an `__init__.py` file and multiple modules.***

- ***This lets you organise related modules under a namespace.***

### **5.2 Why Do We Use Packages?**

**1. Reasons for Using Packages**

1. **Organization:** A large project has too many modules for one folder. Packages group related modules together.
2. **Namespaces:** Two modules with the same name can live in different packages without clashing.
3. **Reuse and Sharing:** A package can be shared with other programmers (for example, `pandas`, `numpy`, `requests`).
4. **Clear Structure:** The folder structure shows how the project is organised.

### **5.3 Module vs Package**

| Feature | Module | Package |
|---------|--------|---------|
| What it is | A single `.py` file | A directory of modules with an `__init__.py` file |
| Contains | Functions, variables, classes | Modules (and other packages) |
| Example | `mymath.py` | `mypackage/` |
| Import | `import mymath` | `import mypackage.math_utils` |

````python
import math, json

print(hasattr(math, "__path__"))       # False -> math is a module
print(hasattr(json, "__path__"))       # True  -> json is a package

import json.decoder                    # a module inside the json package
print(json.decoder.__name__)
````

### **5.4 Benefits of Packages**

| Benefit | Meaning |
|---------|---------|
| Better Organization | Related modules are grouped in one folder |
| No Name Conflicts | The package name acts as a namespace |
| Easy Reuse | The whole group of modules can be imported and shared |
| Scalability | A project can grow without becoming messy |
| Teamwork | Different teams can own different packages |

### **5.5 Understanding Package Organization**

**1. A Simple Package**

```
mypackage/
    __init__.py
    math_utils.py
    string_utils.py
```

**2. A Larger Package with a Sub-package**

```
mypackage/
    __init__.py
    math_utils.py
    string_utils.py
    extras/              <- sub-package
        __init__.py
        helpers.py
```

**3. Usage:**

```python
from mypackage import math_utils
from mypackage.string_utils import capitalize_all
```

# **06. Package Structure**

### **6.1 Creating a Package Directory**

**1. The Package Directory**

- ***A package starts as an ordinary folder.***

- ***The folder name becomes the package name, so use a short, lowercase name without spaces.***

````python
import os, importlib

os.makedirs("mypackage", exist_ok=True)
print(os.path.isdir("mypackage"))
````

### **6.2 Understanding `__init__.py`**

**1. What is `__init__.py`?**

- ***`__init__.py` is a special file that marks a directory as a package.***

- ***It can be empty.***

- ***Python runs it automatically when the package is imported, so it can also hold start-up code.***

````python
with open("mypackage/__init__.py", "w") as f:
    f.write("""# mypackage/__init__.py
print("mypackage is being imported")
""")
````

### **6.3 Creating Multiple Modules inside a Package**

**1. Modules inside a Package**

- ***Modules are ordinary `.py` files placed inside the package folder.***

- ***Here, `math_utils.py` holds number functions and `string_utils.py` holds text functions.***

````python
with open("mypackage/math_utils.py", "w") as f:
    f.write("""# mypackage/math_utils.py
def add(a, b):
    return a + b


def multiply(a, b):
    return a * b


def square(n):
    return n * n
""")

with open("mypackage/string_utils.py", "w") as f:
    f.write("""# mypackage/string_utils.py
def capitalize_all(words):
    return [w.capitalize() for w in words]


def shout(text):
    return text.upper() + "!"
""")

importlib.invalidate_caches()
print(sorted(os.listdir("mypackage")))
````

### **6.4 Organizing Related Modules**

**1. Guidelines**

- ***Put modules that work on the same topic in the same package.***

- ***Use a sub-package (a folder inside a package, with its own `__init__.py`) for a smaller group of modules.***

- ***Keep module names short and meaningful.***

````python
os.makedirs("mypackage/extras", exist_ok=True)

with open("mypackage/extras/__init__.py", "w") as f:
    f.write("# mypackage/extras/__init__.py\n")

with open("mypackage/extras/helpers.py", "w") as f:
    f.write("""# mypackage/extras/helpers.py
def greet(name):
    return f"Welcome, {name}!"
""")

importlib.invalidate_caches()

for folder, subfolders, files in os.walk("mypackage"):
    if "__pycache__" in folder:
        continue
    print(folder, sorted(files))
````

### **6.5 Understanding Package Initialization**

**1. Package Initialization**

- ***The code in `__init__.py` runs once, the first time the package (or anything inside it) is imported.***

- ***Later imports reuse the already loaded package, so the code does not run again.***

````python
import mypackage             # first import: __init__.py runs
import mypackage             # second import: nothing is printed

print(mypackage.__name__)
````

# **07. Importing from Packages**

### **7.1 Importing a Module from a Package**

**1. `import package.module`**

- ***Importing a module from a package uses the dotted name `package.module`.***

- ***Functions are then called with the full dotted name.***

````python
import mypackage.math_utils

print(mypackage.math_utils.add(2, 3))          # 5

import mypackage.extras.helpers                # a module inside a sub-package
print(mypackage.extras.helpers.greet("Asha"))
````

### **7.2 Importing Specific Functions from a Module**

**1. Importing Chosen Functions**

- ***Use `from package.module import name1, name2` to import more than one function from a module of a package.***

- ***The functions can then be used without any prefix.***

````python
from mypackage.math_utils import add, multiply, square

print(add(2, 3), multiply(4, 5), square(6))
````

### **7.3 `from package import module`**

**1. Importing a Module**

- ***`from package import module` imports the module itself.***

- ***The module name is then used as a prefix.***

- ***It is shorter than writing the full dotted name each time.***

````python
from mypackage import math_utils

print(math_utils.add(2, 3))                    # 5
print(math_utils.square(4))

from mypackage.extras import helpers           # sub-package
print(helpers.greet("Ravi"))
````

### **7.4 `from package.module import function`**

**1. Importing a Single Function**

- ***`from package.module import function` imports one function directly.***

````python
from mypackage.string_utils import capitalize_all

print(capitalize_all(["asha", "ravi", "meena"]))

from mypackage.string_utils import shout as loud      # with an alias
print(loud("hello"))
````

**2. Summary of Import Forms**

| Statement | How the function is used |
|-----------|--------------------------|
| `import mypackage.math_utils` | `mypackage.math_utils.add(2, 3)` |
| `from mypackage import math_utils` | `math_utils.add(2, 3)` |
| `from mypackage.math_utils import add` | `add(2, 3)` |
| `from mypackage.math_utils import add as plus` | `plus(2, 3)` |

### **7.5 Using `__init__.py` to Expose Functions**

**1. Exposing Functions**

- ***`__init__.py` can import functions from the modules of the package.***

- ***Then users can import them directly from the package, without knowing the module names.***

- ***A dot in `from .math_utils import add` means "from this same package".***

- ***`__all__` lists the names that are exported by `from package import *`.***

````python
os.makedirs("toolkit", exist_ok=True)

with open("toolkit/__init__.py", "w") as f:
    f.write("""# toolkit/__init__.py
from .math_utils import add, multiply
from .string_utils import capitalize_all

__all__ = ["add", "multiply", "capitalize_all"]
""")

with open("toolkit/math_utils.py", "w") as f:
    f.write("""def add(a, b):
    return a + b


def multiply(a, b):
    return a * b
""")

with open("toolkit/string_utils.py", "w") as f:
    f.write("""def capitalize_all(words):
    return [w.capitalize() for w in words]
""")

importlib.invalidate_caches()
````

````python
from toolkit import add, capitalize_all        # no need to know the module names

print(add(2, 3))
print(capitalize_all(["python", "modules"]))
````

**2. Before and After**

| Without exposing | With `__init__.py` exposing |
|------------------|-----------------------------|
| `from mypackage.math_utils import add` | `from toolkit import add` |
