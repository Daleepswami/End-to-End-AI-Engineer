# **01. Introduction to Modules**

### **1.1 What is a Module?**

**1. What is a Module?**

- ***A module is any Python file (`.py`). You can import it to reuse its code.***

- ***A module is simply a file containing Python code (functions, classes, variables) that you can reuse in other programs.***

- ***Modules keep code organised and avoid rewriting the same logic.***

**2. Types of Modules**

| Type of Module | Description | Examples |
|----------------|-------------|----------|
| Built-in modules | Ship with Python; no installation needed | `math`, `random`, `os`, `datetime` |
| External modules | Created by others; installed with `pip` | `pandas`, `requests`, `numpy` |
| User-defined modules | Your own `.py` files | `utils.py`, `calculator.py` |

````python
import math

print(type(math))                    # <class 'module'>
print(math.sqrt(16))                 # 4.0
print([n for n in dir(math) if not n.startswith("_")][:8])    # dir() lists what a module provides
````

**Tip:**

- ***Use `dir(module)` to list everything a module provides, and `help(module.function)` to read its documentation.***

### **1.2 Why Do We Use Modules?**

**1. Reasons for Using Modules**

1. **Real programs are rarely written in a single file.** Splitting code into modules keeps each file small and clear.
2. **Reuse:** Write a function once and use it in many programs.
3. **Organization:** Related code is kept together (for example, all maths functions in one module).
4. **Ready-made Tools:** Python's built-in and external modules give thousands of ready-made features.
5. **Avoid Repetition:** There is no need to copy and paste the same code again and again.

### **1.3 Benefits of Modules**

| Benefit | Meaning |
|---------|---------|
| Reusability | The same code can be used in many programs |
| Organization | Related functions, classes and variables stay together |
| Maintainability | A bug is fixed in one place, and every program gets the fix |
| Separate Namespace | Names inside a module do not clash with names in other modules |
| Teamwork | Different people can work on different modules |
| Readability | Smaller files are easier to read and understand |

### **1.4 Python `.py` File as a Module**

**1. Every `.py` File is a Module**

- ***Any Python file can be imported as a module from the same folder.***

- ***The module name is the file name without `.py` (for example, `calculator.py` becomes the module `calculator`).***

**2. Rules for Module Names**

- ***Use only letters, digits and underscores, and do not start with a digit.***

- ***Do not use hyphens or spaces.***

- ***Do not use the name of a standard module (for example, do not create your own `random.py`).***

````python
import math, os

print(math.__name__)       # every module has a __name__ attribute
print(os.__name__)
````

### **1.5 Understanding Module Reusability**

**1. Write Once, Use Many Times**

- ***A module is written once and imported wherever it is needed.***

- ***Here, `calculator.py` is a module file, and two different programs use it.***

````python
with open("File/calculator.py", "w") as f:
    f.write("""# calculator.py
def add(a, b):
    return a + b


def multiply(a, b):
    return a * b
""")

import importlib
importlib.invalidate_caches()              # let Python notice the new file
````

````python
# Program A
from File.calculator import add, multiply
print("Program A:", add(4, 5))
print("Program A:", multiply(3, 7))

# Program B reuses the same module
import File.calculator
print("Program B:", File.calculator.add(100, 200))
````

# **02. Creating a Module**

### **2.1 Creating a Python Module**

**1. Steps to Create a Module**

1. ***Create a new file with the `.py` extension (for example, `mymath.py`).***
2. ***Write functions, variables or classes in it.***
3. ***Save it in the same folder as the program that will use it.***

**2. The Module File:**

```python
# mymath.py
PI = 3.14159
VERSION = "1.0"

def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def area_of_circle(radius):
    return PI * radius ** 2
```

````python
with open("mymath.py", "w") as f:
    f.write("""# mymath.py
PI = 3.14159
VERSION = "1.0"


def add(a, b):
    return a + b


def subtract(a, b):
    return a - b


def area_of_circle(radius):
    return PI * radius ** 2
""")

importlib.invalidate_caches()
print("mymath.py created")
````

### **2.2 Defining Functions inside a Module**

**1. Functions in a Module**

- ***A module can contain any number of functions.***

- ***After importing the module, each function is used as `module_name.function_name()`.***

````python
import mymath

print(mymath.add(2, 3))                # 5
print(mymath.subtract(10, 4))          # 6
print(mymath.area_of_circle(2))        # 12.56636
````

**2. Documenting a Function with a Docstring**

- ***A docstring is the first string inside a function. It is available through `help()` and `__doc__`.***

````python
def area(radius):
    """Return the area of a circle with the given radius."""      # docstring
    return 3.14159 * radius ** 2

print(area.__doc__)
````

### **2.3 Defining Variables inside a Module**

**1. Variables in a Module**

- ***A module can also store variables (often constants such as `PI`).***

- ***They are created when the module is imported, and are used as `module_name.variable_name`.***

- ***A module can contain functions, variables and classes.***

````python
print(mymath.PI)               # 3.14159
print(mymath.VERSION)          # 1.0
````

### **2.4 Using One Python File in Another Python File**

**1. Importing a Module into Another File**

- ***A program in the same folder can import the module with `import`.***

- ***`main.py` is the program file that uses the module `mymath`.***

**2. The Program File:**

```python
# main.py
import mymath

print(mymath.add(2, 3))     # 5
print(mymath.PI)            # 3.14159
```

````python
import subprocess, sys

with open("main.py", "w") as f:
    f.write("""# main.py
import mymath

print(mymath.add(2, 3))     # 5
print(mymath.PI)            # 3.14159
""")

result = subprocess.run([sys.executable, "main.py"], capture_output=True, text=True)
print(result.stdout)
````

**3. Running a File vs Importing It**

- ***When a file is run directly, its `__name__` is `"__main__"`.***

- ***When it is imported, its `__name__` is the module name.***

- ***`if __name__ == "__main__":` lets a file contain code that runs only when the file is run directly.***

````python
with open("greetings.py", "w") as f:
    f.write("""# greetings.py
def say_hello(name):
    return f"Hello, {name}!"


if __name__ == "__main__":
    print("greetings.py is being run directly")
    print(say_hello("Asha"))
""")

importlib.invalidate_caches()

print("--- Running greetings.py directly ---")
print(subprocess.run([sys.executable, "greetings.py"], capture_output=True, text=True).stdout)

print("--- Importing greetings.py ---")
import greetings                         # the block under __main__ does not run
print(greetings.say_hello("Ravi"))
````

# **03. Importing Modules**

### **3.1 What is Importing?**

**1. What is Importing?**

- ***Importing means loading a module so that its functions, variables and classes can be used in your program.***

**2. What Python Does When You Import**

1. ***Python searches for the module in the current folder, then in the standard library and installed packages (the list `sys.path`).***
2. ***It runs the module file once and creates a module object.***
3. ***The module is stored in `sys.modules`, so importing it again does not run the file again.***

````python
with open("loaddemo.py", "w") as f:
    f.write("""print("loaddemo.py is being loaded")
VALUE = 42
""")

importlib.invalidate_caches()

import loaddemo              # first import: the file runs
import loaddemo              # second import: reused from sys.modules, so nothing is printed
print(loaddemo.VALUE)
print("loaddemo" in sys.modules)
````

### **3.2 `import` Statement**

**1. What is the `import` Statement?**

- ***`import module` imports the whole module.***

- ***Names are then used with the module name as a prefix.***

**2. Syntax:**

```python
import module_name
```

````python
import math                                 # import the whole module
print(math.sqrt(16))                        # 4.0
````

**3. Importing Several Modules**

- ***Several modules can be imported by separating their names with commas, but one import per line is easier to read.***

```python
import math, random
```

### **3.3 `from ... import` Statement**

**1. What is `from ... import`?**

- ***`from module import name` imports only the chosen names.***

- ***The names can be used directly, without the module name.***

**2. Syntax:**

```python
from module_name import name1, name2
```

````python
from math import pi, ceil                   # import specific names
print(pi, ceil(4.2))                        # 3.141592653589793 5

from mymath import add, PI
print(add(10, 5))                           # 15
print(PI)
````

### **3.4 Importing Multiple Functions**

**1. Importing More Than One Name**

- ***Several names can be imported in one statement by separating them with commas.***

- ***For a long list, wrap the names in parentheses and write one name per line.***

- ***`from module import *` imports every public name, but it is not recommended because it hides where names come from and can overwrite existing names.***

````python
from mymath import add, subtract, area_of_circle

print(add(7, 3), subtract(7, 3), area_of_circle(1))

from mymath import (
    add,
    subtract,
    area_of_circle,
)
````

### **3.5 Difference between `import` and `from ... import`**

| Feature | `import module` | `from module import name` |
|---------|-----------------|---------------------------|
| What is imported | The whole module | Only the chosen names |
| How a function is called | `mymath.add(2, 3)` | `add(2, 3)` |
| Source of a name | Clear from the prefix | Not visible when used |
| Name clashes | Unlikely, because of the prefix | Possible, because names are copied into your program |
| Best for | Many names from one module | A few names that are used often |

**1. Name Clash Example**

- ***`from math import pow` replaces Python's built-in `pow()` with `math.pow()`.***

````python
print(pow(2, 3))               # 8   (built-in pow, integer result)

from math import pow
print(pow(2, 3))               # 8.0 (math.pow, float result)

del pow                        # remove the imported name, so the built-in works again
print(pow(2, 3))
````

**2. Three Ways of Importing `sqrt`**

````python
import math
print(math.sqrt(16))

from math import sqrt
print(sqrt(16))

import math as m
print(m.sqrt(16))
````

# **04. Aliasing Modules**

### **4.1 What is Aliasing?**

**1. What is Aliasing?**

- ***Aliasing means giving an imported module or function a different (usually shorter) name.***

- ***It makes long names easier to type, and it avoids name clashes.***

### **4.2 `as` Keyword**

**1. The `as` Keyword**

- ***The `as` keyword creates the alias while importing.***

**2. Syntax:**

```python
import module_name as alias
from module_name import name as alias
```

### **4.3 Creating an Alias for a Module**

**1. Module Alias**

- ***`import module as alias` lets you use the alias instead of the full module name.***

````python
import datetime as dt                       # import with an alias
print(dt.date(2026, 1, 1).year)             # 2026

import mymath as mm
print(mm.add(1, 2))
````

### **4.4 Creating an Alias for a Function**

**1. Function Alias**

- ***`from module import function as alias` renames the imported function.***

- ***This is useful for shortening a name, or when two modules have functions with the same name.***

````python
from mymath import area_of_circle as circle_area
from math import sqrt as square_root
from math import pow as math_pow             # avoids clashing with the built-in pow()

print(circle_area(2))
print(square_root(25))
print(math_pow(2, 3), pow(2, 3))
````

### **4.5 Common Aliases: `np`, `pd`, `plt`**

**1. Standard Aliases**

- ***Some external libraries have aliases that every Python programmer uses.***

- ***Follow these standard aliases, so that your code is easy for others to read.***

- ***External modules are installed with `pip` (for example, `pip install numpy`).***

| Library | Purpose | Common Alias |
|---------|---------|--------------|
| NumPy | Numerical computing: arrays, linear algebra, broadcasting | `np` |
| Pandas | Data manipulation: DataFrames, CSV/Excel, groupby, merge | `pd` |
| Matplotlib (`pyplot`) | 2D plotting: line, bar, scatter, histogram, heatmap | `plt` |
| Seaborn | Statistical visualisation built on Matplotlib | `sns` |
| TensorFlow | Deep learning framework by Google | `tf` |

**2. Quick NumPy Example**

````python
try:
    import numpy as np

    a = np.array([1, 2, 3, 4, 5])
    print(a.mean())             # 3.0
    print(a * 2)                # [ 2  4  6  8 10]
except ModuleNotFoundError:
    print("NumPy is not installed. Install it with: pip install numpy")
````
