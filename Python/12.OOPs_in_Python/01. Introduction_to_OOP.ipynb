## **Introduction to Object-Oriented Programming (OOP)**

#### **What is OOP?**

*Object-Oriented Programming (OOP) is a way of writing programs where we organize code around **objects** instead of only functions and logic.*

*An object is a single unit that holds both:*

*1. **Data** → called **attributes** (what the object has)*

*2. **Behaviour** → called **methods** (what the object can do)*

**Key Points**

*1. OOP models real-world things (student, car, bank account) as objects in code.*

*2. A **class** is the blueprint and an **object** is the real thing made from that blueprint.*

*3. Data and the functions that work on that data stay together in one place.*

*4. Python is an object-oriented language, but it also allows procedural programming.*

**Real-Life Analogy**

==>> A **Car class** is like the design drawing of a car.

==>> Every actual car on the road is an **object** made from that design.

==>> All cars have the same structure (colour, brand, speed), but each car has its own values (one is red, one is blue).

#### **Procedural Programming vs OOP**

*Till now we mostly wrote **procedural** code, where data is stored separately and functions are written to work on that data.*

*Let's see the same problem (managing a book) in both styles.*

**Example Without OOP (Procedural Way)**

```python
book1 = {"title": "Python Basics", "author": "Rahul", "is_open": False}
book2 = {"title": "Data Science 101", "author": "Asha", "is_open": False}

def open_book(book):
    book["is_open"] = True
    print(f"'{book['title']}' is now open.")

def book_info(book):
    return f"{book['title']} by {book['author']} | Open: {book['is_open']}"

open_book(book1)
print(book_info(book1))
print(book_info(book2))
```

==>> We have to pass the dictionary to **every** function.

==>> Data (dictionary) and logic (functions) are separate.

==>> If the structure of the dictionary changes, we may need to update many functions.

**Example Using OOP**

```python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author
        self.is_open = False

    def open(self):
        self.is_open = True
        print(f"'{self.title}' is now open.")

    def info(self):
        return f"{self.title} by {self.author} | Open: {self.is_open}"

book1 = Book("Python Basics", "Rahul")
book2 = Book("Data Science 101", "Asha")

book1.open()
print(book1.info())
print(book2.info())
```

==>> Each object keeps its **own data** (title, author, is_open).

==>> Methods work directly on the object's data, so we don't pass the book again and again.

==>> Adding a new attribute (like `publisher`) needs a change in only one place, the class.

#### **Why Do We Use OOP?**

**OOP is used for the following reasons:**

*1. **Modularity:** Code is divided into self-contained classes, so each part is easier to understand and debug.*

*2. **Reusability:** Existing classes can be reused and extended using inheritance.*

*3. **Maintainability:** A change inside one class does not break the whole program.*

*4. **Scalability:** New features can be added without disturbing old code.*

*5. **Real-World Modelling:** Real-life problems map naturally to classes and objects.*

*6. **Industry Standard:** Frameworks like Django, TensorFlow and PyTorch are built using OOP.*

#### **Four Pillars of OOP**

*These four concepts are the base of OOP. Each one has its own notebook later.*

| Pillar | One-line Meaning | Real-Life Example |
|---|---|---|
| **Encapsulation** | Bundling data + methods together and hiding internal data | Medicine capsule hides the powder inside |
| **Abstraction** | Showing only what is needed and hiding complex details | Using a TV remote without knowing the circuit |
| **Inheritance** | A class reusing and extending another class | A child gets features from parents |
| **Polymorphism** | Same method name, different behaviour | A person acts as a student, a son and a friend |

#### **Basic OOP Terms**

| Term | Meaning |
|---|---|
| **Class** | Blueprint or template for creating objects |
| **Object / Instance** | A real item created from a class |
| **Attribute** | A variable that stores the object's data |
| **Method** | A function defined inside a class |
| **self** | Refers to the current object |
| **Constructor** | Special method `__init__` that runs when an object is created |

#### **Everything in Python is an Object**

*Even the basic data types we have been using are objects of some class.*

```python
print(type(10))
print(type("Python"))
print(type([1, 2, 3]))
print(type(print))
```

==>> `10` is an object of class `int`, `"Python"` is an object of class `str`, and so on.

==>> That's why we can call methods like `"python".upper()`, because `upper()` is a method of the `str` class.

```python
name = "python"
print(name.upper())        # upper() is a method of str class

numbers = [3, 1, 2]
numbers.sort()             # sort() is a method of list class
print(numbers)
```

#### **When to Use Procedural and When to Use OOP?**

**Use Procedural Programming when:**

*1. The script is small and simple.*

*2. The task is mainly transforming data step by step.*

*3. You only need a few independent utility functions.*

**Use OOP when:**

*1. The application is large and has many related parts.*

*2. You have many similar "things" with both data and behaviour (users, products, orders).*

*3. You want to reuse code using inheritance.*

*4. The code will be maintained or extended by a team over time.*
