## **Constructors and Destructors in Python**

*Constructors and destructors are **special methods** (also called magic or dunder methods, because of the double underscores).*

*1. **Constructor** → runs when an object is created.*

*2. **Destructor** → runs when an object is about to be destroyed.*

### **1. Constructor (`__init__`)**

*1. `__init__` is the constructor in Python.*

*2. It is called **automatically** right after an object is created.*

*3. Its main job is to **initialize** the object's attributes.*

*4. The first parameter is always `self`, followed by any values needed to set up the object.*

**Syntax**

```python
class ClassName:
    def __init__(self, param1, param2):
        self.attribute1 = param1
        self.attribute2 = param2
```

**Constructor is Called Automatically**

```python
class Demo:
    def __init__(self):
        print("Constructor called!")

d1 = Demo()
d2 = Demo()
```

==>> We never call `__init__` ourselves. Creating the object is enough.

### **Types of Constructors**

*1. **Default Constructor:** No `__init__` is written. Python provides an empty one.*

*2. **Non-Parameterized Constructor:** `__init__` takes only `self`.*

*3. **Parameterized Constructor:** `__init__` takes extra parameters.*

**1. Default Constructor**

```python
class Car:
    pass

c = Car()          # works fine, Python provides a default constructor
print(c)
```

**2. Non-Parameterized Constructor**

```python
class Car:
    def __init__(self):
        self.brand = "Tata"
        self.wheels = 4

c = Car()
print(c.brand, c.wheels)
```

**3. Parameterized Constructor**

```python
class Car:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

c1 = Car("Tata", "Nexon")
c2 = Car("Mahindra", "XUV700")

print(c1.brand, c1.model)
print(c2.brand, c2.model)
```

**What happens if we don't pass the required values?**

```python
try:
    c3 = Car("Honda")
except TypeError as e:
    print("TypeError:", e)
```

#### **Constructor with Default Values**

*We can give default values to parameters, just like normal functions.*

```python
class Student:
    def __init__(self, name, student_id, major="Undeclared"):
        self.name = name
        self.student_id = student_id
        self.major = major
        self.courses = []              # every student starts with an empty list

    def enroll(self, course):
        if course not in self.courses:
            self.courses.append(course)
            print(f"{self.name} enrolled in {course}.")
        else:
            print(f"{self.name} is already enrolled in {course}.")

    def details(self):
        return f"{self.name} | {self.student_id} | {self.major} | {self.courses}"

s1 = Student("Priya", "S101", "Computer Science")
s2 = Student("Aman", "S102")            # uses default major

s1.enroll("Python Basics")
s1.enroll("Python Basics")
print(s1.details())
print(s2.details())
```

#### **Constructor Overloading in Python?**

*1. Python does **not** support multiple constructors like Java or C++.*

*2. If we write `__init__` twice, the **last one replaces** the first one.*

```python
class Sample:
    def __init__(self):
        print("First constructor")

    def __init__(self, a):
        print("Second constructor, a =", a)

s = Sample(10)

# s = Sample()
# TypeError: Sample.__init__() missing 1 required positional argument: 'a'
```

==>> To get the same effect, we use **default values** or `*args`.

```python
class Sample:
    def __init__(self, a=None, b=None):
        if a is None and b is None:
            print("No values given")
        elif b is None:
            print("One value:", a)
        else:
            print("Two values:", a, b)

s1 = Sample()
s2 = Sample(5)
s3 = Sample(5, 10)
```

#### **`__init__` Must Return None**

*The constructor only initializes the object. It should not return any value.*

```python
class Wrong:
    def __init__(self):
        return 10

try:
    w = Wrong()
except TypeError as e:
    print("TypeError:", e)
```

#### **`__new__` vs `__init__` (Behind the Scenes)**

*1. `__new__` actually **creates** the object (allocates memory).*

*2. `__init__` then **initializes** that object.*

*3. We rarely need to write `__new__` ourselves, but it is good to know the order.*

```python
class Person:
    def __new__(cls, *args):
        print("1. __new__ called -> object created")
        return super().__new__(cls)

    def __init__(self, name):
        print("2. __init__ called -> object initialized")
        self.name = name

p = Person("Rahul")
print(p.name)
```

### **2. Destructor (`__del__`)**

*1. `__del__` is called when an object is about to be destroyed.*

*2. An object is destroyed when **no variable refers to it anymore** (its reference count becomes zero).*

*3. It is used for cleanup work like closing files or database connections.*

**Syntax**

```python
class ClassName:
    def __del__(self):
        print("Object is being destroyed")
```

**Example**

```python
class Laptop:
    def __init__(self, brand):
        self.brand = brand
        print(f"{self.brand} laptop created")

    def __del__(self):
        print(f"{self.brand} laptop destroyed")

l1 = Laptop("Dell")
del l1                 # reference removed, so __del__ runs
print("End of cell")
```

**What if two variables refer to the same object?**

```python
l1 = Laptop("HP")
l2 = l1                # both point to the same object

del l1
print("l1 deleted, but object still alive because l2 refers to it")

del l2                 # now no reference left
```

==>> `del` removes the **variable name**, not directly the object.

==>> The destructor runs only when the **last** reference is gone.

**Example: Closing a File in Destructor**

```python
class FileHandler:
    def __init__(self, filename):
        self.filename = filename
        self.file = open(filename, "w")
        print(f"File '{filename}' opened.")

    def write_line(self, line):
        self.file.write(line + "\n")
        print(f"Wrote: {line}")

    def __del__(self):
        if not self.file.closed:
            self.file.close()
            print(f"File '{self.filename}' closed by destructor.")

handler = FileHandler("my_log.txt")
handler.write_line("First log entry")
handler.write_line("Second log entry")
del handler
```

**Important Notes on `__del__`**

*1. It is **not guaranteed** to run immediately. Python's garbage collector decides the exact time.*

*2. For files and connections, the `with` statement (context manager) is a safer choice because it always cleans up, even if an error occurs.*

*3. Use `__del__` mostly for learning, debugging or very special cases.*

```python
# Better way for files
with open("my_log.txt", "a") as f:
    f.write("Third log entry\n")

print("File closed automatically:", f.closed)
```

| Constructor (`__init__`) | Destructor (`__del__`) |
|---|---|
| Runs when an object is created | Runs when an object is destroyed |
| Used to initialize attributes | Used to clean up resources |
| Can take parameters | Takes only `self` |
| Called once per object creation | Called when reference count becomes zero |
