## **Classes and Objects in Python**

#### **What is a Class?**

*1. A class is a **blueprint** or template for creating objects.*

*2. It defines which **attributes** (data) and **methods** (behaviour) the objects will have.*

*3. A class itself does not hold the object's data. It only defines the structure.*

**Analogy:** A cookie cutter is a class. It defines the shape of every cookie.

#### **What is an Object?**

*1. An object (also called an **instance**) is a real thing created from a class.*

*2. Each object has its own values for the attributes.*

*3. One class can create many objects.*

**Analogy:** The cookies are objects. Same shape (class), but each can have different toppings (values).

**Basic Syntax of a Class**

```python
class ClassName:
    # class attributes (optional)
    # methods
    pass        # pass is used when the class body is empty
```

*1. **class:** Keyword used to define a class.*

*2. **ClassName:** Name of the class. By convention it is written in **PascalCase** (`Student`, `BankAccount`).*

*3. **: =>** Starts the class body.*

*4. **pass:** A placeholder so that an empty class does not give an error.*

#### **Creating Objects (Instantiating a Class)**

*We create an object by calling the class name like a function.*

```python
class Dog:
    pass

dog1 = Dog()
dog2 = Dog()

print(dog1)
print(dog2)
print(type(dog1))
```

==>> `dog1` and `dog2` are two **different objects** stored at different memory locations.

==>> Both belong to the same class `Dog`.

```python
print(id(dog1))
print(id(dog2))
print(dog1 == dog2)     # False, because they are two different objects
```

#### **Attributes in a Class**

*Attributes are variables that store the data of an object. There are two types:*

*1. **Instance Attributes:** Belong to one particular object. Each object has its own copy.*

*2. **Class Attributes:** Belong to the class and are **shared** by all objects.*

**1. Instance Attributes**

*Instance attributes are usually created inside the constructor `__init__` using `self`.*

```python
class Student:
    def __init__(self, name, age):
        self.name = name        # instance attribute
        self.age = age          # instance attribute

s1 = Student("Rahul", 21)
s2 = Student("Asha", 22)

print(s1.name, s1.age)
print(s2.name, s2.age)
```

==>> `s1` and `s2` have different values for `name` and `age`.

==>> We access attributes using **dot notation**: `object.attribute`.

**2. Class Attributes**

*Class attributes are defined directly inside the class (outside any method).*

```python
class Student:
    college = "ABC Engineering College"     # class attribute

    def __init__(self, name):
        self.name = name                    # instance attribute

s1 = Student("Rahul")
s2 = Student("Asha")

print(s1.name, "-", s1.college)
print(s2.name, "-", s2.college)
print(Student.college)                      # can also access using class name
```

**What happens if we change a class attribute?**

```python
Student.college = "XYZ Institute"

print(s1.college)
print(s2.college)
# Changed for ALL objects, because the value is shared
```

**What happens if we change it using an object?**

```python
s1.college = "My Own College"

print(s1.college)        # s1 now has its OWN instance attribute
print(s2.college)        # s2 still uses the class attribute
print(Student.college)
```

==>> Assigning `s1.college = ...` does **not** change the class attribute.

==>> It creates a new instance attribute only for `s1`, which hides (shadows) the class attribute.

| Instance Attribute | Class Attribute |
|---|---|
| Defined inside methods using `self` | Defined directly inside the class |
| Each object has its own copy | One copy shared by all objects |
| Accessed using the object | Accessed using the class or the object |
| Example: name, age, roll number | Example: college name, school name |

#### **Methods and the self Keyword**

*1. A method is a function defined inside a class.*

*2. The first parameter of an instance method is `self`.*

*3. `self` refers to the **current object** that called the method.*

*4. Python passes `self` automatically, so we never pass it manually.*

```python
class Dog:
    species = "Canis familiaris"

    def __init__(self, name, breed):
        self.name = name
        self.breed = breed
        self.is_hungry = True

    def bark(self):
        print(f"{self.name} says Woof!")

    def eat(self):
        if self.is_hungry:
            print(f"{self.name} is eating...")
            self.is_hungry = False
        else:
            print(f"{self.name} is not hungry.")

my_dog = Dog("Buddy", "Golden Retriever")
your_dog = Dog("Lucy", "Labrador")

my_dog.bark()
your_dog.eat()
your_dog.eat()          # called again, now not hungry
```

**What does `my_dog.bark()` actually do?**

==>> Python internally converts it to `Dog.bark(my_dog)`.

==>> That is why `self` receives `my_dog`.

```python
Dog.bark(my_dog)        # same as my_dog.bark()
```

**What happens if we forget self?**

```python
class Test:
    def show():                 # self is missing
        print("Hello")

t = Test()

try:
    t.show()
except TypeError as e:
    print("TypeError:", e)
```

==>> Python still passes the object automatically, but the method has no parameter to receive it, so we get a `TypeError`.

#### **Checking Object Attributes using `__dict__`**

*Every object stores its instance attributes in a dictionary called `__dict__`.*

```python
s = Student("Aman")
print(s.__dict__)
print(Student.__dict__["college"])
```

#### **Adding, Modifying and Deleting Attributes**

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

emp = Employee("Ravi", 40000)

emp.salary = 45000          # modify
emp.city = "Jaipur"         # add new attribute (only for this object)
print(emp.__dict__)

del emp.city                # delete
print(emp.__dict__)
```

**Built-in Functions for Attributes**

| Function | Use |
|---|---|
| `getattr(obj, "name")` | Get the value of an attribute |
| `setattr(obj, "name", value)` | Set or create an attribute |
| `hasattr(obj, "name")` | Check if an attribute exists |
| `delattr(obj, "name")` | Delete an attribute |

```python
print(getattr(emp, "name"))
setattr(emp, "department", "IT")
print(hasattr(emp, "department"))
delattr(emp, "department")
print(hasattr(emp, "department"))

# getattr with a default value, useful when attribute may not exist
print(getattr(emp, "bonus", 0))
```
