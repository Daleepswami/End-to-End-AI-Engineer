# **03. The with Statement (Context Manager)**

### **3.1 Introduction to with Statement**

**1. What is the `with` Statement?**

- ***The `with` statement is used to manage resources automatically in Python.***

- ***In file handling, it opens a file and automatically closes it when the block of code is completed, even if an exception occurs.***

**2. Why Use the `with` Statement?**

1. ***Automatically closes the file.***

2. ***Eliminates the need to call `close()` manually.***

3. ***Makes code cleaner and easier to read.***

4. ***Helps prevent resource leaks.***

5. ***Safely manages files when exceptions occur.***

**3. Syntax:**

```python
with open("filename.txt", "mode") as file:
    # File operations
```

````python
with open("data.txt", "w") as file:
    file.write("Hello, Python!")
````

**4. Checking Automatic File Closing**

````python
with open("data.txt", "w") as file:
    file.write("Learning Python File Handling.")

    print(file.closed)

print(file.closed)
````

### **3.2 with open() Syntax and Working in Detail**

**1. Syntax of `with open()`**

- ***The `with open()` statement is used to open a file and manage it automatically.***

**Syntax:**

```python
with open("filename", "mode") as file_object:
    # File operations
```

**2. Components of `with open()`**

| Component | Description |
|-----------|-------------|
| `with` | Starts the context management block. |
| `open()` | Opens the specified file. |
| `filename` | Specifies the file name or path. |
| `mode` | Defines the file operation mode. |
| `as` | Assigns the returned file object to a variable. |
| `file_object` | Used to perform file operations. |
| Indentation | Defines the block in which the file remains open. |

**3. Working of `with open()`**

***The execution follows these steps:***

1. ***Python encounters the `with` statement.***
2. ***`open()` opens the specified file.***
3. ***The returned file object is assigned to the variable after `as`.***
4. ***Python executes the indented block.***
5. ***After the block finishes, Python automatically closes the file.***

**4. Using Different File Modes**

````python
# Read mode
with open("data.txt", "r") as file:
    content = file.read()

# Write mode
with open("data.txt", "w") as file:
    file.write("Hello Python")

# Append mode
with open("data.txt", "a") as file:
    file.write("\nLearning File Handling")
````

**5. Working with Multiple Files**

- ***Python allows multiple files to be opened in a single `with` statement.***

**Syntax:**

```python
with open("file1.txt", "r") as file1, open("file2.txt", "w") as file2:
    # Operations on both files
```

### **3.3 Automatic File Closing and Exception Safety**

**1. Automatic File Closing**

- ***When a file is opened using the `with` statement, Python automatically closes it after the execution of the `with` block.***

- ***There is no need to call the `close()` method manually.***

````python
with open("data.txt", "w") as file:
    file.write("Hello, Python!")

print(file.closed)
````

**2. Exception Safety**

- ***Exception safety means that resources are properly managed even when an error occurs during program execution.***

- ***The `with` statement automatically closes the file even if an exception occurs inside its block.***

````python
with open("data.txt", "w") as file:
    file.write("Hello, Python!")

    result = 10 / 0

print(file.closed)
````

**3. Comparison: Without `with` vs With `with`**

- **Without `with`:**

````python
file = open("data.txt", "w")

try:
    file.write("Hello, Python!")
    result = 10 / 0

finally:
    file.close()
````

- **With `with`:**

````python
with open("data.txt", "w") as file:
    file.write("Hello, Python!")
    result = 10 / 0
````
