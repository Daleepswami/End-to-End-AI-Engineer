# **11. CSV File Handling**

### **11.1 Introduction to CSV Files**

**1. What is a CSV File?**

- ***CSV stands for comma-separated values.***

- ***A `.csv` file is a text file that stores tabular data, with values separated by commas.***

### **11.2 Python `csv` Module**

**1. What is the `csv` Module?**

- ***The `csv` module reads and writes tabular data.***

- ***Import it with `import csv`.***

**2. Tip**

- ***Open CSV files with `newline=""` to avoid blank lines on Windows.***

````python
import csv
````

### **11.3 `csv.reader()`**

**1. What is `csv.reader()`?**

- ***`csv.reader(file_object)` reads a CSV file row by row.***

- ***Each row is returned as a list of strings.***

- ***All values are read as strings, so convert numbers using `int()` or `float()`.***

- ***`next(reader)` can be used to skip or read the header row.***

**2. Syntax:**

```python
reader = csv.reader(file_object)
```

````python
import csv

# First, create a CSV file
with open("staff.csv", "w", newline="") as f:
    f.write("name,dept,salary\nAsha,Tech,72000\nRavi,Sales,54000\n")

with open("staff.csv", newline="") as f:
    reader = csv.reader(f)
    header = next(reader)                      # first row = header
    print("Header:", header)
    for row in reader:
        print(row, "->", row[0], "earns", int(row[2]))
````

### **11.4 `csv.writer()`**

**1. Writing a CSV File**

- ***`csv.writer(f)` creates a writer object, and `writerows()` writes many rows at once.***

````python
import csv

rows = [["name", "dept", "salary"], ["Asha", "Tech", 72000], ["Ravi", "Sales", 54000]]
with open("staff.csv", "w", newline="") as f:
    csv.writer(f).writerows(rows)

with open("staff.csv") as f:
    print(f.read())
````

### **11.5 `csv.DictReader()`**

**1. What is `csv.DictReader()`?**

- ***`csv.DictReader` turns each row into a dictionary.***

````python
import csv

with open("staff.csv", newline="") as f:
    for row in csv.DictReader(f):                  # each row becomes a dict
        print(row["name"], "works in", row["dept"], "earning", row["salary"])
````

**2. Another Example**

````python
# First, create data.csv
with open("data.csv", "w", newline="") as f:
    csv.writer(f).writerows([["name", "score"], ["Alice", 90], ["Bob", 85]])

# CSV read
with open('data.csv') as f:
    reader = csv.DictReader(f)
    for row in reader:
        print(row['name'], row['score'])
````

### **11.6 `csv.DictWriter()`**

**1. What is `csv.DictWriter()`?**

- ***`csv.DictWriter` writes dictionaries as rows of a CSV file.***

- ***The `fieldnames` argument gives the column names (the keys of the dictionaries).***

**2. Main Methods:**

- ***`writeheader()` writes the column names as the first row.***

- ***`writerow(dict)` writes one dictionary as a row.***

- ***`writerows(list_of_dicts)` writes many dictionaries at once.***

**3. Syntax:**

```python
writer = csv.DictWriter(file_object, fieldnames=["col1", "col2"])
```

````python
import csv

students = [
    {"name": "Asha", "score": 91},
    {"name": "Ravi", "score": 78},
    {"name": "Meena", "score": 85},
]

with open("students.csv", "w", newline="") as f:
    writer = csv.DictWriter(f, fieldnames=["name", "score"])
    writer.writeheader()                       # column names
    writer.writerows(students)                 # all rows at once

with open("students.csv", newline="") as f:
    for row in csv.DictReader(f):
        print(row["name"], row["score"])
````
