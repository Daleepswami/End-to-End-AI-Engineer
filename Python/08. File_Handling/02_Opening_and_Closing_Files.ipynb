# **02. Opening and Closing Files**

### **2.1 The open() Function**

1. ***The `open()` function is a built-in Python function used to open a file for reading, writing, appending, or other file operations.***

2. ***It returns a file object (also called a file handle), which provides methods to interact with the file.***

**Syntax:**

```python
file_object = open(file, mode)
```

### **2.2 File Opening Modes in Python**

**1. What are File Opening Modes?**

- ***File opening modes specify how a file should be opened and what operations can be performed on it, such as reading, writing, or appending data.***

- ***Modes are passed as the second argument to the `open()` function.***

**2. Types of File Opening Modes**

***Python provides several file opening modes:***

| Mode | Description |
|------|-------------|
| `"r"` | Opens a file for reading. File must exist. |
| `"w"` | Opens a file for writing. Creates a new file or overwrites existing content. |
| `"a"` | Opens a file for appending. Adds data at the end. Creates the file if it does not exist. |
| `"x"` | Creates a new file. Raises an error if the file already exists. |
| `"t"` | Opens a file in text mode. This is the default mode. |
| `"b"` | Opens a file in binary mode. |
| `"r+"` | Opens a file for both reading and writing. File must exist. |
| `"w+"` | Opens a file for reading and writing. Overwrites existing content or creates a new file. |
| `"a+"` | Opens a file for reading and appending. Creates the file if it does not exist. |

**A. Read Mode (`"r"`)**

- ***Opens an existing file for reading.***
- ***Does not allow writing.***
- ***Raises `FileNotFoundError` if the file does not exist.***

````python
file = open("data.txt", "r")
content = file.read()
print(content)
file.close()
````

**B. Write Mode (`"w"`)**

- ***Opens a file for writing.***
- ***Creates the file if it does not exist.***
- ***Deletes the existing content before writing.***

````python
file = open("data.txt", "w")
file.write("Hello, Python!")
file.close()
````

**C. Append Mode (`"a"`)**

- ***Opens a file for adding new content.***
- ***Adds data at the end of the file.***
- ***Preserves existing content.***
- ***Creates the file if it does not exist.***

````python
file = open("data.txt", "a")
file.write("\nLearning File Handling")
file.close()
````

**D. Exclusive Creation Mode (`"x"`)**

- ***Creates a new file.***
- ***Raises `FileExistsError` if the file already exists.***
- ***Prevents accidentally overwriting an existing file.***

````python
file = open("new_data.txt", "x")
file.write("New file created.")
file.close()
````

**E. Text and Binary Modes (`"t"` and `"b"`)**

| Mode | Description | Data Type |
|------|-------------|-----------|
| `"rt"` | Read text file | `str` |
| `"rb"` | Read binary file | `bytes` |
| `"wt"` | Write text file | `str` |
| `"wb"` | Write binary file | `bytes` |

````python
# Text mode
file = open("data.txt", "rt")
content = file.read()
file.close()

# Binary mode
file = open("image.png", "rb")
content = file.read()
file.close()
````

**F. Read and Write Modes (`+`)**

***The `+` symbol allows both reading and writing.***

| Mode | Reading | Writing | Existing Content |
|------|---------|---------|------------------|
| `"r+"` | Yes | Yes | Preserved initially |
| `"w+"` | Yes | Yes | Erased |
| `"a+"` | Yes | Yes | Preserved; writing occurs at the end |

### **2.3 close() Method in Python**

**What is the `close()` Method?**

- ***The `close()` method is used to close an opened file after completing file operations.***

- ***It releases the resources associated with the file and ensures that pending data is properly written.***

**Syntax:**

```python
file.close()
```

**Why is `close()` Needed?**

1. **Resource Management:** Releases resources used by the file.
2. **Save Changes:** Ensures buffered data is written to the file.
3. **Prevent Data Loss:** Helps ensure pending write operations are completed.
4. **File Safety:** Prevents accidental operations on a file after closing it.
5. **System Efficiency:** Avoids keeping unnecessary files open.

**Example of `close()`**

````python
file = open("data.txt", "w")

file.write("Hello, Python!")

file.close()
````

**Checking Whether a File is Closed**

- ***Python provides the `closed` attribute to check the file's status.***

````python
file = open("data.txt", "r")

print(file.closed)

file.close()

print(file.closed)
````

**What Happens If We Use a Closed File?**

- ***Performing file operations after closing a file raises a `ValueError`.***

````python
file = open("data.txt", "w")
file.close()

# file.write("Hello")

# Output:
# ValueError: I/O operation on closed file.
````
