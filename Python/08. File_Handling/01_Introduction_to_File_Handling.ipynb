# **01. Introduction to File Handling**

### **1.1 What is File Handling?**

1. ***File Handling in Python is the process of creating, opening, reading, writing, updating, and managing files stored on a computer.***

2. ***It allows Python programs to store and retrieve data from files instead of keeping all data temporarily in memory.***

**Key Points:**

1. ***File handling allows programs to interact with files stored on a computer.***

2. ***It is used to read existing data from files.***

3. ***It allows programs to write new data into files.***

4. ***It supports creating, updating, and deleting files.***

5. ***It helps maintain data even after the program terminates.***

### **1.2 Why File Handling is Needed**

1. ***File handling is needed to store, retrieve, and manage data permanently so that information can be accessed even after a program has finished executing.***

***Main Reasons for Using File Handling:***

1. **Permanent Data Storage:** Stores data permanently on storage devices.

2. **Data Retrieval:** Allows previously stored data to be accessed whenever required.

3. **Data Modification:** Allows existing data to be updated or new data to be added.

4. **Data Sharing:** Enables different programs to share and access data.

5. **Handling Large Amounts of Data:** Stores and processes large datasets efficiently.

6. **Record Keeping:** Maintains records such as student details, employee information, transactions, and application logs.

### **1.3 Data Persistence**

1. ***Data Persistence is the ability to store data so that it remains available even after a program terminates or the computer is restarted.***

**Key Points:**

1. ***Variables generally store data temporarily in RAM.***

2. ***Data stored in RAM is lost when the program terminates, unless saved elsewhere.***

3. ***Files allow data to be stored on permanent storage devices such as HDDs and SSDs.***

4. ***Persisted data can be retrieved and reused whenever required.***

5. ***File handling is one of the common ways to achieve data persistence.***

### **1.4 File I/O (Input/Output)**

**What is File I/O?**

1. ***File I/O (Input/Output) refers to the process of reading data from files and writing data into files using a Python program.***

**File I/O consists of two main operations:**

1. **Input (Reading)**

    - ***Input means retrieving data from an existing file.***
    - ***Python reads the stored information and makes it available for processing.***

2. **Output (Writing)**

    - ***Output means storing data into a file.***
    - ***Python writes new information into a file for future use.***

```python
# Output (Writing)
with open("data.txt", "w") as file:
    file.write("Hello, Python!")

# Input (Reading)
with open("data.txt", "r") as file:
    content = file.read()

print(content)
```

### **1.5 Text Files vs Binary Files**

**What are File Types?**

- ***Files are generally classified into two main categories based on how data is stored and represented:***

1. Text Files

2. Binary Files

**1. Text Files**

- ***Text files store data as characters using a character encoding such as UTF-8. Their contents can generally be read and understood using a text editor.***

**Key Points:**

- ***Store data in a human-readable text format.***
- ***Use character encoding such as UTF-8.***
- ***Can be opened and edited using text editors.***
- ***Commonly used for storing text, records, and structured text data.***

**Examples:**

- `.txt` – Plain text files
- `.csv` – Comma-separated values
- `.py` – Python source code
- `.html` – HTML documents
- `.json` – JSON data

**2. Binary Files**

- ***Binary files store data as bytes in a format defined by a particular file format or application. Their contents are not generally human-readable when opened directly in a text editor.***

***Key Points:***

- ***Store data as bytes.***
- ***Usually require specific applications or software to interpret their contents.***
- ***Commonly used for images, audio, videos, documents, and executable files.***
- ***Python handles binary files using binary modes such as `rb` and `wb`.***

**Examples:**

- `.jpg` / `.png` – Image files
- `.mp3` – Audio files
- `.mp4` – Video files
- `.pdf` – PDF documents
- `.exe` – Executable files
- `.docx` – Word documents
