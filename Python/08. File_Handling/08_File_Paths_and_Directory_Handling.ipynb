# **08. File Paths and Directory Handling**

### **8.1 Introduction to File Paths**

**1. What is a File Path?**

- ***The `filename` given to `open()` is a string representing the path to the file you want to open (for example, `"my_data.txt"` or `"data/report.csv"`).***

**2. Building Paths Safely**

- ***`os.path.join(a, b)` joins paths safely, because it handles the operating system separators.***

```python
import os

# Safe cross-platform path building
full = os.path.join('folder', 'sub', 'file.txt')
print(full)
# Windows: folder\sub\file.txt / Linux: folder/sub/file.txt
```

### **8.2 Relative Paths**

**1. What is a Relative Path?**

- ***A relative path describes the location of a file starting from the current working directory (the folder from which the program is running).***

- ***It does not contain the full location from the root of the file system.***

- ***If the current working directory changes, the same relative path points to a different place.***

**2. Examples of Relative Paths**

| Relative Path | Meaning |
|---------------|---------|
| `"my_file.txt"` | A file in the current directory |
| `"data/report.csv"` | A file inside the `data` folder of the current directory |
| `"./notes.txt"` | A file in the current directory (`.` means the current folder) |
| `"../notes.txt"` | A file in the parent folder (`..` means one level up) |

```python
# Creates my_file.txt in the same directory as the script
file_object = open("File/my_file.txt", "w")
file_object.close()
print("my_file.txt created in the current directory.")
```

**3. Using a Relative Path with a Folder**

```python
import os

os.makedirs("data", exist_ok=True)

with open("data/report.txt", "w") as f:          # relative path with a folder
    f.write("Relative path example")

print("Current directory:", os.getcwd())
print("Full location    :", os.path.abspath("data/report.txt"))
```

### **8.3 Absolute Paths**

**1. What is an Absolute Path?**

- ***An absolute path is the complete location of a file, starting from the root of the file system.***

- ***It does not depend on the current working directory.***

- ***On Linux and macOS, it starts with `/` (for example, `/home/user/documents/notes.txt`).***

- ***On Windows, it starts with a drive letter (for example, `C:\Users\user\documents\notes.txt`).***

```python
import os

path = '/home/user/documents/notes.txt'
print(os.path.basename(path))   # notes.txt
print(os.path.dirname(path))    # /home/user/documents
print(os.path.splitext(path))   # ('/home/user/documents/notes', '.txt')

print(os.path.abspath("my_file.txt"))   # absolute path of a file in the current directory
```

**2. Checking and Converting Paths**

- ***`os.path.abspath(p)` converts a relative path into an absolute path.***

- ***`os.path.isabs(p)` returns `True` if the path is absolute.***

```python
print(os.path.isabs("my_file.txt"))                    # False -> relative
print(os.path.isabs(os.path.abspath("my_file.txt")))   # True  -> absolute
```

**3. Relative Path vs Absolute Path**

| Feature | Relative Path | Absolute Path |
|---------|---------------|---------------|
| Starts from | Current working directory | Root of the file system |
| Example | `data/report.csv` | `/home/user/data/report.csv` |
| Depends on current directory | Yes | No |
| Length | Short | Long |
| Moving the project to another computer | Usually keeps working | Usually stops working |

### **8.4 Current Working Directory (`os.getcwd()`)**

**1. What is `os.getcwd()`?**

- ***`os.getcwd()` returns the current working directory.***

- ***`os.chdir(path)` changes the directory.***

- ***`os.listdir('.')` lists the files in the current directory.***

```python
import os

print(os.getcwd())          # Current working directory
# os.chdir('/path/to/folder')   # Change directory (replace with a real path)
print(os.listdir('.'))      # List files in current dir
```

### **8.5 `os` Module**

**1. What is the `os` Module?**

- ***The `os` module provides a way to interact with the operating system: navigate directories, manage files and folders, read environment variables, etc.***

**2. `os.path` — Path Utilities**

| Function | Description |
|----------|-------------|
| `os.path.join(a, b)` | Join paths safely (handles OS separators) |
| `os.path.exists(p)` | Check if path exists |
| `os.path.isfile(p)` | True if p is a file |
| `os.path.isdir(p)` | True if p is a directory |
| `os.path.basename(p)` | File name from full path |
| `os.path.dirname(p)` | Directory portion of path |
| `os.path.splitext(p)` | Split name and extension: `('file', '.txt')` |
| `os.path.abspath(p)` | Absolute path |

**3. Example: `os` and `shutil`**

```python
import os, shutil

os.makedirs("project/data/raw", exist_ok=True)         # create nested folders
for name in ["a.txt", "b.txt", "c.csv"]:
    with open(os.path.join("project/data/raw", name), "w") as f:
        f.write("sample")

print(sorted(os.listdir("project/data/raw")))
os.rename("project/data/raw/a.txt", "project/data/raw/alpha.txt")
shutil.copy("project/data/raw/b.txt", "project/data/b_copy.txt")
os.remove("project/data/raw/c.csv")

for folder, subfolders, files in os.walk("project"):
    print(folder, sorted(files))

copy_path = "project/data/b_copy.txt"
print(os.path.exists(copy_path), os.path.getsize(copy_path))
shutil.rmtree("project")                                # delete the folder tree
print(os.path.exists("project"))
```

### **8.6 `pathlib` Module**

**1. What is `pathlib`?**

- ***`pathlib` is the modern way to work with paths and files.***

- ***The `/` operator of `pathlib` builds paths safely across operating systems.***

```python
from pathlib import Path

base = Path("reports")
(base / "2026").mkdir(parents=True, exist_ok=True)
(base / "2026" / "jan.txt").write_text("January sales")
(base / "2026" / "feb.txt").write_text("February sales")

for path in sorted(base.rglob("*.txt")):
    print(path.name, path.stem, path.suffix, path.read_text())
```

**2. `os` / `shutil` vs `pathlib`**

| Task | `os` / `shutil` | `pathlib` |
|------|-----------------|-----------|
| Current folder | `os.getcwd()` | `Path.cwd()` |
| Create folders | `os.makedirs(p, exist_ok=True)` | `Path(p).mkdir(parents=True, exist_ok=True)` |
| List contents | `os.listdir(p)` | `Path(p).iterdir()` |
| Join paths | `os.path.join(a, b)` | `Path(a) / b` |
| Exists? | `os.path.exists(p)` | `Path(p).exists()` |
| Delete file | `os.remove(p)` | `Path(p).unlink()` |
| Delete folder tree | `shutil.rmtree(p)` | — |
| Copy / move | `shutil.copy` / `shutil.move` | — |

### **8.7 Creating and Checking Directories**

**1. Creating Directories**

- ***`os.mkdir(name)` creates a single directory.***

- ***`os.makedirs(path)` creates nested directories.***

- ***`os.rmdir(name)` removes an empty directory.***

**2. Checking Directories**

- ***`os.path.exists(p)` checks whether a path exists.***

- ***`os.path.isdir(p)` checks whether it is a directory, and `os.path.isfile(p)` checks whether it is a file.***

**3. Tip**

- ***Use `exist_ok=True` to avoid errors when folders already exist.***

```python
import os

os.mkdir('new_folder')                       # Create single directory
os.makedirs('File/a/b/c', exist_ok=True)          # Create nested directories

print(os.path.exists('new_folder'))          # True
print(os.path.isdir('new_folder'))           # True
print(os.path.isfile('new_folder'))          # False

os.rmdir('new_folder')                       # Remove empty directory
print(os.path.exists('new_folder'))          # False
```
