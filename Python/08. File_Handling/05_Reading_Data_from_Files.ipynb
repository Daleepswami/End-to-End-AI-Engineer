# **05. Reading Data from Files**

### **5.1 Introduction to Reading Data from Files**

**1. What is File Reading?**

***File reading is the process of retrieving data stored in a file and accessing it through a Python program.***

***Python provides different methods to read file contents, such as:***

- `read()`
- `readline()`
- `readlines()`

**2. Why is File Reading Needed?**

1. ***To retrieve previously stored data.***
2. ***To access saved information.***
3. ***To process data for analysis.***
4. ***To display file contents.***
5. ***To reuse stored data in programs.***

**3. Basic Syntax**

````python
with open("filename.txt", "r") as file:
    content = file.read()
    print(content)
````

**4. Example**

***First, create a file and write some data:***

````python
with open("data.txt", "w") as file:
    file.write("Hello, Python!\n")
    file.write("Learning File Handling.")
````

***Now, read the file:***

````python
with open("data.txt", "r") as file:
    content = file.read()
    print(content)
````

### **5.2 read() Method in Python**

**1. What is the `read()` Method?**

- ***The `read()` method is used to read data from a file.***

- ***By default, it reads the entire remaining content of the file and returns it as a string in text mode.***

**2. Syntax:**

```python
file.read(size)
```

**3. Reading the Entire File**

````python
with open("data.txt", "w") as file:
    file.write("Hello, Python!\n")
    file.write("Learning File Handling.")

with open("data.txt", "r") as file:
    content = file.read()
    print(content)
````

**4. Reading a Specific Number of Characters**

- ***The `read(size)` method can read a specified number of characters.***

````python
with open("data.txt", "r") as file:
    content = file.read(5)
    print(content)
````

### **5.3 readline() Method in Python**

**What is the `readline()` Method?**

- ***The `readline()` method is used to read one line at a time from a file.***

- ***It reads characters from the current file position up to and including the next newline character (`\n`), or until the end of the file.***

**Syntax:**

```python
file.readline(size)
```

**Example:**

````python
with open("data.txt", "w") as file:
    file.write("Hello, Python!\n")
    file.write("Learning File Handling.\n")
    file.write("Deep Learning\n")
    file.write("Data science.")

with open("data.txt", "r") as file:
    content = file.readline()
    print(content)
````

**Reading Multiple Lines**

***We can call `readline()` multiple times to read subsequent lines.***

````python
with open("data.txt", "r") as file:
    print(file.readline())
    print(file.readline())
    print(file.readline())
````

### **5.4 readlines() Method in Python**

**1. What is the `readlines()` Method?**

- ***The `readlines()` method is used to read all remaining lines from a file and return them as a list of strings.***

- ***Each string generally represents one line, including its newline character (`\n`) if present.***

**Syntax:**

```python
file.readlines(hint)
```

````python
with open("data.txt", "r") as file:
    lines = file.readlines()
    print(lines)
````

**Accessing Individual Lines**

***Since `readlines()` returns a list, we can access individual lines using list indexing.***

````python
with open("data.txt", "r") as file:
    lines = file.readlines()

    print(lines[0])
    print(lines[1])
    print(lines[2])
````

**Reading Lines Using a Loop**

***We can iterate over the list returned by `readlines()`.***

````python
with open("data.txt", "r") as file:
    lines = file.readlines()

    for line in lines:
        print(line, end="")
````

**Return Value**

***The `readlines()` method returns a list of strings.***

````python
with open("data.txt", "r") as file:
    result = file.readlines()

    print(type(result))
````

**Comparison: `read()`, `readline()` and `readlines()`**

| Method | Reads | Return Type |
|---|---|---|
| `read()` | Entire remaining content by default | `str` |
| `readline()` | One line | `str` |
| `readlines()` | All remaining lines | `list` |

### **5.5 Reading Files Using Loops (Line-by-Line Reading)**

**1. Introduction**

- ***Python allows us to read a file line by line using a `for` loop.***

- ***Instead of loading the entire file into memory, a loop reads one line at a time. This approach is useful when working with large files.***

**2. Syntax:**

```python
with open("filename.txt", "r") as file:
    for line in file:
        print(line)
```

**Example: Reading a File Line by Line**

````python
lines = [
    "Python is easy to learn.",
    "Python is a programming language.",
    "Python is widely used in Data Science."
]

with open("python_notes.txt", "w") as file:
    for line in lines:
        file.write(line + "\n")

print("File successfully written!")
````

````python
# read
with open("python_notes.txt", "r") as file:
    for line in file:
        print(line, end="")
````

### **5.6 File Cursor in Python**

**1. What is a File Cursor?**

- ***A file cursor is a pointer that represents the current position inside a file during reading or writing operations.***

- ***Python uses the file cursor to keep track of where the next read or write operation will take place.***

- ***When the file is opened for reading, the cursor initially starts at position 0 (the beginning of the file).***

**How Does a File Cursor Work?**

- ***The cursor moves automatically when you perform reading or writing operations.***

````python
with open("data.txt", "r") as file:
    content = file.read(6)
    print(content)
````

````python
# Create a file
with open("data.txt", "w") as file:
    file.write("Python Programming")

# Read data using cursor
with open("data.txt", "r") as file:
    print("First Reading:", file.read(6))
    print("Second Reading:", file.read(5))
    print("Third Reading:", file.read())
````

### **5.7 tell() Method in Python**

**What is the `tell()` Method?**

- ***The `tell()` method is used to find the current position of the file cursor.***

- ***It returns the current cursor position as an integer.***

**Definition:**

- ***The `tell()` method returns the index of the next character or byte that will be read or written in a file.***

**Syntax:**

```python
file.tell()
```

**Checking the Initial Cursor Position**

````python
with open("data.txt", "w") as file:
    file.write("Python Programming")

with open("data.txt", "r") as file:
    print(file.tell())
````

**Checking Cursor Position After Reading**

````python
with open("data.txt", "r") as file:
    print(file.tell())

    print(file.read(6))

    print(file.tell())

    print(file.read(5))

    print(file.tell())
````

````python
with open("sample.txt", "w") as file:
    file.write("Hello Python Programming")

with open("sample.txt", "r") as file:
    print("Initial Position:", file.tell())

    print("First Reading:", file.read(5))
    print("Current Position:", file.tell())

    print("Second Reading:", file.read(6))
    print("Current Position:", file.tell())

    print("Remaining Data:", file.read())
    print("Final Position:", file.tell())
````

### **5.8 seek() Method in Python**

**What is the `seek()` Method?**

- ***The `seek()` method is used to move the file cursor to a specific position in a file.***

- ***It allows us to control where the next read or write operation will take place.***

**Definition:**

- ***The `seek()` method changes the current position of the file cursor.***

**Syntax:**

```python
file.seek(offset, whence)
```

- ***`offset` – Number of bytes or position units to move, depending on the reference point.***
- ***`whence` – Specifies the reference point from which the cursor moves.***

**Values of `whence`**

| Value | Constant | Meaning |
|---|---|---|
| 0 | `os.SEEK_SET` | Beginning of the file |
| 1 | `os.SEEK_CUR` | Current cursor position |
| 2 | `os.SEEK_END` | End of the file |

**Moving Cursor to the Beginning**

````python
with open("data.txt", "w") as file:
    file.write("Python Programming")

with open("data.txt", "r") as file:
    print(file.read(6))

    file.seek(0)

    print(file.read())
````

**Using `seek()` with `whence`**

````python
with open("data.txt", "rb") as file:

    file.seek(7, 0)
    print(file.read())

    file.seek(2, 1)
    print(file.read())

    file.seek(-5, 2)
    print(file.read())
````
