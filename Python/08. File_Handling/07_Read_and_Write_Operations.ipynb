# **07. Read and Write Operations**

### **7.1 Introduction to Read and Write Modes**

**1. What are Read and Write Modes?**

- ***The `+` symbol allows a file to be opened for both reading and writing.***

- ***Python provides three such modes: `r+`, `w+` and `a+`.***

| Mode | Meaning | If File Exists | If File Does Not Exist | Cursor Position |
|------|---------|----------------|------------------------|-----------------|
| `"r+"` | Read and write; no truncation | Reads from start | Raises `FileNotFoundError` | Beginning |
| `"w+"` | Read and write (truncate); similar to `"w"`, but allows reading too | Truncates | Creates new | Beginning |
| `"a+"` | Read and append; writing always happens at the end | Appends | Creates new | End |

### **7.2 `r+` Mode**

**1. What is `r+` Mode?**

- ***The `r+` mode opens a file for both reading and writing.***

- ***The file pointer (cursor) is placed at the beginning.***

- ***This means if you write immediately, you will overwrite the beginning of the file.***

- ***To add data, you often need to read to the end or use `seek()` to move the cursor.***

- ***This mode can be tricky for beginners because writing happens at the current cursor position.***

**2. Example**

- ***`r+` needs an existing file, so create it first.***

```python
with open("File/sample.txt", "w") as f:
    f.write("")
```

```python
# Using r+ carefully
file_rplus = open("File/sample.txt", "r+")   # r+ needs an existing file (created above)
file_rplus.write("Original line 1\nOriginal line 2\n")

file_rplus.seek(0)                      # Move cursor back to beginning to read
print("Content after initial write:")
print(file_rplus.read())

# After read(), the cursor is at the end, so this line is added at the end
file_rplus.write("NEW LINE ADDED at end!\n")

file_rplus.seek(0)                      # Move cursor back to beginning to read
print("\nContent after writing in r+ mode:")
print(file_rplus.read())
file_rplus.close()
```

**3. Writing Immediately Overwrites the Beginning**

```python
with open("File/sample.txt", "r+") as f:
    f.write("NEW")          # cursor is at the beginning, so this overwrites existing data
    f.seek(0)
    print(f.read())
```

### **7.3 `w+` Mode**

**1. What is `w+` Mode?**

- ***The `w+` mode opens a file for both reading and writing.***

- ***If the file exists, it is truncated (existing content is erased).***

- ***If the file does not exist, a new one is created.***

- ***It is similar to `"w"`, but allows reading too.***

```python
with open("File/wplus.txt", "w+") as f:
    f.write("Hello, w+ mode!")
    f.seek(0)               # move to the beginning to read what was written
    print(f.read())
```

### **7.4 `a+` Mode**

**1. What is `a+` Mode?**

- ***The `a+` mode opens a file for both reading and appending.***

- ***If the file exists, new data is appended.***

- ***If the file does not exist, a new one is created.***

- ***Writing always happens at the end of the file.***

- ***If you want to read and then add to the end, `a+` is often safer than `r+`.***

- ***To read the existing content, move the cursor to the beginning using `seek(0)`.***

```python
with open("File/aplus.txt", "w") as f:
    f.write("Line 1\n")

with open("File/aplus.txt", "a+") as f:
    f.write("Line 2\n")     # always written at the end
    f.seek(0)               # move to the beginning to read
    print(f.read())
```

### **7.5 Reading and Writing in the Same File**

**1. Common Pattern**

- ***For beginners, it is generally clearer to separate reading and writing operations, or to use `a+` if you want to read from the beginning but always write at the end.***

- ***`r+` is powerful but requires careful management of the file cursor.***

**2. Better Pattern: Read All, Then Write All Modified Content**

```python
try:
    with open("File/data_to_modify.txt", "w") as f:        # Create initial content
        f.write("Item A\nItem B\n")

    # Read all content
    with open("File/data_to_modify.txt", "r") as f:
        existing_content = f.read()
        print("Existing content:\n", existing_content)

    # Modify/Add to content
    new_content = existing_content + "Item C (added after read)\n"

    # Write all content back (this overwrites the original file)
    with open("File/data_to_modify.txt", "w") as f:
        f.write(new_content)

    print("\nContent after modification and re-write:")
    with open("file/data_to_modify.txt", "r") as f:
        print(f.read())

except FileNotFoundError:
    print("File not found for modification.")
```

### **7.6 File Cursor Management**

**1. Moving the Cursor**

- ***`file_object.seek(0)` moves the cursor back to the beginning of the file.***

- ***`file_object.seek(0, 2)` moves the cursor to the end of the file.***

- ***`file_object.tell()` returns the current cursor position.***

- ***Writing happens at the current cursor position, so manage the cursor carefully in `r+` mode.***

```python
with open("File/sample.txt", "r+") as f:
    f.seek(0, 2)                     # move to the end of the file
    print("End position:", f.tell())
    f.write("Added at the end\n")

    f.seek(0)                        # move back to the beginning
    print(f.read())
```
