# **06. Appending Data to Files**

### **6.1 Introduction to Appending Data**

**1. What is Appending?**

- ***When you want to add new content to an existing file without deleting its original content, you open it in append mode (`"a"`).***

- ***If the file does not exist, it will be created.***

**2. Why Use Append Mode?**

1. ***Adds new data to the end of a file.***
2. ***Keeps the existing content safe.***
3. ***Useful for log files, journals, and records that keep growing.***

### **6.2 Append Mode (`a`)**

**1. What is Append Mode?**

- ***Append mode (`"a"`) opens a file for writing.***

- ***If the file exists, new data is written to the end of the file.***

- ***If the file does not exist, a new one is created.***

- ***The cursor starts at the end of the file.***

| Mode | Meaning | If File Exists | If File Does Not Exist | Cursor Position |
|------|---------|----------------|------------------------|-----------------|
| `"a"` | Append: opens the file for writing; new data is written at the end | Appends | Creates new | End |

### **6.3 Writing Data Using Append Mode**

**1. Basic Example**

- ***A file is first created in write mode, then new content is added using append mode.***

```python
f = open("notes.txt", "w")
f.write("Python file handling\n")
f.close()                                # always close what you open

with open("notes.txt", "a") as f:        # preferred way
    f.write("The with statement closes the file automatically\n")
print("Closed?", f.closed)
```

**2. Reading the File to Verify**

```python
with open("notes.txt", "r") as f:
    print(f.read())
```

**3. Writing Another Line (Module 4 Example)**

```python
# Write (overwrites file if it exists)
with open("output.txt", "w") as f:
    f.write("Hello, File!\n")
    f.writelines(["Line 1\n", "Line 2\n"])

# Append (adds to end)
with open("output.txt", "a") as f:
    f.write("New line added\n")

with open("output.txt", "r") as f:
    print(f.read())
```

**4. Example: Adding Log Entries**

- ***Each time the script runs, new lines are added to the end of the log file.***

```python
import datetime  # To get current timestamp

def log_message(message):
    timestamp = datetime.datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    log_entry = f"{timestamp} - {message}\n"

    # Open in 'a' (append) mode. Cursor starts at the end.
    log_file = open("application.log", "a")
    log_file.write(log_entry)
    log_file.close()
    print(f"Logged: {log_entry.strip()}")

print("--- Starting Application Logging ---")
log_message("Application started successfully.")
log_message("User 'Alice' logged in.")
log_message("Processing data...")
log_message("Application shutting down.")
print("--- Logging Finished ---")
```

**Note:** Timestamps will vary based on the current time. After running, check `application.log`. Each time you run the script, new lines are added to the end of the file.

### **6.4 Difference Between Write and Append**

| Feature | Write Mode (`"w"`) | Append Mode (`"a"`) |
|---------|--------------------|---------------------|
| Existing content | Erased (truncated) | Preserved |
| File does not exist | Creates a new file | Creates a new file |
| Cursor position | Beginning | End |
| Use | Start with fresh content | Add new content to existing data |

**Key Takeaway:**

- ***`"w"` overwrites the file; use `"a"` to add to it.***

```python
# Write mode: second write erases the first
with open("demo_w.txt", "w") as f:
    f.write("First\n")
with open("demo_w.txt", "w") as f:
    f.write("Second\n")
with open("demo_w.txt") as f:
    print("After two writes:", f.read())

# Append mode: second write is added after the first
with open("demo_a.txt", "w") as f:
    f.write("First\n")
with open("demo_a.txt", "a") as f:
    f.write("Second\n")
with open("demo_a.txt") as f:
    print("After write + append:", f.read())
```
