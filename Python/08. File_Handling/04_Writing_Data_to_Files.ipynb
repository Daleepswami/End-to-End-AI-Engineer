# **04. Writing Data to Files**

### **4.1 Introduction to Writing Data in Files**

**1. What is File Writing?**

- ***File writing is the process of storing data from a Python program into a file.***

- ***Python provides different methods to write data into files, such as:***

1. `write()`
2. `writelines()`

**2. Why is File Writing Needed?**

1. ***To store information permanently.***
2. ***To save program output.***
3. ***To maintain records and reports.***
4. ***To store user-generated data.***
5. ***To save data for future use.***

**3. Basic Syntax**

````python
with open("filename.txt", "w") as file:
    file.write("data")
````

### **4.2 write() Method in Python**

**1. What is the `write()` Method?**

- ***The `write()` method is used to write a string of data into a file.***

- ***It returns the number of characters written to the file.***

**2. Syntax:**

```python
file.write(string)
```

**3. Basic Example**

````python
with open("data.txt", "w") as file:
    file.write("Hello, Python!")
````

**4. Writing Multiple Strings**

- ***We can call `write()` multiple times to write different strings into the same file.***

````python
with open("student.txt", "w") as file:
    file.write("Name: Daleep\n")
    file.write("Age: 22\n")
    file.write("Course: Python")
````

**5. Return Value of `write()`**

- ***The `write()` method returns the number of characters written.***

````python
with open("data.txt", "w") as file:
    result = file.write("Python")

    print(result)
````

**6. Writing Different Data Types**

- ***The `write()` method accepts strings, not integers, floats, or other non-string values directly.***

````python
with open("data.txt", "w") as file:
    file.write(str(100))
````

### **4.3 writelines() Method in Python**

**1. What is the `writelines()` Method?**

- ***The `writelines()` method is used to write multiple strings into a file at once.***

- ***It accepts an iterable of strings, such as a list or tuple, and writes each string into the file.***

**2. Syntax:**

```python
file.writelines(iterable)
```

**3. Basic Example**

````python
lines = ["Hello\n", "Python\n", "File Handling"]

with open("data.txt", "w") as file:
    file.writelines(lines)
````

**4. Using `writelines()` Without Newline Characters**

- ***The `writelines()` method does not automatically insert newline characters between strings.***

````python
lines = ["Hello", "Python", "File Handling"]

with open("data.txt", "w") as file:
    file.writelines(lines)
````

### **4.4 Writing Multiple Lines and Newline Character (`\n`)**

**1. Writing Multiple Lines in a File**

- ***Python allows us to write multiple lines of text into a file using `write()` or `writelines()`.***

- ***To separate lines, we use the newline character (`\n`).***

**2. What is the Newline Character (`\n`)?**

***The newline character (`\n`) is an escape sequence used to move text to the next line.***

````python
# Example:
with open("data.txt", "w") as file:
    file.write("Hello\nPython\nFile Handling")
````

**3. Writing Multiple Lines Using `write()`**

***We can use multiple `write()` calls to write different lines.***

````python
with open("data.txt", "w") as file:
    file.write("Line 1\n")
    file.write("Line 2\n")
    file.write("Line 3\n")
````

**4. Writing Multiple Lines Using `writelines()`**

***We can also use `writelines()` with a list of strings.***

````python
lines = [
    "Line 1\n",
    "Line 2\n",
    "Line 3\n"
]

with open("data.txt", "w") as file:
    file.writelines(lines)
````

**`write()` vs `writelines()`**

| Feature | `write()` | `writelines()` |
|---------|-----------|----------------|
| Purpose | Writes a single string per call. | Writes an iterable of strings. |
| Input | Accepts a string. | Accepts a list, tuple, or another iterable of strings. |
| Newline | Does not add newline automatically. | Does not add newline automatically. |
| Return value | Returns the number of characters written. | Returns `None`. |
