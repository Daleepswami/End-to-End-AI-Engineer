# **10. Binary File Handling**

### **10.1 Introduction to Binary Files**

**1. What are Binary Files?**

- ***Binary files store data in a format specific to the program or system (for example, images, videos, executables, `.docx`, `.xlsx`).***

- ***You typically cannot read these directly in a text editor.***

- ***Binary mode reads and writes bytes exactly as they are stored.***

- ***Use it for images, audio, PDFs and any non-text data.***

### **10.2 Binary Modes (`rb`, `wb`, `ab`)**

**1. Binary Modes**

- ***The letter `"b"` is added to another mode to open a file in binary mode.***

- ***In binary mode, data is read and written as `bytes`, not as strings.***

| Mode | Description | If File Exists | If File Does Not Exist |
|------|-------------|----------------|------------------------|
| `"rb"` | Read binary (images, PDFs, etc.) | Reads from start | Raises `FileNotFoundError` |
| `"wb"` | Write binary | Erases existing content | Creates new |
| `"ab"` | Append binary | New bytes are added at the end | Creates new |

**2. What is `"ab"` Mode?**

- ***Append binary mode (`"ab"`) opens a binary file for writing at the end of the file.***

- ***Existing bytes are preserved.***

- ***If the file does not exist, a new one is created.***

```python
with open("append_demo.bin", "wb") as f:
    f.write(b"Hello")

with open("append_demo.bin", "ab") as f:
    f.write(b" World")                   # added at the end

with open("append_demo.bin", "rb") as f:
    print(f.read())
```

### **10.3 Reading Binary Data**

**1. Reading Bytes**

- ***In `"rb"` mode, `read()` returns `bytes`, not a string.***

- ***Use `decode("utf-8")` to convert bytes into text.***

```python
# Create a binary file first (writing is explained in 10.4)
data = bytes([80, 121, 116, 104, 111, 110])        # raw byte values
with open("sample.bin", "wb") as f:
    f.write(data)

with open("sample.bin", "rb") as f:
    content = f.read()
print(content, list(content))
print(content.decode("utf-8"))
```

**2. File Position: `seek()` and `tell()`**

- ***You can move the file pointer with `seek()` and check its position with `tell()`.***

```python
with open("data.bin", "wb") as f:
    f.write(b"HEADER")
    f.write(bytes(range(65, 71)))                  # bytes for A..F

with open("data.bin", "rb") as f:
    print(f.read(6))                               # first 6 bytes
    print("Position:", f.tell())
    f.seek(8)                                      # jump to byte 8
    print(f.read(2))
```

### **10.4 Writing Binary Data**

**1. Writing Bytes**

- ***In `"wb"` mode, `write()` accepts `bytes`.***

```python
data = bytes([80, 121, 116, 104, 111, 110])        # raw byte values
with open("sample.bin", "wb") as f:
    f.write(data)

with open("data.bin", "wb") as f:
    f.write(b"HEADER")
    f.write(bytes(range(65, 71)))                  # bytes for A..F
```

**2. Saving Python Objects with `pickle`**

- ***The `pickle` module saves whole Python objects to binary files.***

```python
import pickle

record = {"name": "Asha", "skills": ["Python", "SQL"], "exp": 3}
with open("record.pkl", "wb") as f:
    pickle.dump(record, f)

with open("record.pkl", "rb") as f:
    print(pickle.load(f))
```

**Warning:**

- ***Only unpickle files you trust: loading a pickle can run arbitrary code.***
