### **Lists in Python (Data Structure)**
**Introduction**

A list is one of the most commonly used data structures in Python.
It stores multiple values in a single variable and keeps them in order.

**Features of a List**

1. Ordered (elements have an index).
2. Mutable (elements can be changed).
3. Allows duplicate values.
4. Can store different data types.
5. Supports nested lists.


==>> A list is created using square brackets [].

```python
#Syntax
list_name = ["item1", "item2", "elemet"]
```

```python
# Examples
nums = [1, 2, 3, 4, 5]
mixed = [42, "hello", 3.14, True]
nested = [[1, 2], [3, 4], [5, 6]]
empty = []
```

**Accessing Elements**

Python uses indexing to access list elements.

```python
nums = [1, 2, 3, 4, 5]

print(nums[0])
print(nums[-1])
```

```python
# Nested List Access
nested = [[1, 2], [3, 4], [5, 6]]

print(nested[1][0])
```

**List Index**

- Positive indexing starts from 0.
- Negative indexing starts from -1.

**len() Function**

len() returns the number of elements in a list.

```python
nums = [1, 2, 3, 4, 5]

print(len(nums))
```

**List Mutability**

Lists are mutable, which means their contents can be changed after creation.

```python
# Modifying Elements

fruits = ["apple", "banana", "cherry"]
fruits[1] = "mango"
print(fruits)
```

**Lists are Passed by Reference**

This is one of the most important concepts.

```python
a = [1, 2, 3]
b = a

b[0] = 99

print(a)
print(b)
```

**Creating an Independent Copy**

Use copy().

```python
# Use copy().
a = [1, 2, 3]
b = a.copy()

b[0] = 99

print(a)
print(b)
```

**List Slicing**

Slicing extracts a portion of a list.

```python
# Syntax
# list[start : stop : step]

# start → starting index.
# stop → ending index (excluded).
# step → movement.
```

**Reverse a list**

```python
print(nums[::-1])
```

```python
# Slicing Rules
# Slice	          Meaning
# [: ]	            Full copy
# [:n]	            Beginning to n-1
# [n:]	            n to end
# [::2]	            Every second element
# [::-1]	          Reverse
```

### **List Methods**

List methods modify or work with list objects.

**1. append()**

Adds one element at the end.

```python
numbers = [1,2,3]

numbers.append(4)

print(numbers)
```

**2. insert()**

Inserts an element at a specific position.

```python
fruits = ["apple","banana"]
fruits.insert(1,"mango")
print(fruits)
```

**3. extend()**

Adds multiple elements from another iterable.

```python
a = [1,2]
a.extend([3,4])

print(a)
```

**4. remove()**

Removes the first matching value.

```python
nums = [1,2,3,2]

nums.remove(2)

print(nums)
```

**5. pop()**

Removes and returns an element.

```python
nums = [1,2,3]

print(nums.pop())
print(nums)

# Specific index
num = [10,20,30]

print(nums.pop(1))
```

**6.clear()**

Removes every element.

```python
nums = [1,2,3]

nums.clear()

print(nums)
```

**7. index()**

Returns the first matching index.

```python
nums = [10,20,30]

print(nums.index(20))
```

**8. count()**

Counts occurrences.

```python
nums = [1,2,2,3,2]

print(nums.count(2))
```

**9. sort()**

Sorts the original list

```python
nums = [5,1,4,2]
nums.sort()
print(nums)

# Descending
nums.sort(reverse=True)
```

**10.sorted()**

Creates a new sorted list.

```python
nums = [5,1,4]

new = sorted(nums)

print(nums)
print(new)
```

**11. reverse()**

Reverses the original list.

```python
nums = [1,2,3]

nums.reverse()

print(nums)
```

**12.copy()**

Creates a shallow copy.

```python
a = [1,2,3]

b = a.copy()

b[0] = 99

print(a)
print(b)
```

### **List Comprehension**
List comprehension creates lists in a shorter and cleaner way.

```python
# Syntax
# [expression for item in iterable if condition]
```

```python
squares = [x**2 for x in range(1,6)]

print(squares)
```

```python
evens = [x for x in range(20) if x%2==0]

print(evens)
```

**Nested Comprehension**

```python
matrix = [[i*j for j in range(1,4)] for i in range(1,4)]

print(matrix)
```
