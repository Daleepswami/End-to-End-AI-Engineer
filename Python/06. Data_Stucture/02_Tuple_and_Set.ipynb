## **Tuple:-**
**A tuple is an ordered, immutable collection of elements.**
- Ordered: Elements maintain their position and can be accessed using indexes.
- Immutable: Existing elements cannot be changed, added, or removed directly.
- Duplicates allowed: The same value can appear multiple times.
- Mixed data types: A tuple can contain integers, strings, floats, and other objects.

### **Creating a tuple**

```python
coords = (10, 20)
rgb = (255, 128, 0)

print(coords)
print(rgb)
print(coords[0])
print(coords[1])
```

### **Why is a tuple immutable?**

```python
numbers = (10, 20, 30)
numbers[0] = 100
# 'tuple' object does not support item assignment
```

### **Creating a single-element tuple**

```python
a = (42,)
b = (42)

print(type(a))
print(type(b))
```

### **Tuple Unpacking**

Tuple unpacking means assigning individual tuple elements to separate variables.

**Basic unpacking**

```python
point = (3, 7)

x, y = point

print(x)
print(y)
```

### Extended unpacking with *

```python
first, *rest = (1, 2, 3, 4, 5)

print(first)
print(rest)
print(type(rest))
```

### **Swapping variables using tuples**

```python
a = 10
b = 20

a, b = b, a

print(a, b)
```

### **Named Tuples**

1. With a regular tuple, you generally access values using indexes.

2. A named tuple lets you access values using both indexes and named fields.

```python
from collections import namedtuple

Point = namedtuple("Point", ["x", "y"])

p = Point(3, 7)

print(p.x)
print(p.y)
print(p[0])
print(p[1])
```
