## **Tuple:-**
**A tuple is an ordered, immutable collection of elements.**
- Ordered: Elements maintain their position and can be accessed using indexes.
- Immutable: Existing elements cannot be changed, added, or removed directly.
- Duplicates allowed: The same value can appear multiple times.
- Mixed data types: A tuple can contain integers, strings, floats, and other objects.

### **Creating a tuple**

```python
coords = (10, 20)
rgb = (255, 128, 0)

print(coords)
print(rgb)
print(coords[0])
print(coords[1])
```

### **Why is a tuple immutable?**

```python
numbers = (10, 20, 30)
numbers[0] = 100
# 'tuple' object does not support item assignment
```

### **Creating a single-element tuple**

```python
a = (42,)
b = (42)

print(type(a))
print(type(b))
```

### **Tuple Unpacking**

Tuple unpacking means assigning individual tuple elements to separate variables.

**Basic unpacking**

```python
point = (3, 7)

x, y = point

print(x)
print(y)
```

### Extended unpacking with *

```python
first, *rest = (1, 2, 3, 4, 5)

print(first)
print(rest)
print(type(rest))
```

### **Swapping variables using tuples**

```python
a = 10
b = 20

a, b = b, a

print(a, b)
```

### **Named Tuples**

1. With a regular tuple, you generally access values using indexes.

2. A named tuple lets you access values using both indexes and named fields.

```python
from collections import namedtuple

Point = namedtuple("Point", ["x", "y"])

p = Point(3, 7)

print(p.x)
print(p.y)
print(p[0])
print(p[1])
```

## **Set**

**A set is a mutable collection of unique, hashable elements.**

**Its main properties are:**
- Unordered: Sets do not guarantee a fixed positional order.
- Unique elements: Duplicate values are automatically eliminated.
- Mutable: You can add and remove elements.
- No indexing: You cannot access elements using s[0].
- Hashable elements only: Lists and dictionaries cannot be stored directly inside a set.

### **Creating a set**



```python
s = {1, 2, 3, 2, 1}

print(s)
print(type(s))
print(len(s))
```

### **Removing duplicates from a list**

```python
list = [1, 2, 2, 3, 3, 4]

unique = set(list)

print(unique)
```

### **Creating an empty set**

```python
a = {}
b = set()

print(type(a))
print(type(b))
```

### **Set Operations**
Python sets support mathematical operations such as union, intersection, difference, and symmetric difference.

```python
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}
```

**1. Union (|)**

Union combines all unique elements from both sets.

```python
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

print(A | B)
print(A.union(B))
```

**2. Intersection (&)**

Intersection returns only the elements common to both sets.

```python
print(A & B)
print(A.intersection(B))
```

**3. Difference (-)**

Difference returns elements present in the left-hand set but absent from the right-hand set.

```python
print(A - B)
print(B - A)
```

**4. Symmetric difference (^)**

Symmetric difference returns elements that belong to either set, but not both.

```python
print(A ^ B)
print(A.symmetric_difference(B))
```

### **Set Methods**

Set methods allow you to add elements, remove elements, and compare sets.

**1. add(x) — Add an element**

```python
s = {1, 2, 3}

s.add(4)
print(s)

s.add(2)
print(s)
```

**2. remove(x)**

 — Remove an element

```python
s = {1, 2, 3}

s.remove(2)
print(s)
```

**3. discard(x)**

— Remove without an error if absent

```python
s = {1, 2, 3}

s.discard(2)
s.discard(5)

print(s)
```

**4. pop()**

— Remove and return an element

```python
s = {10, 20, 30}

value = s.pop()

print(value)
print(s)
```

**5. clear()**

— Remove everything

```python
s = {1, 2, 3}

s.clear()

print(s)
print(len(s))
```

**6. union(), intersection(), and difference()**

These methods correspond to the mathematical operations described above.

```python
A = {1, 2, 3}
B = {3, 4, 5}

print(A.union(B))        
print(A.intersection(B)) 
print(A.difference(B)) 
```

**7. isdisjoint()**

— Check whether sets have no common elements

```python
A = {1, 2}
B = {3, 4}
C = {2, 5}

print(A.isdisjoint(B))
print(A.isdisjoint(C))
```

**Membership Testing and Time Complexity**

Sets are particularly useful when you need to check whether an element exists in a collection.

```python
numbers = {10, 20, 30, 40, 50}

print(30 in numbers)
print(99 in numbers)
```
