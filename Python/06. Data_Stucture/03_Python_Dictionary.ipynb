# **01. Introduction to Dictionary**

### **1.1 What is a Dictionary?**

**1. What is a Dictionary?**

- A dictionary (`dict`) stores data as **key-value pairs**.

- Think of your phone's contact list: you look up a **name (key)** to get a **number (value)**.

- ***Each key acts as a unique label used to retrieve its associated value.***

- **Key:** the label you use to look something up.

- **Value:** the data stored under that label.

- ***Dictionaries are written with curly braces `{ }`.***

- ***Each pair is written as `key: value`, and pairs are separated by commas.***

**2. Properties of a Dictionary**

| Property | Description |
|---|---|
| Syntax | Curly braces `{ }`, with pairs written as `key: value` |
| Keys | Must be **unique** and **immutable** (str, int, float, tuple) |
| Values | Any data type, including lists and other dictionaries |
| Ordering | Insertion order is preserved (Python 3.7+) |
| Mutability | Mutable: pairs can be added, changed and removed |

````python
student = {"name": "Asha", "age": 21, "course": "Data Analytics"}
print(student)
print(type(student))
````

### **1.2 Dictionary vs Other Collections**

| Feature | List | Tuple | Set | Dictionary |
|---|---|---|---|---|
| Syntax | `[1, 2]` | `(1, 2)` | `{1, 2}` | `{"k": 1}` |
| Ordered | Yes | Yes | No | Yes (3.7+) |
| Mutable | Yes | No | Yes | Yes |
| Duplicates | Allowed | Allowed | Not allowed | Keys unique |
| Access | By index | By index | No indexing | By key |
| Best for | Changing sequences | Fixed records | Unique items | Lookups by key |

**Key Point:**

- Choose the collection by asking: does order matter, will it change, are duplicates allowed, do I look up by key?

### **1.3 Rules for Keys and Values**

**1. Keys Must Be Unique**

- ***A repeated key keeps the last value.***

````python
d = {"a": 1, "a": 2}
print(d)                 # {'a': 2}
````

**2. Keys Must Be Immutable**

- ***A list cannot be a key, but a tuple can.***

- ***A tuple can be a key because it is hashable; a list is mutable, so it is not hashable.***

````python
try:
    d = {[1, 2]: "x"}        # list as a key
except TypeError as e:
    print("TypeError:", e)   # unhashable type: 'list'

d = {(1, 2): "x"}            # tuple as a key works
print(d)                     # {(1, 2): 'x'}
````

**3. Values Can Be Anything**

- ***Values can be numbers, strings, lists, and even other dictionaries.***

````python
student = {
    "name": "Asha",
    "subjects": ["Maths", "Python", "SQL"],        # list as a value
    "address": {"city": "Chennai", "pin": 600001}  # dictionary as a value
}

print(student["subjects"][1])          # Python
print(student["address"]["city"])      # Chennai
````

# **02. Creating a Dictionary**

### **2.1 Creating a Dictionary with `{ }` and `dict()`**

**1. Ways to Create a Dictionary**

1. ***With values, using curly braces (literal syntax).***
2. ***An empty dictionary, which is filled later.***
3. ***Using the `dict()` constructor.***

**2. Important Note**

- ***`{}` creates an empty dictionary, not a set. Use `set()` for an empty set.***

````python
# 1. With values
student = {"name": "Asha", "age": 21}

# 2. Empty dictionary (we fill it later)
empty = {}

# 3. Using dict()
person = dict(name="Ravi", age=25)

print(student)   # {'name': 'Asha', 'age': 21}
print(empty)     # {}
print(person)    # {'name': 'Ravi', 'age': 25}
````

### **2.2 Creating a Dictionary from a List of Pairs**

**1. `dict()` with Pairs**

- ***`dict()` can also convert a list of `(key, value)` pairs into a dictionary.***

````python
print(dict([("a", 1), ("b", 2)]))        # {'a': 1, 'b': 2}
````

# **03. Accessing Values**

### **3.1 Accessing with Square Brackets**

**1. Using the Key**

- ***Use the key inside square brackets to get its value.***

````python
student = {"name": "Asha", "age": 21, "course": "Data Analytics"}

print(student["name"])     # Asha
print(student["age"])      # 21
````

### **3.2 Accessing with `get()`**

**1. What if the Key Does Not Exist?**

- ***`student["city"]` raises a `KeyError` and the program crashes.***

- ***Use `get()` to return a default value instead.***

````python
try:
    print(student["city"])
except KeyError as e:
    print("KeyError:", e)                # KeyError: 'city'

print(student.get("city"))               # None
print(student.get("city", "Unknown"))    # Unknown
print(student.get("name", "Unknown"))    # Asha (key exists, so real value)
print(student.get("gpa", 0))             # 0 (default if key missing)
````

**2. Comparison**

| Approach | Syntax | If key is missing |
|---|---|---|
| Square brackets | `d[key]` | Raises `KeyError` |
| `get()` method | `d.get(key, default)` | Returns `default` (or `None`) |

**Best Practice:**

- ***Use `d[key]` when you are sure the key exists. Use `d.get(key, default)` when you are not sure, or for optional data.***

# **04. Adding, Updating and Deleting**

### **4.1 Adding and Updating**

**1. `d[key] = value`**

- ***The same syntax does both jobs.***

- If the key is **new**, it adds a pair.

- If the key **already exists**, it overwrites the value.

````python
student = {"name": "Asha", "age": 21, "course": "Data Analytics"}

student["city"] = "Chennai"    # key is new, so it is ADDED
student["age"] = 22            # key exists, so it is UPDATED

print(student)
# {'name': 'Asha', 'age': 22, 'course': 'Data Analytics', 'city': 'Chennai'}
````

### **4.2 Deleting**

**1. Ways to Delete**

- ***`del d[key]` deletes the pair.***

- ***`d.pop(key)` removes the pair and returns its value.***

- ***`d.clear()` removes all pairs.***

````python
student = {"name": "Asha", "age": 22, "course": "Data Analytics", "city": "Chennai"}
del student["course"]          # DELETE the pair
print(student)                 # {'name': 'Asha', 'age': 22, 'city': 'Chennai'}

d = {'a': 1, 'b': 2, 'c': 3}
del d['b']                     # Remove key 'b'
print(d.pop('c'))              # Remove & return value -> 3
d.clear()                      # Remove all
print(d)                       # {}
````

# **05. Checking Keys and Length**

### **5.1 Membership Test with `in`**

**1. The `in` Operator**

- The `in` operator checks **keys only**, not values.

- ***To check values, use `value in d.values()`.***

- ***`len(d)` gives the number of pairs.***

````python
student = {"name": "Asha", "age": 22, "city": "Chennai"}

print("name" in student)             # True
print("phone" in student)            # False
print("Asha" in student)             # False (Asha is a value, not a key)
print("Asha" in student.values())    # True
print(len(student))                  # 3 (number of pairs)

if "phone" in student:
    print(student["phone"])
else:
    print("No phone number saved")
````

# **06. Looping over a Dictionary**

### **6.1 Iterating over Keys, Values and Items**

| Method | Iterates over |
|---|---|
| `for k in d` | Keys |
| `d.values()` | Values |
| `d.items()` | (key, value) pairs |

````python
student = {"name": "Asha", "age": 22, "city": "Chennai"}

# Loop over keys (default)
for key in student:
    print(key)

# Loop over values only
for value in student.values():
    print(value)

# Loop over keys AND values together (most useful)
for key, value in student.items():
    print(key, "->", value)
````

**2. Using the Key to Look Up the Value**

````python
info = {'name': 'Bob', 'age': 30, 'city': 'Delhi'}

for key in info:
    print(key, '->', info[key])

# Pythonic way
for key, value in info.items():
    print(f'{key}: {value}')
````

# **07. Important Dictionary Methods**

### **7.1 Methods of a Dictionary**

| Method | Description |
|---|---|
| `keys()` | Returns all keys |
| `values()` | Returns all values |
| `items()` | Returns all (key, value) pairs |
| `get(k, default)` | Returns the value for `k`, or `default` if missing |
| `update(other)` | Adds or overwrites pairs from another dictionary |
| `pop(k)` | Removes `k` and returns its value |
| `setdefault(k, v)` | Returns the value of `k`; inserts `v` if `k` is missing |
| `clear()` | Removes all pairs |

````python
marks = {"maths": 90, "science": 85}

# keys(), values(), items()
print(marks.keys())      # dict_keys(['maths', 'science'])
print(marks.values())    # dict_values([90, 85])
print(marks.items())     # dict_items([('maths', 90), ('science', 85)])

# update(): add or overwrite several pairs at once
marks.update({"science": 88, "english": 76})
print(marks)             # {'maths': 90, 'science': 88, 'english': 76}

# pop(): remove a key and get its value back
removed = marks.pop("english")
print(removed)           # 76
print(marks)             # {'maths': 90, 'science': 88}

# setdefault(): add only if the key is missing
print(marks.setdefault("history", 70))   # 70 (key was missing, so added)
print(marks.setdefault("maths", 0))      # 90 (key exists, so unchanged)
print(marks)             # {'maths': 90, 'science': 88, 'history': 70}
````

# **08. Nested Dictionaries**

### **8.1 Dictionary inside a Dictionary**

**1. Nested Dictionary**

- ***A dictionary can hold other dictionaries as values.***

- ***Use one key after another to reach the inner value.***

````python
company = {"sales": {"head": "Ravi", "size": 12},
           "tech":  {"head": "Asha", "size": 25}}       # nested dict

print(company["tech"]["head"])                           # Asha
print(company["sales"]["size"])                          # 12
````

# **09. Dictionary Comprehension**

### **9.1 Building Dictionaries Concisely**

**1. What is a Dictionary Comprehension?**

- ***A dictionary comprehension creates a dictionary from an iterable in one line.***

**2. Syntax:**

```python
{key_expression: value_expression for item in iterable if condition}
```

````python
# Squares as dict
squares = {x: x**2 for x in range(1, 6)}
print(squares)                       # {1: 1, 2: 4, 3: 9, 4: 16, 5: 25}

# Filter only even values
evens = {k: v for k, v in squares.items() if v % 2 == 0}
print(evens)                         # {2: 4, 4: 16}

# Invert a dictionary
inverted = {v: k for k, v in squares.items()}
print(inverted)                      # {1: 1, 4: 2, 9: 3, 16: 4, 25: 5}

# Keep only passed students
marks = {"Asha": 91, "Ravi": 64, "Meena": 85}
passed = {name: m for name, m in marks.items() if m >= 70}
print(passed)                        # {'Asha': 91, 'Meena': 85}
````

# **10. Sorting, Merging and Counting**

### **10.1 Sorting a Dictionary**

**1. Sorting by Value**

- ***`sorted(d.items(), key=lambda kv: kv[1])` sorts the pairs by value.***

- ***`reverse=True` sorts in descending order.***

````python
marks = {"Asha": 91, "Ravi": 64, "Meena": 85}
print(sorted(marks.items(), key=lambda kv: kv[1], reverse=True))
# [('Asha', 91), ('Meena', 85), ('Ravi', 64)]
````

### **10.2 Merging Dictionaries**

**1. The `|` Operator**

- ***The `|` operator merges two dictionaries into a new dictionary (Python 3.9+).***

- ***If a key is in both, the value from the right-hand dictionary is used.***

- ***`update()` changes the dictionary in place, while `|` creates a new one.***

````python
merged = marks | {"Kiran": 77}
print(merged)       # {'Asha': 91, 'Ravi': 64, 'Meena': 85, 'Kiran': 77}
````

### **10.3 Counting with `Counter`**

**1. What is `Counter`?**

- ***`Counter` (from the `collections` module) counts how many times each item appears.***

- ***`most_common(n)` returns the `n` most frequent items.***

````python
from collections import Counter

print(Counter("mississippi").most_common(2))      # [('i', 4), ('s', 4)]
````

### **10.4 Grouping with `defaultdict`**

**1. What is `defaultdict`?**

- ***`defaultdict` is a dictionary that creates a default value for a missing key.***

- ***`defaultdict(list)` is useful for grouping items.***

````python
from collections import defaultdict

groups = defaultdict(list)
for name, dept in [("Asha", "tech"), ("Ravi", "sales"), ("Meena", "tech")]:
    groups[dept].append(name)
print(dict(groups))                  # {'tech': ['Asha', 'Meena'], 'sales': ['Ravi']}
````
