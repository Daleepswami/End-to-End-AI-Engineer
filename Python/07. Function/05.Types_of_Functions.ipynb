## **Types of Functions in Python**

### **1. Introduction**

*Python provides different types of functions to perform tasks, reuse code, and simplify programming.*

**The main types covered in this topic are:**

1. ***Built-in Functions***

2. ***User-Defined Functions***

3. ***Lambda Functions***

4. ***Recursive Functions***

### 1. **Built-in Functions**

Built-in functions are functions that Python provides by default. You can use them directly without defining them yourself or importing a module.

| Function | Description | Example |
|---|---|---|
| `print()` | Displays output | `print("Hello")` |
| `len()` | Returns the number of items | `len("Python")` |
| `sum()` | Calculates the sum of items | `sum([10, 20, 30])` |
| `max()` | Returns the largest item | `max([10, 50, 30])` |
| `min()` | Returns the smallest item | `min([10, 50, 30])` |
| `type()` | Returns an object's type | `type(10)` |
| `round()` | Rounds a number | `round(3.14159, 2)` |

### 2. **User-Defined Functions**

1. User-defined functions are functions that programmers create themselves using the def keyword.

2. They are useful when you want to perform a specific task that you may need to execute multiple times.

```python
def calculate_square(number):
    return number * number

print(calculate_square(5))
print(calculate_square(8))
print(calculate_square(10))
```

### **3. Lambda Functions**

1. A lambda function is a small, anonymous function defined using the lambda keyword.

2. It can contain any number of parameters but only one expression. The expression's result is returned automatically.

**Syntax**

```python
lambda parameters: expression
```

**Example 1: Square a number**

```python
square = lambda x: x * x

print(square(6))
print(square(9))
```

### **4. Recursive Functions**

1. A recursive function is a function that calls itself to solve a problem.

2. Recursion can break a large problem into smaller versions of the same problem.

**A recursive function generally needs two important parts:**

1. **Base case:** The condition that stops recursion.

2. **Recursive case:** The part where the function calls itself with a smaller or simpler input

```python
def factorial(n):
    if n == 0 or n == 1:
        return 1

    return n * factorial(n - 1)

print(factorial(5))
print(factorial(3))
print(factorial(0))
```
