## **Parameters and Arguments in Python**

**1. Introduction to Parameters and Arguments**

Parameters and arguments are used to pass information to a function so that it can perform a task using different input values.

They allow us to create flexible and reusable functions instead of writing separate functions for different values.

```python
def greet(name):
    print("Hello,", name)

greet("Rahul")
greet("Aman")
```

  ==>> name is a parameter because it appears in the function definition.

  ==>> "Rahul" and "Aman" are arguments because they are passed during function calls.

  ==>> The function uses the supplied value to display a personalized greeting.

**What Is a Parameter?**

1. A parameter is a variable specified in a function definition that receives a value when the function is called.

2. Parameters act as placeholders for input values

```python
def greet(name):
    print("Hello,", name)
```

**What Is an Argument?**

1. An argument is a value or expression supplied to a function when it is called.

2. Arguments provide the actual input that the function uses.

```python
def greet(name):
    print("Hello,", name)

greet("Rahul")
```

```python
def add(a, b):
    print(a + b)

add(10, 20)
```

**Function With One Parameter**

A function can accept a single parameter.

```python
def square(number):
    print(number * number)

square(5)
square(8)
```

**Function With Multiple Parameters**

A function can accept two or more parameters.

```python
def add(a, b):
    print(a + b)

add(10, 20)
add(50, 30)
```

## **Return Statement in Python**

**What is the return Statement?**

1. The return statement is used to send a value back from a function to the place where the function was called.

2. It allows a function to calculate a result and make that result available for further use in the program.

```python
# Syntax:

def function_name():
    # Perform some operation
    return value
```

```python
def add_numbers(a, b):
    return a + b

result = add_numbers(10, 20)
print(result)
```

### **Why Do We Use return?**

The return statement is useful when we want to use a function's result somewhere else in our program

```python
def calculate_square(number):
    return number * number

square = calculate_square(5)

print(square)
print(square + 10)
```

### **Difference Between print() and return**

*This is one of the most important concepts to understand.*

*1. print() displays information on the screen.*

*2. return sends a value back to the caller.*

```python
def add(a, b):
    print(a + b)

result = add(5, 3)

print("Result:", result)
```

### **A Function Without an Explicit return**

*1. If a function does not contain an explicit return statement, Python automatically returns None when the function finishes.*

```python
def greeting(name):
    print(f"Hello, {name}!")

result = greeting("Alice")

print("Returned value:", result)
```

### **Returning Multiple Values**
*1. Python allows a function to return multiple values in a single return statement.*

*2. These values are automatically packed into a tuple.*

```python
def area_and_perimeter(length, width):
    area = length * width
    perimeter = 2 * (length + width)

    return area, perimeter

a, p = area_and_perimeter(5, 3)

print("Area:", a)
print("Perimeter:", p)
```

### **Returning Different Data Types**

*1. A function can return different kinds of values, such as integers, floats, strings, Boolean values, lists, or tuples.*

```python
def get_name():
    return "Python"

def is_valid():
    return True

def get_numbers():
    return [10, 20, 30]

print(get_name())
print(is_valid())
print(get_numbers())
```
