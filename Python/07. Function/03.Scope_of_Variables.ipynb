### **Scope of Variables in Python**

**What Is Variable Scope?**

1. Variable scope defines where a variable can be accessed or used in a Python program.

2. In Python, variables created in different places may have different scopes.

**The two main types covered in this section are:**

1. Local Variables

2. Global Variables

Understanding scope helps us write functions that are easier to understand, test, and debug.

**1. Local Variables**

1. A local variable is a variable defined inside a function.

2. It can normally be accessed only within the function in which it is defined.

```python
def greet():
    message = "Hello Python"
    ''' This is local variable
    beacuse this variable inside of function'''
    print(message)

greet()
```

**Accessing a Local Variable Outside Its Function**

```python
def greet():
    message = "Hello Python"
    print(message)

greet()

# print(message)
# NameError: name 'message' is not defined
```

**Lifecycle of a Local Variable**



1. A local variable is created when the function starts executing.

2. It exists during that function call.

3. When the function finishes, its local name is no longer available outside that function.

```python
def calculate():
    result = 10 + 20
    return result

answer = calculate()

print(answer)

# Returning a value from a function does not automatically
# make its local variable accessible outside the function.
# You can return the value and store it in another variable.
```

**Global Variables**

1. A global variable is a variable defined outside all functions, usually at the top level of a Python script.

2. It can be accessed from inside functions, provided there is no conflicting local name.

```python
message = "Hello Python"

def greet():
    print(message)

greet()
print(message)
```

**Lifecycle of a Global Variable**

1. A global variable is created when the relevant top-level assignment executes and generally remains available for the rest of that program's execution.

*The two variables have the same name but belong to different scopes.*

| Local variable | Global variable |
|---|---|
| Defined inside a function | Defined outside functions |
| Normally accessible within its function | Can be read from functions and top-level code |
| A local assignment affects the local name | A global assignment changes the global name |
| Useful for self-contained function logic | Useful for values shared across a program, but should be used carefully |


### **The global Keyword**

1. By default, when you assign a value to a variable inside a function, Python treats that name as local unless you declare it global.

2. The global keyword tells Python that the name refers to a variable in the module's global scope.

#### **Example Without global**



```python
counter = 0

def increment():
    counter = 100
    print("Inside function:", counter)

increment()
print("Outside function:", counter)
```

#### **Example With global**



```python
counter = 0

def increment():
    global counter
    counter = counter + 1
    print("Inside function:", counter)

increment()
print("Outside function:", counter)
```

**Reading vs Modifying a Global Variable**


**Reading a Global Variable**

1. You can read a global variable inside a function without using global.

```python
price = 500

def display_price():
    print(price)

display_price()
```

**Modifying a Global Variable**

1. If you want to reassign a global variable inside a function, use global.

```python
price = 500

def update_price():
    global price
    price = 700

update_price()
print(price)
```
