## **Types of Arguments in Python**

**1. Introduction**

*1. In Python, arguments are the actual values passed to a function when it is called.*

*2. They provide input to a function so that it can perform a task using different values.*

**Python supports several ways to pass arguments:**

1. *Positional Arguments*

2. *Keyword Arguments*

3. *Default Arguments*

4. **Variable-Length Arguments**
(*args and **kwargs)**

*   1. Variable-Length Arguments(*args)
*   2. keyword Variable-Length Arguments(**kwargs)




### **1. Positional Arguments**

*Positional arguments are matched to function parameters according to their position or order.*

```python
def describe_pet(animal_type, pet_name):
    print(f"I have a {animal_type}.")
    print(f"Its name is {pet_name}.")

describe_pet("dog", "Buddy")
```

**1. What happens if the order changes?**

```python
describe_pet("Buddy","dog" )
```

**2. What happens if an argument is missing?**

```python
describe_pet("dog")
# TypeError: describe_pet() missing 1 required positional argument: 'pet_name'
```

### **2. Keyword Arguments**

1. *Keyword arguments are passed by explicitly specifying the parameter name along with its value.*

2. *They make function calls more readable and allow arguments to be passed in a different order.*

```python
def describe_pet(animal_type, pet_name):
    print(f"I have a {animal_type}.")
    print(f"Its name is {pet_name}.")

describe_pet(animal_type="cat", pet_name="Mittens")
```

**Changing the order:-** 1. The order does not matter here because Python matches each value using its parameter name.

```python
describe_pet(pet_name="Harry", animal_type="hamster")
```

**Default Arguments**

1. *A default argument is a parameter that has a predefined value in the function definition.*

2. *If the caller does not provide a value for that parameter, Python uses the default value.*

```python
def greet_person(name="Guest"):
    print(f"Hello, {name}!")

greet_person("Sarah")
greet_person()
```

**Important rule**

*Parameters with default values must come after parameters without default values.*

```python
def calculate(a, b=10):
    return a + b
calculate(5)
```

**If that were not the case**

```python
def calculate(a=10, b):
    return a + b
calculate(10)
# The second definition raises a SyntaxError
```

**Variable-Length Arguments**

**Sometimes we do not know in advance how many arguments a function will receive. Python provides two special forms for handling this:**

 * **args — collects extra positional arguments into a tuple*

 * ***kwargs — collects extra keyword arguments into a dictionary.)*

***args**

Use *args when a function needs to accept any number of additional positional arguments.

```python
def add_numbers(*args):
    print(args)
    print("Total:", sum(args))

add_numbers(10, 20)
add_numbers(10, 20, 30, 40)
```

****kwargs**

Use **kwargs when a function needs to accept any number of additional keyword arguments.

```python
def show_details(**kwargs):
    print(kwargs)

show_details(name="Asha", city="Jaipur")
show_details(name="Ravi", age=25, role="Developer")
```

### **Combining Argument Types**

```python
def make_profile(name, role="Analyst", *skills, **details):
    print("Name:", name)
    print("Role:", role)
    print("Skills:", skills)
    print("Details:", details)

make_profile("Asha")

print("-" * 30)

make_profile(
    "Ravi",
    "Engineer",
    "Python",
    "SQL",
    city="Chennai",
    exp=3
)
```
