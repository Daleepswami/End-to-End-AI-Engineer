## **1. Docstrings**
1. *A docstring is a string written as the first statement inside a function, class, or module. It explains the purpose of that code.*

2. *Docstrings are useful when you or another programmer need to understand what a function does without reading its entire implementation.*

 ### **1.Writing a docstring**

```python
def calculate_square(number):
    """Return the square of the given number."""
    return number * number

print(calculate_square(5))
```

###2. **Accessing a docstring using __doc__**

*Python stores a function's docstring in its __doc__ attribute.*

```python
def greet(name):
    """Print a greeting for the given name."""
    print(f"Hello, {name}!")

print(greet.__doc__)
```

### 3.**Accessing documentation using help()**
***Python's built-in help() function displays documentation about an object.***

```python
def add(a, b):
    """Return the sum of two numbers."""
    return a + b

help(add)

# The help output includes the function's name and its docstring.
```

## **2. Type Hints**


1. Type hints indicate the expected data types of function parameters and return values.

2. They help readers understand how a function should be used and allow development tools to detect some potential type-related mistakes.

### **1. Type hints for parameters**

```python
def add(a: int, b: int):
    return a + b

print(add(10, 20))
```

### **2. Type hints for the return value**

```python
def calculate_square(number: int) -> int:
    return number * number

result = calculate_square(5)
print(result)
```

### **3. Different parameter and return types**

```python
def format_price(price: float) -> str:
    return f"Price: ₹{price:.2f}"

print(format_price(99.5))
```

## **3.First-Class Functions**
*In Python, functions are first-class objects. This means functions can be treated like other values in several important ways:*



*   Assigned to variables.
*   Passed as arguments to other functions
*   Returned from other functions.


### **3.1 Assigned to variables**

```python
def greet(name):
    return f"Hello, {name}!"

message_function = greet

print(message_function("Asha"))
```

### **3.2 Passed as arguments to other functions.**

```python
def square(number):
    return number * number

def display_result(func, value):
    result = func(value)
    print("Result:", result)

display_result(square, 5)
```

### **3.3 Returned from other functions.**

```python
def get_multiplier():
    def multiply_by_two(number):
        return number * 2

    return multiply_by_two

double = get_multiplier()

print(double(5))
print(double(10))
```

## **4. Higher-Order Functions**

1. A higher-order function is a function that takes another function as an argument, returns a function, or does both.

### **4.1. A function that accepts another function**

```python
def add(a, b):
    return a + b

def multiply(a, b):
    return a * b

def calculate(func, x, y):
    return func(x, y)

print(calculate(add, 10, 5))
print(calculate(multiply, 10, 5))
```

### **4.2 A higher-order function that applies a function twice**

```python
def apply_twice(func, value):
    return func(func(value))

def increment(number):
    return number + 1

print(apply_twice(increment, 5))
```

## **5. Functional Tools: map(), filter(), and reduce()**
***Python provides tools that work with functions to transform, select, or combine data.***

### **5.1 map()**

***The map() function applies a given function to every item in an iterable.***

```python
numbers = [1, 2, 3, 4, 5]

squares = list(map(lambda x: x * x, numbers))

print(squares)
```

### **5.2 filter()**
***The filter() function selects items for which a function returns a truthy result.***

```python
numbers = [1, 2, 3, 4, 5, 6]

even_numbers = list(filter(lambda x: x % 2 == 0, numbers))

print(even_numbers)
```

### ***5.3 reduce()***
***1. The reduce() function repeatedly combines items to produce a single accumulated result. It must be imported from functools.***

```python
from functools import reduce

numbers = [1, 2, 3, 4, 5]

product = reduce(lambda a, b: a * b, numbers)

print(product)
```
