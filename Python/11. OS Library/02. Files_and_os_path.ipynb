# **03. Working with Files**

### **3.1 `os.rename()` – Rename Files**

**1. What is `os.rename()`?**

- ***`os.rename(old, new)` renames a file or a directory.***

- ***It raises `FileNotFoundError` if the old name does not exist.***

- ***If the new path is in a different folder, the file is also moved there.***

**2. Syntax:**

```python
os.rename(old_name, new_name)
```

````python
import os

with open("old.txt", "w") as f:             # create a file to rename
    f.write("sample")

os.rename('old.txt', 'new.txt')
print(os.path.exists('old.txt'), os.path.exists('new.txt'))

try:
    os.rename("no_such_file.txt", "x.txt")
except FileNotFoundError as e:
    print("Error:", e)
````

**3. Moving a File While Renaming**

````python
os.makedirs("archive", exist_ok=True)
os.rename("new.txt", "archive/renamed.txt")
print(os.listdir("archive"))
````

### **3.2 `os.remove()` – Delete Files**

**1. What is `os.remove()`?**

- ***`os.remove(path)` deletes a file.***

- ***It raises `FileNotFoundError` if the file does not exist.***

- ***It cannot delete directories (use `os.rmdir()` for an empty directory).***

- ***Deleted files do not go to the recycle bin, so use it carefully.***

**2. Syntax:**

```python
os.remove(path)
```

````python
os.remove('archive/renamed.txt')
print(os.path.exists('archive/renamed.txt'))

try:
    os.remove("archive/renamed.txt")        # already deleted
except FileNotFoundError as e:
    print("Error:", e)

os.rmdir("archive")
````

### **3.3 `os.path.exists()` – Check Path Existence**

**1. What is `os.path.exists()`?**

- ***`os.path.exists(p)` checks whether a path exists.***

- ***It returns `True` for both files and directories, and `False` if the path does not exist.***

**2. Syntax:**

```python
os.path.exists(path)
```

````python
with open("file.txt", "w") as f:
    f.write("sample")

print(os.path.exists('file.txt'))           # True / False
print(os.path.exists('no_such_file.txt'))   # False
print(os.path.exists('.'))                  # True -> a directory also exists
````

**3. Use: Avoiding an Error**

````python
if os.path.exists("file.txt"):
    os.remove("file.txt")
    print("file.txt deleted")
else:
    print("file.txt not found")
````

### **3.4 `os.path.isfile()` – Check Whether a Path Is a File**

**1. What is `os.path.isfile()`?**

- ***`os.path.isfile(p)` returns `True` only if the path exists and is a file.***

- ***It returns `False` for directories and for paths that do not exist.***

**2. Syntax:**

```python
os.path.isfile(path)
```

````python
with open("file.txt", "w") as f:
    f.write("sample")

print(os.path.isfile("file.txt"))           # True  -> it is a file
print(os.path.isfile("."))                  # False -> it is a directory
print(os.path.isfile("no_such_file.txt"))   # False -> does not exist
````

### **3.5 `os.path.isdir()` – Check Whether a Path Is a Directory**

**1. What is `os.path.isdir()`?**

- ***`os.path.isdir(p)` returns `True` only if the path exists and is a directory.***

- ***It returns `False` for files and for paths that do not exist.***

**2. Syntax:**

```python
os.path.isdir(path)
```

````python
print(os.path.isdir("."))                   # True  -> it is a directory
print(os.path.isdir("file.txt"))            # False -> it is a file
print(os.path.isdir("no_such_folder"))      # False -> does not exist
````

**3. Comparison of the Three Checks**

| Path | `exists()` | `isfile()` | `isdir()` |
|------|------------|------------|-----------|
| An existing file | `True` | `True` | `False` |
| An existing directory | `True` | `False` | `True` |
| A path that does not exist | `False` | `False` | `False` |

````python
for path in ["file.txt", ".", "no_such_path"]:
    print(f"{path:14} exists={os.path.exists(path)}  isfile={os.path.isfile(path)}  isdir={os.path.isdir(path)}")

os.remove("file.txt")
````

# **04. Working with os.path**

### **4.1 Introduction to `os.path`**

**1. What is `os.path`?**

- ***`os.path` is a submodule of `os` that provides utilities for working with file and folder paths.***

- ***Most of its functions only work on path strings, so they handle the differences between operating systems for you.***

| Function | Description |
|----------|-------------|
| `os.path.join(a, b)` | Join paths safely (handles OS separators) |
| `os.path.exists(p)` | Check if path exists |
| `os.path.isfile(p)` | True if p is a file |
| `os.path.isdir(p)` | True if p is a directory |
| `os.path.basename(p)` | File name from full path |
| `os.path.dirname(p)` | Directory portion of path |
| `os.path.split(p)` | Split path into directory and file name |
| `os.path.splitext(p)` | Split name and extension: `('file', '.txt')` |
| `os.path.abspath(p)` | Absolute path |
| `os.path.getsize(p)` | Size of a file in bytes |

### **4.2 `os.path.join()` – Join Path Components**

**1. What is `os.path.join()`?**

- ***`os.path.join(a, b, ...)` joins path components using the correct separator for the operating system.***

- ***It is safer than joining strings with `+` or hard-coded `/` and `\`.***

- ***If a later component is an absolute path, the earlier components are discarded.***

**2. Syntax:**

```python
os.path.join(path1, path2, ...)
```

````python
import os

# Safe cross-platform path building
full = os.path.join('folder', 'sub', 'file.txt')
print(full)
# Windows: folder\sub\file.txt / Linux: folder/sub/file.txt

print(os.path.join("data", "/absolute/file.txt"))     # absolute part wins (Linux / macOS)
````

### **4.3 `os.path.basename()` – Get File Name**

**1. What is `os.path.basename()`?**

- ***`os.path.basename(p)` returns the last part of a path, which is usually the file name.***

**2. Syntax:**

```python
os.path.basename(path)
```

````python
path = '/home/user/documents/notes.txt'
print(os.path.basename(path))               # notes.txt
````

### **4.4 `os.path.dirname()` – Get Directory Name**

**1. What is `os.path.dirname()`?**

- ***`os.path.dirname(p)` returns the directory part of a path (everything before the file name).***

**2. Syntax:**

```python
os.path.dirname(path)
```

````python
print(os.path.dirname(path))                # /home/user/documents
````

### **4.5 `os.path.split()` – Split Path**

**1. What is `os.path.split()`?**

- ***`os.path.split(p)` splits a path into a tuple `(head, tail)`.***

- ***`head` is the directory part and `tail` is the file name.***

- ***It gives the same result as `(os.path.dirname(p), os.path.basename(p))`.***

**2. Syntax:**

```python
os.path.split(path)
```

````python
print(os.path.split(path))                  # ('/home/user/documents', 'notes.txt')

folder, file_name = os.path.split(path)     # unpack the tuple
print(folder)
print(file_name)

print(os.path.split("/home/user/"))         # a trailing slash gives an empty tail
````

### **4.6 `os.path.splitext()` – Split File Extension**

**1. What is `os.path.splitext()`?**

- ***`os.path.splitext(p)` splits a path into the name and the extension.***

- ***It returns a tuple `(root, ext)`, where `ext` starts with a dot.***

- ***Only the last extension is split off.***

**2. Syntax:**

```python
os.path.splitext(path)
```

````python
print(os.path.splitext(path))               # ('/home/user/documents/notes', '.txt')

print(os.path.splitext("archive.tar.gz"))   # only the last extension is split
print(os.path.splitext("README"))           # no extension -> empty string
````

### **4.7 `os.path.abspath()` – Get Absolute Path**

**1. What is `os.path.abspath()`?**

- ***`os.path.abspath(p)` returns the absolute (full) path of a file or folder.***

- ***A relative path is combined with the current working directory.***

- ***It also cleans the path, for example by removing `..`.***

**2. Syntax:**

```python
os.path.abspath(path)
```

````python
print(os.getcwd())
print(os.path.abspath("data.txt"))          # current directory + data.txt
print(os.path.abspath("sub/../data.txt"))   # '..' is resolved
````

### **4.8 `os.path.getsize()` – Get File Size**

**1. What is `os.path.getsize()`?**

- ***`os.path.getsize(p)` returns the size of a file in bytes.***

- ***It raises `FileNotFoundError` if the path does not exist.***

**2. Syntax:**

```python
os.path.getsize(path)
```

````python
with open("size_demo.txt", "w") as f:
    f.write("Hello, os.path!")

size = os.path.getsize("size_demo.txt")     # Size in bytes
print(size, "bytes")
print(round(size / 1024, 4), "KB")

try:
    os.path.getsize("no_such_file.txt")
except FileNotFoundError as e:
    print("Error:", e)

os.remove("size_demo.txt")
````
