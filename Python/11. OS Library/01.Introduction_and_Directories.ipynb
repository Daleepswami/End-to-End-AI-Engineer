# **01. Introduction to OS Module**

### **1.1 What is the `os` Module?**

**1. What is the `os` Module?**

- ***The `os` module provides a way to interact with the operating system: navigate directories, manage files and folders, read environment variables, etc.***

- ***It is a built-in module of Python, so nothing needs to be installed.***

### **1.2 Why Do We Use the `os` Module?**

**1. Reasons for Using the `os` Module**

1. **Work with Directories:** Find the current folder, change it, create folders and delete folders.
2. **Manage Files:** Rename and delete files, and check whether they exist.
3. **Cross-platform Code:** The same Python code can run on Windows, macOS and Linux.
4. **Environment Variables:** Read information stored in the system's environment.
5. **Automation:** Automate everyday tasks, such as organizing files in a folder.

**2. Example: Reading an Environment Variable**

````python
import os

print(os.getenv("NO_SUCH_VARIABLE", "default value"))     # returns the default if the variable is missing
````

### **1.3 OS Module vs Operating System**

**1. What is the Difference?**

- ***The operating system (Windows, macOS, Linux) is the software that manages the computer's hardware, files and programs.***

- ***The `os` module is a Python library that lets your Python program talk to that operating system.***

| Feature | Operating System | `os` Module |
|---------|------------------|-------------|
| What it is | System software (Windows, macOS, Linux) | A Python module |
| Role | Manages hardware, files and programs | Lets Python code use the operating system's features |
| Installation | Installed on the computer | Comes with Python |
| Example | Creating a folder in File Explorer | `os.mkdir("folder")` |

### **1.4 Importing the `os` Module**

**1. Ways to Import**

- ***`import os` imports the whole module. Functions are then used as `os.function_name()`.***

- ***`from os import getcwd` imports only the chosen function, which can then be used without the `os.` prefix.***

- ***`import os as o` gives the module a short alias (not recommended, because `os` is already short).***

````python
import os                                   # most common way
print(os.getcwd())

from os import listdir                      # import only one function
print(listdir("."))

import os as o                              # alias
print(o.name)
````

### **1.5 Understanding `os.name`**

**1. What is `os.name`?**

- ***`os.name` is a string that tells which type of operating system Python is running on.***

- ***It helps in writing code that behaves differently on different systems.***

| Value | Meaning |
|-------|---------|
| `'posix'` | Linux, macOS and other Unix-like systems |
| `'nt'` | Windows |

````python
import os

print(os.name)

if os.name == "nt":
    print("Running on Windows")
else:
    print("Running on Linux / macOS")

print("Path separator:", os.sep)          # \ on Windows, / on Linux and macOS
````

### **1.6 Difference Between `os` and `os.path`**

**1. What is the Difference?**

- ***`os` provides functions that work with the operating system: directories, files, environment variables.***

- ***`os.path` is a submodule of `os`. It provides functions that work with path strings, such as joining, splitting and checking paths.***

- ***When you write `import os`, `os.path` is available automatically.***

| Feature | `os` | `os.path` |
|---------|------|-----------|
| Purpose | Interact with the operating system | Work with file and folder paths |
| Examples | `os.getcwd()`, `os.listdir()`, `os.mkdir()`, `os.rename()`, `os.remove()` | `os.path.join()`, `os.path.exists()`, `os.path.basename()`, `os.path.splitext()` |
| Changes things on disk? | Yes (creates, renames, deletes) | Mostly no (reads or builds path strings) |

````python
import os

print(os.getcwd())                          # os    -> asks the operating system
print(os.path.join("folder", "file.txt"))   # os.path -> builds a path string
````

# **02. Working with Directories**

### **2.1 What is a Directory?**

**1. What is a Directory?**

- ***A directory (also called a folder) is a location on the disk that stores files and other directories.***

- ***A directory inside another directory is called a subdirectory.***

- ***The directory in which a Python program is currently running is called the current working directory.***

```
project/              <- directory
├── main.py           <- file
├── data/             <- subdirectory
│   └── numbers.txt
└── notes.txt
```

### **2.2 `os.getcwd()` – Get Current Working Directory**

**1. What is `os.getcwd()`?**

- ***`os.getcwd()` returns the current working directory as a string.***

**2. Syntax:**

```python
os.getcwd()
```

````python
import os

print(os.getcwd())                          # Current working directory
````

### **2.3 `os.chdir()` – Change Current Directory**

**1. What is `os.chdir()`?**

- ***`os.chdir(path)` changes the current working directory to the given path.***

- ***It raises `FileNotFoundError` if the directory does not exist.***

**2. Syntax:**

```python
os.chdir(path)
```

````python
start = os.getcwd()
os.makedirs("demo_dir", exist_ok=True)

os.chdir("demo_dir")                        # move into demo_dir
print("Now in :", os.getcwd())

os.chdir(start)                             # go back
print("Back in:", os.getcwd())

try:
    os.chdir("no_such_directory")
except FileNotFoundError as e:
    print("Error:", e)
````

### **2.4 `os.listdir()` – List Directory Contents**

**1. What is `os.listdir()`?**

- ***`os.listdir(path)` returns a list of the names of all files and folders inside a directory.***

- ***If no path is given, it lists the current directory.***

- ***It returns only names, not full paths, and the order is not guaranteed.***

**2. Syntax:**

```python
os.listdir(path=".")
```

````python
for name in ["a.txt", "b.txt"]:
    with open(os.path.join("demo_dir", name), "w") as f:
        f.write("sample")
os.makedirs("demo_dir/sub", exist_ok=True)

print(os.listdir('.'))                      # List files in current dir
print(sorted(os.listdir("demo_dir")))       # List files in demo_dir
````

### **2.5 `os.mkdir()` – Create a Directory**

**1. What is `os.mkdir()`?**

- ***`os.mkdir(path)` creates a single directory.***

- ***It raises `FileExistsError` if the directory already exists.***

- ***It raises `FileNotFoundError` if the parent directory does not exist.***

**2. Syntax:**

```python
os.mkdir(path)
```

````python
try:
    os.mkdir("new_folder")                  # Create single directory
    print("new_folder created")
except FileExistsError:
    print("new_folder already exists")

try:
    os.mkdir("missing_parent/child")        # parent does not exist
except FileNotFoundError as e:
    print("Error:", e)
````

### **2.6 `os.makedirs()` – Create Nested Directories**

**1. What is `os.makedirs()`?**

- ***`os.makedirs(path)` creates a directory along with all the missing parent directories.***

- ***With `exist_ok=True`, no error is raised if the directory already exists.***

**2. Syntax:**

```python
os.makedirs(path, exist_ok=False)
```

````python
os.makedirs('school/classA/students', exist_ok=True)       # Create nested directories
print(os.path.exists('school/classA/students'))
````

**3. `mkdir()` vs `makedirs()`**

| Feature | `os.mkdir()` | `os.makedirs()` |
|---------|--------------|-----------------|
| Creates | One directory | A directory and all missing parents |
| Missing parent | Error | Parents are created |
| `exist_ok` option | No | Yes |

### **2.7 `os.rmdir()` – Remove an Empty Directory**

**1. What is `os.rmdir()`?**

- ***`os.rmdir(path)` removes a directory, but only if it is empty.***

- ***If the directory contains files or folders, an error (`OSError`) is raised.***

**2. Syntax:**

```python
os.rmdir(path)
```

````python
os.rmdir('new_folder')                      # Remove empty directory
print(os.path.exists('new_folder'))

os.makedirs("full_folder", exist_ok=True)
with open("full_folder/file.txt", "w") as f:
    f.write("sample")

try:
    os.rmdir("full_folder")                 # directory is not empty
except OSError as e:
    print("Error:", e)

os.remove("full_folder/file.txt")           # empty it first
os.rmdir("full_folder")
print(os.path.exists("full_folder"))
````

### **2.8 `os.removedirs()` – Remove Nested Empty Directories**

**1. What is `os.removedirs()`?**

- ***`os.removedirs(path)` removes the last directory of the path, and then keeps removing its parent directories, one by one, as long as they are empty.***

- ***All directories to be removed must be empty. It stops without an error when it reaches a parent that is not empty.***

**2. Syntax:**

```python
os.removedirs(path)
```

````python
os.removedirs('school/classA/students')     # removes students, classA and school
print(os.path.exists('school'))
````
