## Conditionals in Python

### Introduction to Conditionals

Conditional statements are used to make decisions in a Python program. They allow a program to execute different blocks of code depending on whether a specified condition is satisfied.

Python evaluates a condition and determines whether it is true or false. Based on the result, the program decides which block of code to execute.

The main types of conditional statements in Python are:

1. if statement

2. if-else statement

3. if-elif-else statement

4. Nested if statements

### **1. The if Statement**

The if statement executes a block of code only when the specified condition is true.

If the condition is false, Python skips the block and continues executing the remaining code.







```python
#Syntax

#if condition:
    # Block of code
```

1. if is the keyword used to introduce a condition.

2. condition is an expression evaluated by Python.

3. : indicates the beginning of the conditional block.

4. The indented statements belong to the if block.

```python
# Example 1: Checking Whether a Number Is Positive

num = 10

if num > 0:
    print("The number is positive.")
```

```python
# Example 2: Checking Whether a Number Is Negative

num = -5

if num < 0:
    print("The number is negative.")
```

```python
#Example 3: Checking Whether a Number Is Even

num = 8

if num % 2 == 0:
    print("The number is even.")
```

### **2. if-else statement**

The if-else statement is used when a program needs to choose between two alternatives.

If the if condition is satisfied, Python executes the if block.

If the if condition is not satisfied, Python executes the else block.

Exactly one of these two blocks executes in a normal if-else statement.





```python
#Syntax

#if condition:
    # Executes when the condition is satisfied
#else:
   # Executes when the condition is not satisfied


```

1. if checks the specified condition.

2. else handles the alternative case when the condition is not satisfied.

3. : marks the beginning of each block.

4. Indentation defines which statements belong to each block.

5. The else keyword does not require a condition.

```python
#Example 1: Checking Whether a Number Is Even or Odd

num = 7

if num % 2 == 0:
    print("The number is even.")
else:
    print("The number is odd.")
```

```python
##Example 2: Checking Whether a Number Is Positive or Non-Positive

num = -5

if num > 0:
    print("The number is positive.")
else:
    print("The number is non-positive.")
```

```python
#Example 3: Checking Whether a Person Is Eligible to Vote

age = 20

if age >= 18:
    print("Eligible to vote.")
else:
    print("Not eligible to vote.")
```

### **3. if-elif-else statement**
The if-elif-else statement is used when a program needs to choose between multiple conditions.

**In Python step:**

1. if checks the first condition.

2. elif (short for else if) checks another condition if the preceding conditions were not satisfied.

3. else executes when none of the preceding conditions are satisfied.

4. Python evaluates conditions from top to bottom. Once a condition is satisfied, its block executes, and the remaining conditions in that chain are skipped.



 

```python
#Syntax

#if condition1:
    # Executes if condition1 is satisfied
#elif condition2:
    # Executes if condition1 is not satisfied
    # and condition2 is satisfied
#elif condition3:
   # Executes if the preceding conditions are not satisfied
   # and condition3 is satisfied
#else:
    # Executes if none of the conditions are satisfied



```

**Understanding the Syntax**

if is the first condition and is required.

elif adds another condition. You can use multiple elif blocks.

else is optional and handles all remaining cases.

Each condition is checked in order until one is satisfied.

The else block does not contain a condition.

```python
# Example 1: Checking Whether a Number Is Positive, Negative, or Zero

num = 0

if num > 0:
    print("The number is positive.")

elif num < 0:
    print("The number is negative.")

else:
    print("The number is zero.")
```

```python
#Example 2: Assigning Grades Based on Marks

marks = int(input("Enter your marks"))

if marks >= 90:
    print("Grade A")

elif marks >= 80:
    print("Grade B")

elif marks >= 70:
    print("Grade C")

elif marks >= 60:
    print("Grade D")

else:
    print("Grade F")
```

```python
# Example 3: Comparing Three Numbers

a = 20
b = 15
c = 10

if a > b and a > c:
    print("a is the largest.")

elif b > a and b > c:
    print("b is the largest.")

elif c > a and c > b:
    print("c is the largest.")

else:
    print("Some numbers are equal.")
```

### **Nested if Statements in Python**

A nested if statement is an if statement placed inside another if statement. It is used when a program needs to check an additional condition after an outer condition has been satisfied.

Python first evaluates the outer condition. If it is satisfied, Python enters the outer block and checks the inner condition. If the outer condition is not satisfied, the inner block is skipped.

Nested if statements can contain multiple levels of conditions depending on the program's requirements.


```python
#syntax

#if condition1:
    # Outer if block

    #if condition2:
        # Inner if block
```

**Understanding the Syntax**

1. The first if is called the outer if statement.

2. The second if is called the inner if statement.

3. The inner if is evaluated only when the outer condition is satisfied.

4. The inner block requires additional indentation.

5. Each condition controls whether its own block executes.

```python
#Example 1: Checking Eligibility to Vote

age = 25

if age >= 18:
    print("Age requirement satisfied.")

    if age >= 21:
        print("Eligible for this example's age requirement.")

print("Program completed.")
```

```python
# Example 2: Checking Username and Password

username = "admin"
password = "python123"

if username == "admin":
    print("Username is correct.")

    if password == "python123":
        print("Login successful.")

    else:
        print("Incorrect password.")

else:
    print("Incorrect username.")
```

```python
#Example 3: Nested if with User Input

age = int(input("Enter your age: "))

if age >= 18:
    income = int(input("Enter your monthly income: "))

    if income >= 25000:
        print("Basic eligibility criteria satisfied.")

    else:
        print("Income requirement not satisfied.")

else:
    print("Age requirement not satisfied.")
```
