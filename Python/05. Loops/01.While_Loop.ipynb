### **Introduction to Iteration**

Iteration is the process of executing a block of code repeatedly until a specified condition is met or the loop is terminated.

**Python provides two main types of loops:**

1. while loop

2. for loop

#### **The while Loop**

A while loop repeatedly executes a block of code as long as the given condition evaluates to True.

```python
#syntax

#while condition:
#    statement(s)



#while: Starts the loop.
#condition: Determines whether the loop should continue.
#statement(s): The block of code executed during each iteration.
```

```python
# Example: Print Numbers from 1 to 5

count = 1
while count <= 5:
    print(count)
    count += 1
```

**Infinite Loop**

An infinite loop occurs when a loop's condition never becomes False, causing the loop to continue indefinitely.

**Using break to Exit a Loop**

The break statement immediately terminates the nearest enclosing loop and transfers control to the statement following that loop.

```python
count = 1

while count <= 10:
    if count == 6:
        break
    print(count)
    count += 1
```

**Using continue to Skip an Iteration**

The continue statement skips the remaining statements in the current iteration and moves to the next iteration of the loop.

```python
count = 0

while count < 10:
    count += 1

    if count % 2 == 0:
        continue

    print(count)
```

**Nested while Loops**

A nested while loop is a while loop placed inside another while loop.

The inner loop executes completely for each iteration of the outer loop, provided its condition remains True.

```python
i = 1
while i <= 3:
    j = 1

    while j <= 3:
        print(f"i = {i}, j = {j}")
        j += 1

    i += 1
```
