#### **For Loop and Range in Python**

1. A for loop is used to iterate over a sequence of elements, such as a list, tuple, string, or range of numbers. It executes a block of code once for each element in the sequence.

2. Unlike a while loop, which continues based on a condition, a for loop typically iterates over a predefined sequence or iterable.

```python
#Syntax

# for variable in iterable:
#     statement(s)
```

1. for: Starts the loop.

2. variable: Stores the current element during each iteration.

3. in: Connects the variable to the iterable.

4. iterable: The collection or sequence being iterated over.

5. statement(s): The code executed during each iteration.

```python
#Example 1: Iterating Over a List

numbers = [10, 20, 30, 40, 50]

for num in numbers:
    print(num)
```

#### **The range() Function**

1. The range() function generates a sequence of integers. It is commonly used with a for loop when you want to repeat an operation a specific number of times.

2. The range() function has three forms.

#### **1. range(stop)**

Generates integers starting from 0 up to, but not including, stop

```python
#Syntax:
#range(stop)


#Example:

for i in range(5):
    print(i)
```

#### **2. range(start, stop)**

Generates integers starting from start up to, but not including, stop

```python
#Syntax:
#range(start, stop)

#Example:
for i in range(1, 6):
    print(i)
```

#### **3. range(start, stop, step)**

Generates integers starting from start, increasing or decreasing by step, until the stop boundary is reached.

```python
#Syntax:
#range(start, stop, step)
for i in range(2, 11, 2):
    print(i)
```

```python
#Negative Step

for i in range(5, 0, -1):
    print(i)
```

**Using range() to Calculate a Sum**

The range() function can also be used to perform calculations repeatedly.

```python
#Example: Sum of Numbers from 1 to 5

total = 0

for i in range(1, 6):
    total += i

print("Sum:", total)
```

**Using break in a for Loop**

The break statement terminates the loop immediately.

```python
for i in range(1, 11):
    if i == 6:
        break

    print(i)
```

**Using continue in a for Loop**

The continue statement skips the remaining code in the current iteration and proceeds to the next iteration.

```python
for i in range(1, 11):
    if i % 2 == 0:
        continue

    print(i)
```

**Nested for Loops**

A nested for loop is a for loop inside another for loop

```python
for i in range(1, 4):
    for j in range(1, 4):
        print(f"i = {i}, j = {j}")
```
